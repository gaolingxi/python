### 文件操作
Python提供了多种文件操作的方法，允许你创建、读取、写入、删除文件，以及执行其他文件管理任务。

````python
import os
import shutil

source_dir = r"C:/Users/gao_l/Desktop/作业"
target_dir1 = r"C:/Users/gao_l/Desktop/作业修改/规律作业名"
target_dir2 = r"C:/Users/gao_l/Desktop/作业修改/不规律作业名"

for filename in os.listdir(source_dir):
    print(filename)
    # 获取文件的完整路径
    file_path = os.path.join(source_dir, filename)

    # 获取文件名
    file_name = os.path.basename(file_path)
    
    # 将正常的文件和规律的文件分开
    if filename.startswith('2024'):
        # 创建目标文件的完整路径
        target_file_path1 = os.path.join(target_dir1, filename)

        # 移动文件到目标目录，如果已经有文件在，直接覆盖
        shutil.move(file_path, target_file_path1)
    
    else:
        # 创建目标文件的完整路径
        target_file_path2 = os.path.join(target_dir2, filename)

        # 移动文件到目标目录，如果已经有文件在，直接覆盖； 
        
        shutil.move(file_path, target_file_path2)
````
