```python
# 检查 df1 的 'province' 列中的每个元素是否在 df2 的 'province' 列的某个元素中
# 并更新 df1 的 'province' 列
df1['provcode'] = df1['provcode'].apply(lambda x: next((y for y in df2['province'] if x in y), x))
df1['year']  = df1['year'].astype(int) + 1


# 查看更新后的 df1
print(df1)
```
