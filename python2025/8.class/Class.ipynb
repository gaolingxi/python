# 面向对象编程
面向对象编程是一种编程思想。面向对象编程是从面向过程编程过渡来的。

## 以洗衣服作为一个案例
### 面向过程编程
如果用面向过程的编程思想，我们将洗衣服拆解成一系列步骤，每个步骤就是一个函数。<br>
```第一步```: 打开洗衣机；<br>
```第二步```: 放衣服和洗衣液；<br>
```第三步```: 选择洗衣模式，开始洗衣服；<br>
```第四步```: 等待洗完，拿出衣服。<br>

### 面向对象编程
拆分人和洗衣机两个对象：<br>
#### 人
```第一步```: 打开洗衣机；<br>
```第二步```: 放衣服和洗衣液；<br>
```第三步```: 洗完衣服后拿出衣服；<br>
#### 洗衣机
```第一步```: 根据洗衣模式洗衣服；<br>

### 总结
* 面向过程编程：更注重事情的每个步骤以及顺序
* 面向对象编程：更注重事情有哪些参与者，参与者需要做些什么事情。将其拆解成一个个模块和对象。

## 用一个简单的例子来通俗地解释面向对象编程（Object-Oriented Programming，简称OOP）。

想象一下，你正在玩一个电子游戏，游戏里有很多角色，比如战士、法师和弓箭手。每个角色都有他们自己的属性（比如生命值、攻击力）和技能（比如战士的剑击、法师的火球术、弓箭手的箭雨）。在面向对象编程中，我们可以把每个角色看作是一个“对象”。

### 1. 什么是“对象”？

在面向对象编程的世界里，“对象”就像是现实世界中的事物。每个对象都有自己的“属性”（就像是事物的特性）和“方法”（就像是事物能做的事情）。比如，一辆车是一个对象，它有颜色（属性）和启动（方法）。

### 2. 什么是“类”？

“类”就像是对象的“蓝图”或“模板”。通过这个蓝图，我们可以创建很多相似的对象。比如，我们有一个“角色”类，它定义了所有角色共有的属性和方法。然后，我们可以用这个类来创建具体的战士、法师和弓箭手对象。

### 3. 面向对象编程的好处

- **封装**：每个对象都把自己的数据（属性）和操作数据的方法（行为）封装在一起，这样可以保护数据不被随意访问和修改。
- **继承**：我们可以创建新的类来继承已有类的属性和方法，这样可以减少重复代码，让编程更高效。
- **多态**：不同的对象可以对同一个方法有不同的响应，这让程序更灵活。

### 总结

面向对象编程就像是在用代码创造一个虚拟世界，你定义了各种“类”（蓝图），然后根据这些蓝图创建具体的“对象”（实例）。每个对象都有自己的属性和方法，就像是每个角色都有自己的生命值和技能。通过这种方式，我们可以更自然地模拟现实世界的复杂性，并且让代码更加模块化、易于理解和维护。


# 面向对象程序设计-类(Class)

````python
class Consumer:
    def __init__(self, w):
        self.wealth = w

    def earn(self, y):
        self.wealth += y

    def spend(self, x):
        new_wealth = self.wealth - x
        if new_wealth < 0:
            print("Insufficent Funds")
        else:
            self.wealth = new_wealth
````

````python
c1 = Consumer(10)
c1.earn(5)
c1.spend(3)
c1.wealth
````
