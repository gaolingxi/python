1.消费者月收入1000元，购买$q_X$、$q_Y$数量的商品，他们的价格分别为$p_X=5$元，$p_Y=4$元，总效用函数为$U_{X,Y}=2XY$，求：(提示：在AI搜索中搜索以下关键字“证明效应最大化边际效用相等”)
a.获得效用最大化时X、Y各购买多少？
b.效用最大时其总效用、单位货币的边际效用是多少？
c.如果$p_X$提高10%，$p_Y$不变，在总效用不变的条件下，收入应增加多少？
将以上过程写成一个名为consumer的类。


```python
class Consumer:
    def __init__(self, income, p_X, p_Y, U_XY):
        self.income = income
        self.p_X = p_X
        self.p_Y = p_Y
        self.U_XY = U_XY

    def maximize_utility(self):
        # 解决效用最大化问题
        X = self.income / (self.p_X + self.p_Y * (self.p_X / self.p_Y))  # 根据效用最大化条件 MU_X / MU_Y = p_X / p_Y
        Y = (self.p_X / self.p_Y) * X
        return X, Y

    def calculate_utility_and_marginal_utilities(self, X, Y):
        U_max = self.U_XY * X * Y
        MU_X = 2 * Y
        MU_Y = 2 * X
        return U_max, MU_X, MU_Y

    def calculate_income_increase(self, new_p_X):
        # 更新p_X为new_p_X
        original_p_X = self.p_X
        self.p_X = new_p_X
        X, Y = self.maximize_utility()
        # 计算新的收入，使得效用保持不变
        new_income = X * self.p_X + Y * self.p_Y
        self.p_X = original_p_X  # 恢复原来的p_X值
        return new_income

# 创建消费者对象
consumer = Consumer(income=1000, p_X=5, p_Y=4, U_XY=2)

# a. 获得效用最大化时X、Y各购买多少？
X, Y = consumer.maximize_utility()
print(f"效用最大化时X、Y各购买多少？\nX = {X}, Y = {Y}")

# b. 效用最大时其总效用、单位货币的边际效用是多少？
U_max, MU_X, MU_Y = consumer.calculate_utility_and_marginal_utilities(X, Y)
print(f"效用最大时其总效用、单位货币的边际效用是多少？\n总效用 = {U_max}, MU_X = {MU_X}, MU_Y = {MU_Y}")

# c. 如果p_X提高10%，p_Y不变，在总效用不变的条件下，收入应增加多少？
new_p_X = 5 * 1.1
income_increase = consumer.calculate_income_increase(new_p_X)
print(f"如果p_X提高10%，p_Y不变，在总效用不变的条件下，收入应增加多少？\n收入应增加 = {income_increase}")
```

1. 在桌面上建立名为homework的文件，在其中生成50个文件夹以"20240X"命名,X=1,2,...,50，对于每个20240X,生成5个文件夹，命名为20240XY，Y=a,b...e。在homework文件夹中生成50个文本文件，文件命名为"2024+X"，X=51,52,...,100,对于每个2024+X,生成5个文本文件，命名为2024+X+Y，Y=a,b...e。在homework文件夹中生成50个表格文件，文件命名为"2024 X"，X=101,102,...,150,对于每个2024 X,生成5个表格文件，命名为2024 XY，Y=a,b...e。

```python
import os

# 定义桌面路径，根据你的操作系统和用户名进行调整
desktop_path = 'C:\\Users\\glx\\Desktop'
homework_folder = os.path.join(desktop_path, 'homework')

# 创建homework文件夹
if not os.path.exists(homework_folder):
    os.makedirs(homework_folder)

# 生成50个文件夹以"20240X"命名
for x in range(1, 51):
    folder_name = f"20240{x}"
    folder_path = os.path.join(homework_folder, folder_name)
    os.makedirs(folder_path, exist_ok=True)
    # 对于每个20240X,生成5个文件夹，命名为20240XY
    for y in ['a', 'b', 'c', 'd', 'e']:
        subfolder_name = f"20240{x}{y}"
        subfolder_path = os.path.join(folder_path, subfolder_name)
        os.makedirs(subfolder_path, exist_ok=True)

# 在homework文件夹中生成50个文本文件，文件命名为"2024+X"
for x in range(51, 101):
    file_name = f"2024+{x}.txt"
    file_path = os.path.join(homework_folder, file_name)
    with open(file_path, 'w') as file:
        file.write("")  # 创建空文件
    # 对于每个2024+X,生成5个文本文件，命名为2024+X+Y
    for y in ['a', 'b', 'c', 'd', 'e']:
        subfile_name = f"2024+{x}{y}.txt"
        subfile_path = os.path.join(homework_folder, subfile_name)
        with open(subfile_path, 'w') as file:
            file.write("")  # 创建空文件

# 在homework文件夹中生成50个表格文件，文件命名为"2024 X"
for x in range(101, 151):
    file_name = f"2024 {x}.xlsx"
    file_path = os.path.join(homework_folder, file_name)
    # 创建空的Excel文件（需要安装openpyxl库）
    from openpyxl import Workbook
    wb = Workbook()
    wb.save(file_path)
    # 对于每个2024 X,生成5个表格文件，命名为2024 XY
    for y in ['a', 'b', 'c', 'd', 'e']:
        subfile_name = f"2024 {x}{y}.xlsx"
        subfile_path = os.path.join(homework_folder, subfile_name)
        wb = Workbook()
        wb.save(subfile_path)

print("文件和文件夹创建完成。")
```
