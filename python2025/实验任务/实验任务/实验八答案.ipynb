```python
# 人均家庭纯收入分位数
print( family_2018["fincome2_per_p"].unique())

import pandas as pd
import numpy as np

# 假设 df 是你的 DataFrame
# 首先计算四分位数
Q1 = family_2018['fincome2_per'].quantile(0.25)
Q2 = family_2018['fincome2_per'].quantile(0.5)
Q3 = family_2018['fincome2_per'].quantile(0.75)

# 然后，使用 np.select 来为每个收入分配四分位数标签
import numpy as np
family_2018['income_quartile'] = np.select(
    [
        (family_2018['fincome2_per'] <= Q1),  # 小于等于Q1
        (family_2018['fincome2_per'] > Q1) & (family_2018['fincome2_per'] <= Q2),  # 大于Q1且小于等于Q2
        (family_2018['fincome2_per'] > Q2) & (family_2018['fincome2_per'] <= Q3),  # 大于Q2且小于等于Q3
        family_2018['fincome2_per'] > Q3  # 大于Q3
    ],
    ['1', '2', '3', '4']
)
```

```python
compare = family_2018[["fincome2_per_p", 'income_quartile']]
print(compare)
```
