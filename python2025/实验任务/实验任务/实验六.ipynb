1.编写一个python函数，该函数能输出Pascal三角形的前n行，n = 6。


[1]
[1, 1]
[1, 2, 1]
[1, 3, 3, 1]
[1, 4, 6, 4, 1]
[1, 5, 10, 10, 5, 1]

````python
def generate_pascals_triangle(n):
    triangle = []
    for i in range(n):
        row = [1]  # 每行以 1 开始
        if triangle:
            last_row = triangle[-1]
            row.extend([last_row[j] + last_row[j + 1] for j in range(len(last_row) - 1)])
            row.append(1)  # 每行以 1 结束
        triangle.append(row)
    return triangle

# 输出 Pascal 三角形的前 6 行
n = 6
pascals_triangle = generate_pascals_triangle(n)
for row in pascals_triangle:
    print(row)
````
