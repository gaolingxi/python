# 家庭线上消费 

定义：家庭线上消费是指受访家庭在过去一年网上购物的消费总额，单位为元。<br>
数据来源：CHFS2019。
数据处理：为了避免异常值的影响，对家庭线上消费数额进行了缩尾处理，即在1%和99%的分位数上进行截断，然后取对数后用于回归分析。

# 清洗样本

```python
# 删除'online_consump'列中含有缺失值的行
family_2019_cleaned = family_2019.dropna(subset=['online_consump'])
```

```python

print(family_2019_cleaned)
```

```python
# 删除'online_consump'列中含有缺失值的行
family_2019_cleaned = family_2019.dropna(subset=['online_consump'])

# 在1%和99%的分位数上进行截断
import pandas as pd

# 假设family_2019_cleaned是您的DataFrame
Q1 = family_2019_cleaned['online_consump'].quantile(0.01)
Q99 = family_2019_cleaned['online_consump'].quantile(0.99)
```

```python
print(family_2019_cleaned)
```

```python
family_2019_cleaned['online_consump_clipped'] = family_2019_cleaned['online_consump'].clip(lower=Q1, upper=Q99)
```

```python

print(family_2019_cleaned)
```

* lower=Q1：这是online_consump列的1%分位数，即该列中1%的最小值。任何小于这个值的元素都会被替换为Q1。
* upper=Q99：这是online_consump列的99%分位数，即该列中99%的最大值。任何大于这个值的元素都会被替换为Q99。
* np.clip()函数将online_consump列中的所有元素限制在Q1和Q99之间。这通常用于数据预处理，以减少极端值的影响，使数据更加稳定，便于分析和建模。

```python
family_2019_drop = family_2019_cleaned[(family_2019_cleaned['online_consump'] >= Q1) & (family_2019_cleaned['online_consump'] <= Q99)]
print(family_2019_drop)
```

# 城市物流效率

定义：该变量是通过家庭报告的快递送达时间来衡量的，具体是将家庭层面的物流时效加权到城市层面。<br>

“CHFS 2019的问卷中问到：‘您家进行网络购物的平均收货时间（从下单到签收）是多久？（单位：天）’。对应的回答分别为1—7，其中：1代表1天；2代表2天；3代表3天；4代表4—5天；5代表6—7天；6代表7天以上。为便于直接纳入回归分析中，在数据处理过程中将回答结果直接转换为物流配送天数。将4—5天定义为4.5天；将6—7天定义为6.5天；将7天以上定义为8天。数值越大，代表物流效率越低。

```python
import numpy as np
family_2019['delivery_time'] = np.nan
family_2019.loc[(family_2019['g1023dg'] == 1),'delivery_time'] = 1
family_2019.loc[(family_2019['g1023dg'] == 2),'delivery_time'] = 2
family_2019.loc[(family_2019['g1023dg'] == 3),'delivery_time'] = 3
family_2019.loc[(family_2019['g1023dg'] == 4),'delivery_time'] = 4.5
family_2019.loc[(family_2019['g1023dg'] == 5),'delivery_time'] = 6.5
family_2019.loc[(family_2019['g1023dg'] == 6),'delivery_time'] = 8

# delivery_time中缺失值填充为0
family_2019['delivery_time'] = family_2019['delivery_time'].fillna(0)
```

家庭层面的物流时效加权到城市层面

```python
print(master_2019)
```

```python
prov = master_2019[['hhid', 'prov', 'total_consump']]
```

```python
print(prov)
```

```python
prov = master_2019[['hhid', 'prov', 'total_consump']]
merged_data = pd.merge(family_2019, prov, on=['hhid'], how='inner')
merged_data= merged_data.groupby('prov', as_index=False)['delivery_time'].mean()
merged_data_df = pd.DataFrame(merged_data)
```

```python
print(merged_data_df)
```

```python
data = pd.merge(family_2019_drop, prov, on=['hhid'], how='inner')
data = pd.merge(data, merged_data_df, on=['prov'], how='inner')
data['online_rate'] = data['online_consump']/data['total_consump']
```

```python
data['is_onlineconsump'] = np.nan
data['is_onlineconsump'] = np.where(pd.isnull(data['g1023d']), 0, data['is_onlineconsump'])
data.loc[(data['is_onlineconsump'] != 0), 'is_onlineconsump'] = 1
```

在这两个示例中，我们都是在检查'g1023d'列中的值是否为缺失值（NaN）。如果是，那么'is_onlineconsump'列对应的值将被设置为0。如果不是，'is_onlineconsump'列将保持原来的值。
