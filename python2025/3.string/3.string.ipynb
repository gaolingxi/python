# 3.字符串
字符串是计算机编程中表示文本数据的一种数据类型。它可以包括字母、数字、符号以及其他所有能在键盘上输入的字符。

## 字符串的编码与转换
### 字符串的Unicode编码
* 最早的字符串是美国标准信息交换码ASCII码：表示10个数字、26个大写英文字母、26个小写英文字母等256个字符；
* 由于各国的文字都需要编码，于是出现了多种编码。我国制定的中文编码标准有：GBK和GB2312;
* UTF-8是国际通用的编码，对全世界所有国家需要用到的字符都进行了编码。<br>
【注】：在Python3.X中，默认采用UTF-8编码

### 字符串与字节序列
* 字符串(String)是用来表示文本的数据类型，可以包含字母、数字、符号等字符。在Python 3中，字符串是Unicode编码的，这意味着它们可以表示世界上大多数语言的字符。
* 字节序列(Byte Sequence)表示二进制数据
【注】：文本在计算机内部用Unicode存储，但网络传输需要二进制形式，所以得把文本转换成字节序列。

### 使用encode函数进行编码
将Unicode编码的字符串转为字节序列。<br>
**encode函数的用法：**
```python
字符串.encode('文本数据具体编码')
```

````python
str1 = "电商发展通过克服交易障碍和空间距离的约束，直接提升家庭消费"
byte1 = str1.encode('GBK')
print('原字符串:', byte1)
print('转换后:', str1)
````

bytes类型是带有b前缀的文本数据

````python
str1 = "电商发展通过克服交易障碍和空间距离的约束，直接提升家庭消费"
byte1 = str1.encode()
print('原字符串:', byte1)
print('转换后:', str1)
````

### 使用decode函数进行编码
将二进制数据转换为Unicode编码的字符串，这称为解码。<br>
**decode函数的用法：**
```python
字符串.decode('文本数据具体编码')
```

````python
str2 = byte1.decode()
print('解码后:',str2)
````

### 分割字符串
把一个字符串str按照制定的分隔符切分为字符串的列表。<br>
分割字符串的用法：
```python
str.split(分隔符，分割次数)
```
返回结果列表的元素最多为(分割次数+1)<br>
默认：<br>

### 检索字符串
#### count()方法

count()方法用于检索指定字符串在另一个字符串中出现的次数。如果被检索的字符串不存在，则返回0，否则返回出现的次数。<br>
**count()方法的用法：**
```python
返回子字符串的次数 = 原字符串.count(子字符串,搜索起始索引,搜索结束索引)
```


````python
# 家庭年度支出记录
family_expenses = """
1月：食品 - 500，水电费 - 200，娱乐 - 100
2月：食品 - 450，水电费 - 250，教育 - 150
3月：食品 - 550，医疗 - 300，水电费 - 210
4月：水电费 - 190，教育 - 200，食品 - 480
5月：食品 - 500，水电费 - 220，医疗 - 100
6月：食品 - 470，水电费 - 230，娱乐 - 50
7月：水电费 - 205，教育 - 300，食品 - 510
8月：食品 - 500，娱乐 - 200，水电费 - 190
9月：医疗 - 400，食品 - 450，水电费 - 220
10月：水电费 - 180，食品 - 470，教育 - 100
11月：食品 - 500，水电费 - 210，医疗 - 200
12月：水电费 - 195，娱乐 - 150，食品 - 480

"""

# 统计“食品”类别出现的频率
food_count = family_expenses.count('食品')

# 统计“水电费”类别出现的频率
ent_count = family_expenses.count('娱乐')

print("一年中家庭有食品消费的月份有", food_count, "次。")
print("一年中家庭有娱乐消费的月份有", ent_count,"次。")
````

#### find()方法
find()方法用于<span style="color: red;">从左至右</span>检索指定字符串是否在另一个字符串中出现的次数。如果被检索的字符串不存在，则返回-1，否则返回子字符串出现的索引。<br>
**find()方法的用法：**
```python
返回子字符串的次数 = 原字符串.find(子字符串,搜索起始索引,搜索结束索引)
```
【注】：python中字符串的检索方法还有rfind()方法，其作用与find()类似，只是<span style="color: red;">从右至左</span>进行检索。

````python
# 找到python中字母t的位置
str = "python"
tth = str.rfind("t")
print(tth)
````

### 检索字符串的开头和结尾
#### 检索字符串是否以指定子字符串开头

startswith()方法用于检索字符串是否以指定的子字符串开头。如果是，则返回True；否则返回False。<br>
**startswith()的用法：**<br>
```python
原字符串.startswith(子字符串,检索范围起始位置的索引,检索范围结束位置的索引)
```

````python
# 通过身份证号来判断是否浙江人
id = input("请输入您的身份证号：")
index1 = id.startswith("33",0,)
print(index1)
if index1:
    print("出生地是浙江省")
else:
    print("出生地不是浙江省")
````

#### 检索字符串是否以指定子字符串结尾
endwith()方法用于检索字符串是否以指定的字符串结尾的。如果是，则返回True；否则返回False。<br>
**endswith()的用法:**<br>
```python
原字符串.endswith(子字符串,检索范围起始位置的索引,检索范围结束位置的索引)
```


````python
address = "浙江省湖州市安吉县天荒坪镇下辖村"
index2 = address.endswith("村")
if index2:
    print("户籍地为农村")
else:
    print("户籍地为城市")
````

### 计算字符串的长度
由于不同字符占字节数不同，因此如果要计算字符串的长度，需要先了解字符所占的字节数
* 在Python中，数字、英文、小数点、下划线和空格占一个字节；
* 汉字在GBK/GB2313编码中占两个字节
* 汉字在UTF-8中一般占3个字节

**len函数计算字符串的长度：**
```python
len(字符串)
```
当使用len()函数计算字符串的长度时，不区分英文、数字和汉字，所有字符都计数1个。

````python
str1 = "供需决定价格"
print(len(str1))
````

### 格式化字符串
格式化字符串是先制定一个模板，在这个模版中预留几个空位，然后根据需要填上相应的内容。这些空位需要通过制定的符号（占位符）进行标记。格式化字符串有以下两种方法：
1. 使用“%”操作符；
2. 使用字符串对象的format()方法。

**1.%操作符的用法:**
```python
%[0][宽度][保留的小数位数]格式字符
```

<img src="figure/format1.jpg" alt="示例图片" width="500" height="200">

````python
template = "社会消费品零售总额%09d亿元"
data = 38726
print(template % data)
````

| 格式字符 | 说明 |
| --- | --- |
| `%s` | 字符串 |
| `%c` | 单个字符 |
| `%d` 或 `%i` | 十进制整数 |
| `%f` 或 `%F` | 浮点数 |
| `%e` | 指数（基底写为e） |

**2.使用字符串对象的format()方法:**
```python
{index:[填充空白的符号][宽度][保留的小数位数]格式字符}
```
<img src="figure/format2.jpg" alt="示例图片" width="600" height="100">

| 格式字符 | 说明 |
| --- | --- |
| s | 字符串类型 |
| d | 十进制整数 |
| e或者E | 科学计数法 |
| f或者F | 浮点数 |
| % | 百分比 |

````python
template = "社会消费品零售总额{0:6.0f}亿元"
context = template.format(12000.123)
print(context)
````

````python
template = "社会消费品零售总额{0:09.0f}亿元"
context = template.format(12000.123)
print(context)
````

````python
template = "社会消费品零售总额{:,.2f}亿元"
context = template.format(1251)
print(context)
````

````python
template = "金融业，交通运输、仓储和邮政业生产指数同比分别增长{0}%、{1}%、{2}%、{3}%"
print(template.format(12.1, 9.4, 5.7, 5.2))
````
