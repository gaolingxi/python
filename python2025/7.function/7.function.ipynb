$$GDP_{rate}= \dfrac{GDP_1-GDP_0}{GDP_0} \times 100\% $$

````python
# 2020年的GDP增长率
gdp_2019 = 21
gdp_2020 = 22.6

gdp_growth_rate_2020 = (gdp_2020 - gdp_2019)/gdp_2019
print("2020年的GDP增长率为: {:.2f}%".format(gdp_growth_rate_2020*100))
````

````python
# 2021年的GDP增长率
gdp_2020 = 22.6
gdp_2021 = 25

gdp_growth_rate_2021 = (gdp_2021 - gdp_2020)/gdp_2020
print("2021年的GDP增长率为: {:.2f}%".format(gdp_growth_rate_2021*100))
````

````python
def growth_rate(value_1, value_0):
    if value_0 == 0:
        print("基期不能为0")
    growth_rate = ((value_1-value_0)/value_0)*100
    return growth_rate
````

````python
gdp_2019 = 21
gdp_2020 = 22.6
gdp_2021 = 25

gdp_growth_rate_2020 = growth_rate(gdp_2020, gdp_2019)
gdp_growth_rate_2021 = growth_rate(gdp_2021, gdp_2020)

print("2020年的GDP增长率为: {:.2f}%".format(gdp_growth_rate_2020))
print("2021年的GDP增长率为: {:.2f}%".format(gdp_growth_rate_2021))
````

## 参数传递
函数中参数的作用是传递数据给函数使用，函数利用接收到的数据进行具体的操作处理。

### 形式参数
在定义函数时，函数名后面括号中的参数为“形式参数”。
### 实际参数
在调用一个函数时，函数名后面括号中的参数为“实际参数”

````python
# 创建一个自定义的函数
def calculate_mpc(initial_income, initial_consumption, final_income, final_consumption):
    """
    计算边际消费倾向（MPC）的函数。
    
    参数:
    initial_income (float): 收入增加前的收入水平。
    initial_consumption (float): 收入增加前的消费水平。
    final_income (float): 收入增加后的收入水平。
    final_consumption (float): 收入增加后的消费水平。
    
    返回:
    float: 边际消费倾向（MPC）。
    """
    # 计算收入的变化量
    income_change = final_income - initial_income
    # 计算消费的变化量
    consumption_change = final_consumption - initial_consumption
    # 计算边际消费倾向（MPC）
    mpc = consumption_change / income_change
    return mpc
````

在这个函数中，initial_income、initial_consumption、final_income 和 final_consumption 是形式参数。它们是在函数定义时创建的变量，用于接收函数调用时传递的实际参数。

````python
# 调用函数

# 实际参数
# 某人在获得奖金前的收入和消费
income_before = 5000
consumption_before = 3000

# 某人在获得奖金后的收入和消费
income_after = 5500
consumption_after = 3200

mpc = calculate_mpc(income_before, consumption_before, income_after, consumption_after)
print(f"边际消费倾向（MPC）是：{mpc:.2f}")
````

### 理解形式参数和实际参数
剧本的角色相当于形式参数，在调用函数时传递进来的参数就是演角色的演员。

### 位置参数
位置参数必须按照正确的顺序将其传到函数中，即调用的数量和位置必须和定义是一样的。

````python
# 如果位置参数的顺序或值不正确，那么函数的输出结果也可能是不正确的。因此，确保按照正确的顺序传递参数是非常重要的。 

mpc = calculate_mpc(consumption_before, income_before, income_after, consumption_after)
print(f"边际消费倾向（MPC）是：{mpc:.2f}")
````

### 关键字参数
关键字参数是指使用形式参数的名字来确定输入的参数值。通过该方式指定实际参数时,它不再需要与形式参数的位置完全一致，只要将参数名书写正确即可。

````python
def calculate_mpc(delta_income, delta_consumption, as_percentage=False):
    """
    计算边际消费倾向（MPC）。
    
    参数:
    delta_income (float): 收入的变化量。
    delta_consumption (float): 消费的变化量。
    as_percentage (bool): 是否将结果返回为百分比形式，默认为False。
    
    返回:
    float: 边际消费倾向。
    """
    mpc = delta_consumption / delta_income
    if as_percentage:
        return mpc * 100
    return mpc
````

````python
# 假设某人收入增加了2000元，消费增加了1500元
delta_income = 2000
delta_consumption = 1500

# 计算MPC，不转换为百分比形式
mpc = calculate_mpc(delta_income, delta_consumption)
print(f"MPC（非百分比形式）: {mpc}")
````

````python
mpc_percentage = calculate_mpc(delta_income, delta_consumption, as_percentage=True)
print(f"MPC（百分比形式）: {mpc_percentage}%")
````

### 可变参数
在python中，还可以定义可变参数，可变参数也被称为不定长参数，即传入函数中的实际参数可以是任意个。
#### *parameter
```*parameter```表示接收任意多个实际参数并将其放到一个元组中。

````python
def print_triangle(*rows, char = '*'):
    """
    使用可变参数打印一个三角形图案。
    
    参数:
    *rows: 可变数量的参数，每个参数代表三角形的一行，应为整数。
    """
    for i in rows:
        print((len(rows) - i)*' ' + char * (2*i - 1))

print_triangle(1, 2, 3)
````

````python
print_triangle(1, 2, 3, 4, 5, char = "0")
````

#### **parameter
```**parameter```表示接收任意多个类似关键字参数一样显示赋值的实际参数，并将其放到一个字典中。

````python
def eco_value(**sign):
    for key, value in sign.items():
        print(f"{key}:{value}")
eco_data = eco_value(GDP_Growth = 2.5, Unemployment = 4.0,  Inflation = 1.5, Trade_Balance =  200)
````

### 理解返回值
* 你在餐厅工作，客人点一个菜，你就按一个步骤来制作它，制作完成之后可以把这个菜交给别人。
* 函数的返回值就像是你完成工作后交给别人的成果，它是函数执行完毕后给出的答案。

````python
def supply_function1(price):
    """
    线性供给函数示例。
    
    参数:
    price -- 商品的价格

    返回值:
    quantity -- 在给定价格下的生产者愿意提供的商品数量
    """
    # 假设供给函数为 Q = a + bP，其中Q是供给量，P是价格，a和b是常数
    a = -100  # 常数项，意味着价格最低为20    
    b = 5  # 斜率，表示价格每增加1单位，供给量增加5单位
    quantity = a + b * price
    return quantity

supply_function1(60)
````

## 变量的作用域
变量的作用域是指程序代码能访问该变量的区域，如果超出该区域，在访问时就会出现错误。在程序中，一般根据变量的“有效范围”将变量分为“局部变量”和“全局变量”。

# 局部变量
局部变量是指在函数内部定义和使用的变量。

````python
# 局部变量

def supply_function2(price):
    """
    线性供给函数示例。
    
    参数:
    price -- 商品的价格

    返回值:
    quantity -- 在给定价格下的生产者愿意提供的商品数量

    局部参数：
    a和b
    """
    par1 = -100  # 常数项，意味着价格最低为20    
    par2 = 5  # 斜率，表示价格每增加1单位，供给量增加5单位
    quantity = par1 + par2 * price
    return quantity
supply_function2(30)
par1 = 1
print(par1)
````

# 全局变量
全局变量能够作用于函数内外部的变量。
* 如果一个变量在函数外部定义的，那么不仅可以在函数外部访问该变量，而且也可以在函数内部访问该变量。

````python
par1 = -100  # 常数项，意味着价格最低为20    
par2 = 5  # 斜率，表示价格每增加1单位，供给量增加5单位

def supply_function3(price):
    """
    线性供给函数示例。
    
    参数:
    price -- 商品的价格

    返回值:
    quantity -- 在给定价格下的生产者愿意提供的商品数量

    局部参数：
    a和b
    """
  
    quantity = par1 + par2 * price
    return quantity
supply_function3(30)
````

### lambda函数
在Python中，lambda 是一个关键字，用于创建匿名函数，即没有名称的函数。lambda 函数通常用于编写简单的、临时的函数，它们在定义后可以立即使用，而不需要使用 def 关键字来定义一个完整的函数。

````python
Supply = lambda P : -100 + 5*P
Supply(30)
````

### 列表推导式中使用lambda函数

````python
# 定义价格列表
prices = [10, 20, 30, 40, 50]

# 使用 lambda 函数定义需求函数 Q = 100 - 0.5P
demand = lambda P: 100 - 0.5 * P

# 使用列表推导式计算每个价格下的总收益
total_revenues = [P * demand(P) for P in prices]

print(total_revenues)
````

### 字典推导式中使用lambda函数

````python
# 定义一个包含产品成本和销售价格的字典
products = {
    'A': {'cost': 50, 'price': 80},
    'B': {'cost': 70, 'price': 120},
    'C': {'cost': 30, 'price': 60}
}

# 使用 lambda 函数定义利润计算公式
profit_margin = lambda cost, price: price - cost

# 使用字典推导式计算每个产品的利润
product_profit_margins = {product: profit_margin(data['cost'], data['price']) for product, data in products.items()}

print(product_profit_margins )
````
