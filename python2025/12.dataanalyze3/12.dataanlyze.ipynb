### 就业质量

本文参照已有研究并结合数据可得性，基于收入、就业身份、单位性质、社会保障和行业性质五个变量构建就业质量指标。收入变量根据月收入加以设定，并取自然对数。在就业身份变量的设定上，雇员赋值为3，雇主和自营劳动者赋值为2，家庭帮工和其他赋值为1。在单位性质变量的设定上，国有部门（包括机关、事业单位以
及国有及国有控股企业）赋值为6，外资企业（包括港澳台独资企业、外商独资和中外合资企业）赋值为5，股份制企业（包括集体企业以及股份、联营企业）赋值为4，私营企业（包括个体工商户、私营企业和社团民办组织）赋值为3，其他单位赋值为2，无单位赋值为 1。社会保障为二值变量，参加城镇职工医疗保险赋值为 1，不参加赋值为
0。行业性质是二值变量，参考宁光杰和姜现（2019）的划分，将采矿业，电煤水热生产供应业，交通运输、仓储和邮政业，金融业，水利、环境和公共设施管理业，公共管理和社会组织归为垄断行业，赋值为1，将其余行业如居民服务、修理和其他服务业，住宿餐饮，文体娱乐等行业归为竞争性行业，赋值为0。采用主成分分析方法对五个变量进行合成，将合成变量转化为0-100之间的数值后，得到就业质量的测度指标。表1给出了就业质量指标构建的具体说明。

# 2013年就业质量

```python
import numpy as np
# 工资性收入
person_2013['wage'] = person_2013[['a3020', 'a3022', 'a3023']].sum(axis = 1)


# 将工资性收入中缺失值替换为0
person_2013['wage'] = person_2013['wage'].fillna(0)

# 就业身份
person_2013['job_status'] = np.nan
person_2013.loc[person_2013['a3003'] == 1, 'job_status'] = 3
person_2013.loc[person_2013['a3003'] == 2, 'job_status'] = 2
person_2013.loc[(person_2013['job_status'] != 2) & (person_2013['job_status'] != 3), 'job_status'] = 1

# 单位性质
person_2013['job_type'] = np.nan
person_2013.loc[(person_2013['a3014']==1) |(person_2013['a3014']==2)|(person_2013['a3014']==5), 'job_type'] = 5
person_2013.loc[person_2013['a3014'].isin([1, 2, 5]), 'job_type'] = 5
#person_2013.loc[person_2013['a3014'] == 6, 'job_type'] = 4
person_2013.loc[person_2013['a3014'].isin([3, 4]), 'job_type'] = 3
person_2013.loc[person_2013['a3014'].isin([4, 10]), 'job_type'] = 2
person_2013.loc[~person_2013['job_type'].isin([2, 3, 5]), 'job_type'] = 1

# 社会保障
person_2013['social_security'] = np.nan
person_2013.loc[person_2013['f2003'].isin([1, 2, 3]), 'social_security'] = 1
person_2013.loc[person_2013['social_security'] != 1, 'social_security'] = 0

# 行业性质
person_2013['industry'] = np.nan
person_2013.loc[person_2013['a3006'].isin([2, 4, 6, 10, 14, 19]), 'industry'] = 1
person_2013.loc[person_2013['industry'] != 1, 'industry'] = 0


person_2013['year'] = 2013

job_2013 = person_2013[['wage', 'job_status', 'job_type', 'social_security', 'industry', 'year', 'hhid_2013']]
province = master_2013[['hhid_2013', 'prov_CHN']]

merged_data2013 = pd.merge(province, job_2013, on=['hhid_2013'], how='inner')
wage_2013 = merged_data2013[['wage', 'year', 'prov_CHN']]
```

```python
df_combined = pd.concat([wage_2013, wage_2015, wage_2017, wage_2019], ignore_index=True)
print(df_combined)
```

```python
# 假设df_combined是你的原始数据框，并且它有一个名为'year'的列
df = pd.DataFrame()

for i in range(2013, 2020, 2):
    # 筛选出特定年份的数据，并按'prov_CHN'分组计算'wage'的平均值
    grouped_data = df_combined[df_combined['year'] == i].groupby('prov_CHN', as_index=False)['wage'].mean()
    
    # 为分组数据添加年份信息
    grouped_data['year'] = i
    
    # 如果df为空，则直接赋值；否则，使用concat函数将新数据追加到df中
    if df.empty:
        df = grouped_data
    else:
        # 确保所有数据都有prov_CHN列，以便能够合并
        df = pd.concat([df, grouped_data], ignore_index=True)

# 现在df是一个包含所有年份数据的DataFrame，其中包含'prov_CHN'和每个年份的'wage'平均值
```

```python
# 学生信息DataFrame
students = pd.DataFrame({
    'student_id': [1, 2, 3],
    'name': ['Alice', 'Bob', 'Charlie']
})

# 学生成绩DataFrame
grades = pd.DataFrame({
    'student_id': [2, 1, 4],
    'grade': ['B', 'A', 'C']
})
```

```python
print(students)
```

```python
print(grades)
```

1. how='inner'（内连接，默认值）
内连接将只保留两个DataFrame中都有的键（student_id）的行

```python
merged_inner = pd.merge(students, grades, on='student_id', how='inner')
print("Inner Join:")
print(merged_inner)
```

2. how='outer'（外连接）
外连接将保留两个DataFrame中所有键的匹配，如果一个DataFrame中没有匹配的键，则在结果中用NaN填充。

```python
merged_outer = pd.merge(students, grades, on='student_id', how='outer')
print("\nOuter Join:")
print(merged_outer)
```

3. how='left'（左连接）
左连接将保留左侧DataFrame（students）的所有行，即使右侧DataFrame（grades）中没有匹配的键，也会在右侧填充NaN。

```python
merged_left = pd.merge(students, grades, on='student_id', how='left')
print("\nLeft Join:")
print(merged_left)
```

4. how='right'（右连接）
右连接将保留右侧DataFrame（grades）的所有行，即使左侧DataFrame（students）中没有匹配的键，也会在左侧填充NaN。

```python
merged_right = pd.merge(students, grades, on='student_id', how='right')
print("\nRight Join:")
print(merged_right)
```
