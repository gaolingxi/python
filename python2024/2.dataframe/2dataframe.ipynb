## 整数类型
### 整数类型的定义
整数包括正整数、负整数和0，并且它的位数是任意的，是用来表示整数数值，即没有小数部分的数值。如果要指定一个非常大的整数，只需要写出其所有位数即可。（当指定的整数超过计算机自身的计算功能时，会自动转用高精度计算。）


````python
large_number = 1234567890123456789012345678901234567890 # 定义一个非常大的整数
print(large_number)  # 打印这个整数
````

整数类型包括十进制整数、八进制整数、十六进制整数和二进制整数
1. 十进制整数。（不能以0作为十进制整数的开头）

````python
#十进制正确写法
num1 = 123
print(num1)
num2 = -123
print(num2)
num3 = 0
print(num3)
````

````python
#错误写法
print(0123)
````

2. 八进制整数、十六进制好和二进制整数。<br>
* 八进制：进位规则是“逢八进一”，“Oo”或者“0O"开头提醒计算机这是八进制整数。
* 十六进制：进位规则是“逢八进一”，“Ox“或者“0X”开头提醒计算机这是十六进制整数。（十六进制0~9，a~f）
* 二进制整数：进位规则是“逢二进一”，“0b”或者“0B”开头提醒计算机这是二进制整数。



````python
num1 = 0o127      #八进制
print(num1)
num2 = 0x12a     #十六进制
print(num2)
num3 = 0b101      #二进制
print(num3)
````

### 整数类型的独有功能
用bit_length()函数获取整数有多少位。<br>
用法:
```python
整数位的数目 = 整数.bit_length()
```

* 在计算机的内存（RAM）和存储设备（如硬盘、SSD、USB驱动器）中，数据都是以位和字节的形式存储的。<br>
1. 位（bit）:<br>
位是计算机存储的最小单位，表示一个二进制数字，即0或1。<br>
2. 字节（byte）:<br>
字节是计算机存储的基本单位，通常由8个位（bit）组成。
3. 位与字节的关系：<br>
1字节 = 8位 <br>
4. 随着存储技术的发展，存储容量的单位也从字节扩展到了更大的单位，如千字节（KB）、兆字节（MB）、吉字节（GB）、太字节（TB）等。<br>
1KB = 1024B（字节） <br>
1MB = 1024KB <br>
1GB = 1024MB <br>

````python
# 获取整数的二进制有多少个位
num1 = 5
result = num1.bit_length()     # 类的方法
print(result)                  # 函数
````

### 整数类型的通用功能-算术运算符
整数类型支持多种常用的运算符，这些运算符可以执行基本的数学运算和比较操作

````python
num1 = 12 + 15           #加
num2 = 15 - 4            #减
num3 = 5 * 12            #乘
num4 = 7 / 2             #除
num5 = 7 % 2             #取余
num6 = 7 // 2            #取整
num7 = 7 ** 2            #幂，7的平方
print(num1)
print(num2)
print(num3)
print(num4)
print(num5)
print(num6)
print(num7)
````

除了基本的运算外，Python也能通过运算符和数学函数进行高级数学运算。<br>


【模块】：模块可以定义函数、类和变量。（有些别人已经开发好的模块，我们可以直接拿来用）

````python
import math                            # 导入Python的math模块，这个模块提供了许多数学函数和常数 
mu = 0                                 # 设置正态分布的均值μ为0
sigma = 1                              # 设置正态分布的标准差σ为1
x = 1.2                                # 设置我们想要计算概率密度的x值为1.2
exponent = (-(x-mu) ** 2)/(2*sigma**2) # 计算正态分布概率密度函数中的指数部分
print(exponent)                        # 打印出计算得到的指数部分的值
````

### 算术运算符的优先级
Python的运算规则是：<br>
* 优先级高的运算先执行；
* 优先级低的运算后执行；
* 同一优先级的操作按照从左到右的顺序进行；
* 如果有括号，首先执行括号内的运算。

| 类型 | 说明 |优先级|
| --- | --- |---|
| ** | 幂 |高|
| *、/、%、// | 算术运算符 |中高|
| +、- | 算术运算符 |中低|
| <、<=、>、>=、!=、== | 比较运算符 |低|

````python
num = 2**3 * 4**2 / 8 * 2
# 首先计算指数运算: 2**3 == 8, 4**2 == 16
# 接下来从左至右进行乘号和除号的运算： 8*16 == 128，128/8 == 16， 16*2 == 32
bool = 2**3 > 65%2 <= 7
# 首先计算指数运算：2**3 == 8
# 接下来进行取余的运算： 65%2 == 1
# 最后从左至右进行比较运算符的运算： 8 > 1,正确；1<=7，正确

print(int(num))
print(bool)
````

## 浮点数
### 浮点数的定义
浮点数有整数部分和小数部分组成，主要用于处理包括小数的数。<br>
【注】因为计算机硬件特点，浮点数不能执行精确运算。计算机其实不认识十进制数，它只认识二进制数，也就是说，当我们进行十进制数的运算时，计算机需要将十进制数转换成二进制数。
比如0.2的二进制表示为：<br>
0.0011001100110011001100110011001100110011...

````python
num1 = 0.1 + 0.3     #结果得到想要的0.4
num2 = 0.1 + 0.2     #结果得到0.30000000000000004
print(num1)
print(num2)
````

### 解决方法
对于浮点数来说，大多数计算机的误差不会超过$2^{53}$，如果需要非常精确的结果可以用如下模块：
* decimal模块
* fraction模块

decimal模块的用法：<br>
```python
较为精确的小数 = decimal.Decimal('原小数')
```



````python
from decimal import *
num1 = Decimal('0.1')
num2 = Decimal('0.2')
num3 = num1 + num2
print(num3 * 1000056)
````

faction模块的用法：<br>
```python
分数（更精确的小数）= Fraction(分子，分母)
```

````python
from fractions import Fraction
frac = Fraction(1,10)+Fraction(2,10)
print(frac*100056)
````
