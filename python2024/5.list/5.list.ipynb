# 列表
列表（List）是一种可变的序列类型，它可以存储不同类型的元素，如整数、浮点数、字符串、甚至其他列表。<br>
这些序列结构有以下几个共同的操作：索引、切片、相加和相乘操作。
## 列表的通用功能（序列类型）

````python
# 索引
list1 = ["Gross Domestic Product", "Inflation", "Unemployment Rate", "Monetary Policy"]
print(list1[2])

# 切片
print(list1[1:3])

# 相加
list2 = ["Public Economics", "Labor Economics"]
print(list1 + list2)

# 相乘
print(list2 * 2)
````

**【例】：多项式函数**
$$p(x)=a_0+a_1x+a_2x^2+\dots+a_Nx^N=\sum_{n=0}^Na_nx^n$$
当$N=4$，$(a_0,a_1,a_2,a_3,a_4)=(1,2,4,5,6)$<br>
计算$x=23$的值

````python
coef = [1, 2, 4, 5, 6]
p = 0
x = 23
for i, a in enumerate(coef):
    p += a * x ** i
print(p)
````

**【例】：收敛函数**
一个简单的迭代公式$x_{n+1}=rx_n$，其中$r=0.1$，这个迭代公式可以模拟一个随着时间衰减的过程，最终收敛到$0$。

````python
# 演示数列收敛到0的过程
values = []
x = 1
r = 0.5
max_iteration = 500
tolerance = 1e-3
for i in range(max_iteration):
    x *= r
    values.append(x)
    if x - 0 > tolerance:
        print(f"{i+1} iterations.")
        print(f"{values}")
    else: 
        print(f"Converged after {i+1} iterations, values is {values[-1]}")
        print(f"{values}")
        break
````

**作业题验证码：采用字符串，不使用循环**

````python
# 有六个会员注册网站Alice、mr、MrAdam、HGW.ALICE、Alice。不区分大小写，验证会员名是否唯一。
str1 = "Alice|mr|MrAdam|HGW|ALICE|Alice|"
str2 = str1.upper()
num1 = str2.count("Alice|".upper())
print(num1)
num2 = str2.count("mr|".upper())
print(num2)
num3 = str2.count("MrAdam|".upper())
print(num3)
num4 = str2.count("HGW|".upper())
print(num4)
num5 = str2.count("ALICE|".upper())
print(num5)
num6 = str2.count("Alice".upper())
print(num6)
````

**作业题验证码：采用字符串，使用循环**

````python
# 有六个会员注册网站Alice、mr、MrAdam、HGW.ALICE、Alice。不区分大小写，验证会员名是否唯一。
str1 = "Alice|mr|MrAdam|HGW|ALICE|Alice|"
str2 = str1.upper()
start = 0 
while start < len(str2):
  
    end = str2.find("|",start)
    str_sub = str2[start: end+1]
    if str2.count(str_sub) > 1:
        print(f"{str_sub}")
    start = end + 1
    if end == -1:
        break
````

**作业题验证码：采用列表对象**

````python
# 有六个会员注册网站Alice、mr、MrAdam、HGW.ALICE、Alice。不区分大小写，验证会员名是否唯一
str1 = "Alice|mr|MrAdam|HGW|ALICE|Alice"
str2 = str1.upper()
list1 = str2.split("|")

for i in list1:
    num = list1.count(i)
    if num > 1:
        print(f"重复会员名为{i}")
    else:
        print("没有重复的")
````

2.用index()方法获取指定元素在列表中首次出现的位置。
**index（）函数的用法：**


```python

```

````python
# 有六个会员注册网站Alice、mr、MrAdam、HGW.ALICE、Alice。不区分大小写，验证会员名是否唯一,找到是第几个会员重复了。
str1 = "Alice|mr|MrAdam|HGW|ALICE|Alice"
str2 = str1.upper()
list1 = str2.split("|")

for i in list1:
    num = list1.count(i)
    ind = list1.index(i)
    if num > 1:
        print(f"重复会员名为{i},是第{ind}位会员")
    else:
        print("没有重复的")
````

### 对列表进行排序
1. 使用列表对象的sort()方法实现对列表进行排序<br>
列表对象提供了sort()方法，用于对原列表进行排序。

**sort()函数的用法：**

```python
原列表.sort()
```


````python
#从高到低
list_sum.sort(reverse = True)
print(list_sum)
#从低到高
list_sum.sort()
print(list_sum)
list_sum.sort(reverse = False)
print(list_sum)
````

2. 使用内置的sorted函数实现对列表进行排序<br>
用于对列表进行排序，使用该函数进行排序后，原列表中的顺序不变。<br>

**sorted()函数的用法：**
```python
新的列表 = sorted(原列表)
```

````python
# 三名学生的语文、数学和外语的成绩为：
grade1 = [98, 88, 70]
grade2 = [88, 64, 80]
grade3 = [80, 90, 96]

list_grade = [grade1, grade2, grade3]
list_id = ["grade1", "grade2", "grade3"]
list_sum = []
for i in list_grade:
    gra_sum = sum(i)
    list_sum.append(gra_sum)
print(list_sum)
````

````python
#从高到低
list_sumN = sorted(list_sum, reverse = True)
print(list_sumN)
#从低到高
list_sumN = sorted(list_sum)
print(list_sumN)
list_sumN = sorted(list_sum, reverse = False)
print(list_sumN)
````

````python
# 找到最高分同学的ID
index1 = list_sum.index(list_sumN[-1])
print(list_id[index1])
````

## 列表推导式
列表推导式（List Comprehensions）是一种创建列表的简洁和直观的语法。<br>

**列表推导式的用法：**
```python
[表达式 for 迭代值 in 可迭代对象 [if 条件表达式]]
```
``` [if 条件表达式] ```不是必须的，可以使用也可以省略。


````python
list1 = [1 for i in range(10)]
print(list1)
````

````python
# 生成指定范围的数值列表
import random as rd
randomNumber = [rd.randint(0, 10) for i in range(10)]
print(f"列表为：\n{randomNumber}")
````

````python
# 根据列表生成指定需求的列表
# 三名学生的语文、数学和外语的成绩为：
grade1 = [98, 88, 70]
grade2 = [88, 64, 80]
grade3 = [80, 90, 96]

list_grade = [grade1, grade2, grade3]
list_sum = [sum(i) for i in list_grade]
print(list_sum)
````

````python
# 从列表中选择符合条件的元素组成新的列表
# 用列表推导式生成一个商品价格高于5000的列表
price = [1200, 5330, 2988, 6200,1998, 8888]
sale = [x for x in price if x>5000]
print(f"原价格为{price}")
print(f"价格高于5000的商品有{sale}")
````

**理解二维列表**

酒店有很多个楼层（1-5）。每个楼层有很多个房间(1-10)。

```python
二维列表 = [[元素11, 元素12, 元素13，..., 元素1n],
```

````python
list2D = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(list2D)
````

````python
grade1 = [98, 88, 70]
grade2 = [88, 64, 80]
grade3 = [80, 90, 96]

list_grade = [grade1, grade2, grade3]
print(list_grade)
````

2. 使用嵌套for循环进行创建<br>
创建一个包含4行5列（$4\times 5$ ）的二维列表，其中所有的元素都是1.

````python
mat = []
for i in range(4):
    mat.append([])
    for j in range(5):
        mat[i].append(1)
print(mat)
````

3. 使用列表推导式进行创建

````python
mat = [[1 for j in range(5)] for i in range(4)]
````

# 元组
元组也是一种序列结构，是一系列按特定顺序排列的序列结构。但是它为不可变序列。

````python
tuple1 = (1,2,3,4,5,6,7)
tuple2 = ("Supply and Demand", "Monetary Policy",  "Fiscal Policy", "Phillips Curve")
print(tuple1)
print(tuple2)
````

````python
gdp_data = (
    ("China", 1434290),
    ("United States", 2143320),
    ("Japan", 508000),
    ("Germany", 384750),
    ("India", 2875220)
)

# 使用for循环遍历元组
for country, gdp in gdp_data:
    print(f"The GDP of {country} is {gdp}.")
````

**【理解元组】**

````python
tuple3 = tuple1 + tuple2
print(tuple3)
````

````python
# 列表排序和添加元素
list1 = [1,2,5,2,3,9,6,7]
list1.sort()
list1.append(2)
print(list1)
````
