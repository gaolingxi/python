# 面向对象编程
面向对象编程是一种编程思想。面向对象编程是从面向过程编程过渡来的。

## 以洗衣服作为一个案例
### 面向过程编程
如果用面向过程的编程思想，我们将洗衣服拆解成一系列步骤，每个步骤就是一个函数。<br>
```第一步```: 打开洗衣机；<br>
```第二步```: 放衣服和洗衣液；<br>
```第三步```: 选择洗衣模式，开始洗衣服；<br>
```第四步```: 等待洗完，拿出衣服。<br>

### 面向对象编程
拆分人和洗衣机两个对象：<br>
#### 人
```第一步```: 打开洗衣机；<br>
```第二步```: 放衣服和洗衣液；<br>
```第三步```: 洗完衣服后拿出衣服；<br>
#### 洗衣机
```第一步```: 根据洗衣模式洗衣服；<br>

### 总结
* 面向过程编程：更注重事情的每个步骤以及顺序
* 面向对象编程：更注重事情有哪些参与者，参与者需要做些什么事情。将其拆解成一个个模块和对象。

## 用一个简单的例子来通俗地解释面向对象编程（Object-Oriented Programming，简称OOP）。

想象一下，你正在玩一个电子游戏，游戏里有很多角色，比如战士、法师和弓箭手。每个角色都有他们自己的属性（比如生命值、攻击力）和技能（比如战士的剑击、法师的火球术、弓箭手的箭雨）。在面向对象编程中，我们可以把每个角色看作是一个“对象”。

### 1. 什么是“对象”？

在面向对象编程的世界里，“对象”就像是现实世界中的事物。每个对象都有自己的“属性”（就像是事物的特性）和“方法”（就像是事物能做的事情）。比如，一辆车是一个对象，它有颜色（属性）和启动（方法）。

### 2. 什么是“类”？

“类”就像是对象的“蓝图”或“模板”。通过这个蓝图，我们可以创建很多相似的对象。比如，我们有一个“角色”类，它定义了所有角色共有的属性和方法。然后，我们可以用这个类来创建具体的战士、法师和弓箭手对象。

### 3. 面向对象编程的好处

- **封装**：每个对象都把自己的数据（属性）和操作数据的方法（行为）封装在一起，这样可以保护数据不被随意访问和修改。
- **继承**：我们可以创建新的类来继承已有类的属性和方法，这样可以减少重复代码，让编程更高效。
- **多态**：不同的对象可以对同一个方法有不同的响应，这让程序更灵活。

### 总结

面向对象编程就像是在用代码创造一个虚拟世界，你定义了各种“类”（蓝图），然后根据这些蓝图创建具体的“对象”（实例）。每个对象都有自己的属性和方法，就像是每个角色都有自己的生命值和技能。通过这种方式，我们可以更自然地模拟现实世界的复杂性，并且让代码更加模块化、易于理解和维护。


# 面向对象程序设计-类(Class)

````python
class Consumer:
    def __init__(self, w):
        self.wealth = w

    def earn(self, y):
        self.wealth += y

    def spend(self, x):
        new_wealth = self.wealth - x
        if new_wealth < 0:
            print("Insufficent Funds")
        else:
            self.wealth = new_wealth
````

````python
c1 = Consumer(10)
c1.earn(5)
c1.spend(3)
c1.wealth
````

### 一个市场
```线性需求函数```: $Q^d = ap + b$ <br>
```线性供给函数```: $Q^s = cp + d$
* 求解市场的短期均衡价格和均衡产量。
* 该市场的消费者剩余、生产者剩余和社会总福利分别是多少？
* 如果政府对每一单位商品征收t元的税，那么该市场的短期均衡价格和均衡产量是多少？另外，消费者剩余、生产者剩余和社会总福利的变化又分别是多少？

````python

class Market:
    def __init__(self, a, b, c, d):
         '''
         设置市场参数
         '''
         self.a, self.b, self.c, self.d = a, b, c, d
         if b < d:
              raise ValueError('Insufficient demand')
     
    def price(self):
         "计算均衡价格"
         return (self.d - self.b) / (self.a - self.c)
    
    def quantity(self):
         "计算均衡数量"
         return self.a * self.price() + self.b
    
    def consumer_surp(self):
         "计算消费者剩余"
         return (-self.b/self.a - self.price()) * self.quantity() * 0.5
    
    def producer_surp(self):
         "计算厂商剩余"
         return (self.price() + self.d/self.c) * self.quantity() * 0.5
    def total_surp(self):
         return self.producer_surp() + self.consumer_surp()
````

````python
class TaxedMarket(Market):
    def __init__(self, a, b, c, d, tax):
        super().__init__(a, b, c, d)
        self.t = tax
    
    def price(self):
        "计算含税均衡价格"
        return (self.d - self.b - self.c * self.t ) / (self.a - self.c)
    
    def producer_surp(self):
        "计算含税厂商剩余"
        # 厂商剩余需要考虑税收影响，因此价格应该是不含税的价格
        return (self.price() + self.d/self.c - self.t) * self.quantity() * 0.5
    
````

````python
baseline_params = -4, 68, 4, -12
m = Market(*baseline_params)
m.price()
m.quantity()
m.consumer_surp()
m.producer_surp()
````

````python
baseline_params = -4, 68, 4, -12, 2
m = TaxedMarket(*baseline_params)
m.price()
m.quantity()
m.consumer_surp()
# m.producer_surp()
````

```一个企业的线性成本函数:```$c(Q)=cQ$

* 古诺均衡<br>
两个厂商都是已知对方产量的情况下，各自确定能够给自己带来最大利润的产量（每一个厂商都是以自己的产量去适应对方已确定的产量）。<br>
$$Y = Y_1 + Y_2$$
* 卡特尔均衡<br>
两个厂商联合起来，实现市场上的利润最大化。

* 斯塔伯格均衡<br>
一个厂商为实力相对雄厚而处于支配地位的领导者，另一个厂商则为追随者。（“领导者-追随者”）



````python
from sympy import symbols, Eq, solve
class Market:
    def __init__(self, a, b, c):
        self.a, self.b, self.c = a, b, c
        q_1, q_2 = symbols('q_1 q_2')

    def margin_cost(self):
        return self.c
    
    def quantity(self):
        return (self.c - self.b)/(2 * self.a)
    def price(self):
        return self.a * self.quantity() + self.b
````

````python
class Market:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def equilibrium_price(self):
        raise NotImplementedError("Each market type must implement this method")

    def equilibrium_quantity(self):
        raise NotImplementedError("Each market type must implement this method")


class PerfectCompetitionMarket(Market):
    def equilibrium_price(self):
        # 在完全竞争市场中，均衡价格等于边际成本
        return self.c

    def equilibrium_quantity(self):
        # 均衡产量由市场需求函数决定，P = MC
        price = self.equilibrium_price()
        return ( price - self.b) / self.a


class CournotMarket(Market):
    def equilibrium_price(self):
        # 古诺均衡价格计算
        q1, q2 = self.equilibrium_quantity()
        total_quantity = q1 + q2
        return self.a * total_quantity + self.b

    def equilibrium_quantity(self):
        # 古诺均衡产量计算
        # 假设两个企业对称，成本相同
        q1 = q2 = (self.c - self.b)/(3 * self.a)
        return q1, q2


class CartelMarket(Market):
    def equilibrium_price(self):
        # 卡特尔均衡价格计算
        q = self.equilibrium_quantity()
        return self.a * q + self.b

    def equilibrium_quantity(self):
        # 卡特尔均衡产量计算
        # 总产量使得价格最大化
        q = (self.c - self.b) / (2 * self.a)
        return q


````

````python
# 参数设置
a = -1  # 需求函数中的a
b = 100    # 需求函数中的b
c = 0    # 成本函数中的c

# 创建市场实例
perfect_competition_market = PerfectCompetitionMarket(a, b, c)
cournot_market = CournotMarket(a, b, c)
cartel_market = CartelMarket(a, b, c)
perfect_competition_market.equilibrium_quantity()
cournot_market.equilibrium_quantity()
# cartel_market.equilibrium_price()
````
