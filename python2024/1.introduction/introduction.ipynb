## Python的创建
Python 是一门**优雅而健壮**的编程语言，由荷兰国家数学与计算机科学研究中心的**吉多·范罗苏姆**于1990年代初设计。它继承了传统编译语言的**强大性和通用性**，同时也借鉴了脚本语言和解释语言的**易用性**。Python被设计成是**符合大脑思维习惯**的，采用极简主义的设计理念，加以统一规范的交互模式。这使得Python易于学习、理解和记忆。Python 开发者的哲学是 **用一种方法，最好是只有一种方法来做一件事**。

# 理解编程中的等号
编程中等号的用途：
等号（=）用作赋值操作符，用于将值赋给变量。赋值操作会改变变量的状态，将其存储的旧值替换为新值。
# 编程中的等号与数学中的等号的区别
* 在数学中，等号表示一种逻辑关系，只表示左右两边相等
* 在计算机中，等号表示赋值操作，是单项操作，从右至左执行，将右侧的值赋给左侧

## 注释规则
<img src="figure/poem1.jpg" alt="示例图片" width="1000" height="400"><br>



Python的注释有以下两个作用：<br>
1. Python的注释就是在代码中添加注释性的文字，进而帮助我们更好地阅读代码;
2. 在Python调试的过程中，注释可以用来临时禁用某些代码段，而不必删除它们。

* 单行注释<br>
在Python中，使用<span style="color: red;">#</span>作为单行注释的符号。从符号#到换行，其后面所有的内容都作为注释的内容而被Python编译器忽略。

````python
# 单行注释的第一种形式
# 要求输入身高
# height = input("请输入您的身高：")
# 单行注释的第二种形式
height = input("请输入您的身高：")      #要求输入身高
````

* 多行注释<br>
将包含在三对引号（<span style="color: red;">即'''注释内容'''或者"""注释内容"""</span>），这样的多行注释将被Python解释器忽略。

````python
# 三对单引号
"""
姓名：高泠溪学号: XXXXXXXX学院：经济学院数字经济系提交日期：2024年9月20日主题：实验一
"""
print("Hello World")
````

````python
# 三对双引号
"""
      姓名：高泠溪
      学号: XXXXXXXX
      学院：经济学院数字经济系
      提交日期：2024年9月20日
      主题：实验一
"""
print("Hello World")
````

````python
# 注意不要将引号与注释内容放在同一行,会被识别为字符串，也就是一段文字。
''' 姓名：高泠溪 学号：XXXXXXXX '''
````

## 变量命名规则


1. 变量必须以字母（a-z,A-Z）或下划线开头；
2. 变量不能以数字开头；

````python
1b = 123
````

````python
b = 123             # 单个小写字母
B = 123             # 单个大写字母
````

3. 建议使用有意义的命名：变量名应具有描述性，能够清楚地表达变量的含义与用途，避免使用单个字母或是数字作为变量名

````python
studentid = "202410312"    # 小写字母串
student_id = "202410312"   # 带下划线的小写字符串
````

````python

STUDENTID = "202410312"     # 大写字符串
STUDENT_ID = "202410312"    # 带下划线的大写字符串
StudentId = "202410312"     # 驼峰式命名
````

4. 建议不要用字符'l'，'O'，或 'I'作为单个字符的变量名。一些字体中，这些字符不能与数字1和0区别。用'L' 代替'l'时。

````python
Ox123 = 123
````

5. 避免使用Python关键字和内置函数名作为变量名，会导致编译错误

````python
print = 123
print(print)
````

6. 在运算符两侧使用空格，如赋值=、比较==、算术运算符+、-、*、/、%等。
