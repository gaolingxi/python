## 8.1 类和对象

> **类（Class）= 模板/蓝图**；**对象（Object）= 用模板造出来的具体实例**。

```python
# 类是模板
class Student:
    pass

# 对象是实例
stu = Student()        # 用类造一个学生
```

> 大白话：类像"户型图"，对象像"按户型图盖出来的房子"。多个对象共享同一个类。

### 📖 例题1：类和对象什么关系

````python
# 📖 例题1：类和对象什么关系

class Animal:
    pass               # 定义类（空类）

a1 = Animal()
a2 = Animal()
print(a1, a2)            # 两个独立的对象
print(a1 == a2)         # False（不是同一个对象）
print(type(a1))         # <class 'Animal'>
````

### 📖 例题2：创建第一个类和对象

````python
# 📖 例题2：创建第一个类和对象
# 金融场景：创建客户
class Consumer:
    def __init__(self, w):
        self.wealth = w

    def earn(self, y):
        self.wealth += y

    def spend(self, x):
        self.wealth -= x

# 用Consumer造一个消费者
c1 = Consumer(10)
c1.earn(5)
c1.spend(3)
print(c1.wealth)  # 12（10+5-3）
````

## 8.2 属性

> 属性 = 对象的数据。用 `self.属性名` 定义，用 `对象.属性名` 访问。

```python
class Student:
    def __init__(self, sid, name):
        self.sid = sid       # 学号
        self.name = name     # 姓名

stu = Student("001", "张三")
print(stu.name)   # 张三
```

### 📖 例题3：两个对象属性独立

````python
# 📖 例题3：两个对象属性独立

class Student:
    def __init__(self, sid, name):
        self.sid = sid
        self.name = name

s1 = Student("001", "张三")
s2 = Student("002", "李四")
s1.name = "小张"               # 改s1的名字
print(s1.name)                 # 小张
print(s2.name)                 # 李四（s2没受影响！）
````

### 📖 例题4：用属性存信息

````python
# 📖 例题4：用属性存信息
# 金融场景：账户属性
c1 = Consumer(10)
c2 = Consumer(50)

c1.earn(5)
c2.earn(20)

print(f"c1财富: {c1.wealth}")  # 15
print(f"c2财富: {c2.wealth}")  # 70

# 两个Consumer对象是独立的，改c1不影响c2
c1.spend(10)
print(f"c1消费后: {c1.wealth}")  # 5
print(f"c2没变: {c2.wealth}")    # 70
````

## 8.3 方法

> 方法 = 对象的行为。定义在类里，第一个参数必须是 `self`。

```python
class Student:
    def __init__(self, sid, name):
        self.sid = sid
        self.name = name

    def greet(self):          # 方法
        return f"我是{self.name}，学号{self.sid}"

stu = Student("001", "张三")
print(stu.greet())
```

> **self = 谁调用这个方法，self就是谁**。

### 📖 例题5：self区分不同对象

````python
# 📖 例题5：self区分不同对象

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def info(self):
        return f"{self.name}: {self.score}"

s1 = Student("张三", 98)
s2 = Student("李四", 88)
print(s1.info())     # 张三: 98
print(s2.info())     # 李四: 88
````

### 方法可以修改属性

### 📖 例题6：方法能改属性

````python
# 📖 例题6：方法能改属性

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def add_score(self, delta):
        self.score = self.score + delta   # 改自己的score

    def info(self):
        return f"{self.name}: {self.score}"

s = Student("张三", 80)
print(s.info())
s.add_score(10)              # 加10分
print(s.info())              # 90
````

### 📖 例题7：定义第一个方法

````python
# 📖 例题7：定义第一个方法
# 金融场景：操作方法
class Student:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def total(self):
        return sum(self.scores)

    def avg(self):
        return sum(self.scores) / len(self.scores)

    def max_score(self):
        return max(self.scores)

    def min_score(self):
        return min(self.scores)

stu = Student("张三", [98, 88, 70, 92, 85])
print(f"总分: {stu.total()}")      # 433
print(f"平均分: {stu.avg():.1f}")  # 86.6
print(f"最高分: {stu.max_score()}") # 98
print(f"最低分: {stu.min_score()}") # 70
````

## 8.4 构造方法 __init__

> `__init__` 是**构造方法**——创建对象时自动调用，用来初始化属性。

```python
class Student:
    def __init__(self, name, score):
        self.name = name      # 创建时就有了
        self.score = score
```

> 大白话：`__init__` 就是对象的"出生设置"——创建对象时传什么，它就设什么。

### 📖 例题8：不写__init__会这样

````python
# 📖 例题8：不写__init__会这样

class A:
    pass              # 没有__init__

a = A()
print(a)              # 对象能创建，但没属性

class B:
    def __init__(self, x):
        self.x = x

b = B(5)
print(b.x)            # 5（有__init__，属性有了）
````

### 📖 例题9：用__init__自动初始化

````python
# 📖 例题9：用__init__自动初始化
# 金融场景：自动初始化
class A:
    pass  # 没有__init__

a = A()
print(a)  # <__main__.A object at ...>（能创建，但没属性）

class B:
    def __init__(self, x, y):
        self.x = x
        self.y = y

b = B(3, 5)
print(b.x)  # 3
print(b.y)  # 5
print(b)    # <__main__.B object at ...>（能创建，有属性）
````

## 8.5 继承

> 子类继承父类——自动拥有父类的所有属性和方法，还能加自己的。

```python
class Student:           # 父类
    def __init__(self, name):
        self.name = name

class CSStudent(Student):  # 子类（计算机学生）
    def __init__(self, name, language):
        super().__init__(name)   # 调用父类的__init__
        self.language = language # 子类额外属性

stu = CSStudent("张三", "Python")
print(stu.name)         # 张三（继承自父类）
print(stu.language)     # Python（子类自己的）
```

### 📖 例题10：子类能用父类方法

````python
# 📖 例题10：子类能用父类方法

class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        return f"{self.name}说话"

class Dog(Animal):
    def speak(self):          # 重写父类的speak
        return f"{self.name}汪汪"

d = Dog("小狗")
print(d.speak())              # 小狗汪汪（用子类版本）
print(d.name)                 # 小狗（用父类属性）
````

### 📖 例题11：继承后加自己的内容

````python
# 📖 例题11：继承后加自己的内容

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def info(self):
        return f"{self.name}, {self.age}岁"

class Student(Person):
    def __init__(self, name, age, major, score):
        super().__init__(name, age)  # 调用父类__init__
        self.major = major
        self.score = score
    def grade(self):
        if self.score >= 90:
            return "优秀"
        elif self.score >= 60:
            return "及格"
        else:
            return "不及格"

s = Student("张三", 20, "经济学", 92)
print(s.info())     # 张三, 20岁（继承自父类）
print(s.major)      # 经济学（子类自己的）
print(s.grade())    # 优秀（子类自己的方法）
````

### 继承的用途

| 场景 | 父类 | 子类 |
|------|------|------|
| 学生 | Student | CSStudent（计算机）、EconStudent（经济学） |
| 家庭 | Family | SingleFamily（单亲）、CoupleFamily（双亲） |
| 函数 | Function | LinearFunction（线性）、ExponentialFunction（指数） |

> 用 `super()` 调用父类方法，避免重复代码。

## 8.6 封装

> **封装 = 把数据和方法打包在一起，外面只能看不能随便改**。

Python用命名约定模拟私有属性：
- `属性名` — 公共，外面能访问
- `_属性名` — 受保护（约定俗成），外面一般不直接用
- `__属性名` — 私有，外面直接访问会报错（但能绕过去）

### 📖 例题12：不同前缀属性效果

````python
# 📖 例题12：不同前缀属性效果

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner        # 公共
        self._balance = balance   # 受保护
        self.__pin = "1234"       # 私有

acc = BankAccount("张三", 10000)
print(acc.owner)          # 张三
print(acc._balance)       # 10000（能访问，但约定不直接用）
# print(acc.__pin)       # 报错！私有不能直接访问
````

### 📖 例题13：_和__前缀区别

````python
# 📖 例题13：_和__前缀区别

class BankAccount:
    def __init__(self, owner, balance, pin):
        self.owner = owner          # 公共属性
        self._balance = balance     # 受保护（约定不直接用）
        self.__pin = pin            # 私有（name mangling）

acc = BankAccount("张三", 10000, "1234")
print(acc.owner)        # 张三（公共，能直接访问）
print(acc._balance)     # 10000（受保护，能访问但约定不直接用）
# print(acc.__pin)     # AttributeError! 私有不能直接访问

# 但可以用方法访问私有属性
class BetterAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance
    def get_balance(self):
        return self._balance
    def deposit(self, amt):
        self._balance += amt

b = BetterAccount("李四", 5000)
print(b.get_balance())    # 5000（通过方法访问）
b.deposit(1000)
print(b.get_balance())    # 6000（通过方法修改）
````

### 魔术方法

> 魔术方法（双下划线开头结尾）在特殊操作时自动调用，让对象能像内置类型一样用。

| 魔术方法 | 触发时机 | 等价操作 |
|----------|----------|----------|
| `__init__` | 创建对象 | `Student()` |
| `__str__` | 转字符串 | `str(stu)` / `print(stu)` |
| `__repr__` | 调试显示 | `repr(stu)` |
| `__len__` | 取长度 | `len(stu)` |
| `__add__` | 加法 | `s1 + s2` |

> 大白话：定义了 `__str__`，对象就能像字符串一样被 print；定义了 `__add__`，对象就能相加。

### 📖 例题14：定义__str__后print

````python
# 📖 例题14：定义__str__后print

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def __str__(self):
        return f"学生{self.name}: {self.score}分"

stu = Student("张三", 98)
print(stu)                # 学生张三: 98分（不再是<__main__.Student>）
````

### 📖 例题15：__repr__和__str__区别

````python
# 📖 例题15：__repr__和__str__区别

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def __str__(self):
        return f"学生{self.name}: {self.score}分"
    def __repr__(self):
        return f"Student('{self.name}', {self.score})"

s = Student("张三", 98)
print(s)         # 学生张三: 98分（__str__，给人看）
print(repr(s))   # Student('张三', 98)（__repr__，给开发者看）
````

### 📖 例题16：__add__让对象相加

````python
# 📖 例题16：__add__让对象相加

class Score:
    def __init__(self, total):
        self.total = total

    def __add__(self, other):
        return Score(self.total + other.total)

    def __str__(self):
        return f"总分{self.total}"

s1 = Score(98)
s2 = Score(88)
print(s1 + s2)            # 总分186
````

### 📖 例题17：封装和魔术方法

````python
# 📖 例题17：封装和魔术方法
# 金融场景：封装保护
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def __eq__(self, other):
        return self.name == other.name and self.score == other.score
    def __len__(self):
        return len(str(self.name))

s1 = Student("张三", 98)
s2 = Student("张三", 98)
s3 = Student("李四", 88)

print(s1 == s2)  # True（同名同分）
print(s1 == s3)  # False（不同名）
print(len(s1))   # 2（名字长度）

# __str__让print好看
def __str__(self):
    return f"{self.name}: {self.score}分"
print(s1)  # 张三: 98分
````
