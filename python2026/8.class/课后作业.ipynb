## 一、选择题

### 1. Python中定义类的关键词是？

A. `class`  
B. `def`  
C. `object`

### 2. 类的构造方法是？

A. `__new__`  
B. `__init__`  
C. `init`  
D. `__start__`

**答案：B**  
`__init__` 是构造方法，创建对象时自动调用。

---

### 3. 方法中第一个参数 `self` 代表什么？

A. 类本身  
B. 调用该方法的对象  
C. 父类  
D. 全局变量

**答案：B**  
`self` 就是"谁调用这个方法，self就是谁"。

---

### 4. 子类继承父类，调用父类方法用？

A. `parent()`  
B. `super()`  
C. `base()`  
D. `inherit()`

**答案：B**  
`super()` 调用父类的方法。

---

### 5. `__str__` 魔术方法在什么时候被调用？

A. 创建对象时  
B. 打印对象时  
C. 删除对象时  
D. 赋值时

**答案：B**  
`__str__` 在 `str()` 或 `print()` 时被调用。

## 三、代码题

### 1. 定义Student类，__init__接收name，打印name。

````python
# 请在下面填写代码：
class Student:
    def __init__(self, name): self.name = name
print(Student("张三").name)
````

**答案：** 张三

---

### 2. 定义方法 greet()，返回 "你好，name"。

````python
# 请在下面填写代码：
class Person:
    def __init__(self, name): self.name = name
    def greet(self): return f"你好，{self.name}"
print(Person("张三").greet())
````

**答案：** 你好，张三

---

### 3. 定义子类 CSStudent 继承 Student，加属性 major。

````python
# 请在下面填写代码：
class Student:
    def __init__(self, name, major="默认"):
        self.name = name; self.major = major
print(Student("张三", "计算机").major)
````

**答案：** 计算机

---

### 4. 定义 __str__ 方法，让 print 显示自定义格式。

````python
# 请在下面填写代码：
class Score:
    def __init__(self, total): self.total = total
    def __str__(self): return f"总分{self.total}"
print(Score(256))
````

**答案：** 总分256

---

### 5. 定义BankAccount类，有deposit和withdraw方法，测试存取款。

````python
# 请在下面填写代码：
class BankAccount:
    def __init__(self, balance): self.balance = balance
    def deposit(self, amt): self.balance += amt
    def withdraw(self, amt): self.balance -= amt
a = BankAccount(100); a.deposit(50); a.withdraw(30)
print(a.balance)
````

**答案：** 120

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
class Student
    def __init__(self, name):
        self.name = name
```

````python
# 请在下面填写改正后的代码：
class Student:
    def __init__(self, name):
        self.name = name
````

**改正说明：** class 行尾必须有冒号 `:`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
class Student:
    def __init__(self, name):
        name = name
print(Student("张三").name)
```

````python
# 请在下面填写改正后的代码：
class Student:
    def __init__(self, name): self.name = name
print(Student("张三").name)
````

**改正说明：** `name = name` 没设到self上，用 `self.name = name`。

---

### 3. 下面的代码运行后报错，请找出问题并改正

```python
class Student:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f"你好，{self.name}"
print(greet())
```

````python
# 请在下面填写改正后的代码：
class Student:
    def __init__(self, name): self.name = name
    def greet(self): return f"你好，{self.name}"
print(Student("张三").greet())
````

**改正说明：** 方法要通过对象调用，不能直接调用 `greet()`。

---

### 4. 下面的代码运行后报错，请找出问题并改正

```python
class Parent:
    def __init__(self):
        self.x = 1
class Child(Parent):
    def __init__(self):
        self.y = 2
print(Child().x)
```

````python
# 请在下面填写改正后的代码：
class Parent:
    def __init__(self): self.x = 1
class Child(Parent):
    def __init__(self): super().__init__(); self.y = 2
print(Child().x)
````

**改正说明：** 子类 __init__ 要调 `super().__init__()` 才能用父类属性。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
class A:
    def __str__(self):
        return "A对象"
    def __repr__(self):
        return "A表示"
print(A())
```

````python
# 请在下面填写改正后的代码：
class A:
    def __str__(self): return "A对象"
    def __repr__(self): return "A表示"
print(A())
````
