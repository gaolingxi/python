## 6.1 字典的创建

> **字典 = 用花括号 `{ }` 装的"键-值对"**。
>
> 键（key）找值（value）—— 键必须唯一，值可以重复。

```python
dict1 = {"键1": "值1", "键2": "值2", ...}
```

> **大白话**：字典像电话本——姓名（键）找电话（值）。

### 📖 例题1：字典通过键取值

````python
# 📖 例题1：根据姓名找到成绩
scores = {"张三": 98, "李四": 88, "王五": 70}
print(scores["张三"])
print(scores["王五"])
````

### 多种创建方式

### 📖 例题2：三种已经能理解的创建方式

````python
# 📖 例题2：创建字典的三种方式
d1 = {"张三": 98, "李四": 88}
print("直接写：", d1)

d2 = {}
d2["张三"] = 98
d2["李四"] = 88
print("从空字典添加：", d2)

d3 = dict(张三=98, 李四=88)
print("使用 dict()：", d3)
````

### 📖 例题3：fromkeys()批量创建字典

````python
# 📖 例题3：fromkeys()批量创建字典

# 给所有科目设默认值0
scores = dict.fromkeys(["语文", "数学", "英语"], 0)
print(f"默认0分: {scores}")

# 注意：所有值指向同一个对象（可变对象时）
# 用列表值会有问题
lists = dict.fromkeys(["语文", "数学"], [])
lists["语文"].append(98)
print(f"陷阱: {lists}")  # 两个都是[98]！
````

### 从两列数据看 `zip()`

姓名和成绩已经分别存好，怎样按相同位置配对？本例先认识 `zip()`：它把两组数据按位置一一配在一起。

### 📖 例题4：zip()创建字典

````python
# 📖 例题4：zip()创建字典

keys = ["语文", "数学", "英语"]
values = [98, 88, 70]

# zip配对后转字典
d = dict(zip(keys, values))
print(f"zip创建: {d}")

# 实际应用：成绩表转字典
names = ["张三", "李四", "王五"]
scores = [98, 88, 70]
score_dict = dict(zip(names, scores))
print(f"成绩字典: {score_dict}")
````

## 6.4 遍历字典

> 遍历字典最常用 `items()`——一次拿到键和值。

### 📖 例题9：items()同时取键和值

````python
# 📖 例题9：items()同时取键和值

d = {"张三": 98, "李四": 88, "王五": 70}

# items()返回键值对
for key, value in d.items():
    print(f"{key}: {value}")

# 实际应用：打印成绩表
scores = {"张三": 98, "李四": 88, "王五": 70}
print("\n成绩表:")
for name, score in scores.items():
    print(f"  {name}: {score}分")
````

### 📖 例题10：遍历字典打印经济指标

````python
# 📖 例题10：遍历字典打印经济指标

indicators = {
    "GDP增长率": 2.5,
    "CPI": 2.1,
    "失业率": 5.2,
    "M2增速": 9.7
}

# 遍历打印
print("经济指标:")
for name, value in indicators.items():
    print(f"  {name}: {value}%")

# 筛选：增长率>2%的
print("\n增长率>2%:")
for name, value in indicators.items():
    if value > 2:
        print(f"  {name}: {value}%")
````

### 遍历的3种写法

### 📖 例题11：三种遍历方式对比

````python
# 📖 例题11：三种遍历方式对比

d = {"张三": 98, "李四": 88}

# 1. 只遍历键（默认）
print("1. 遍历键:")
for key in d:
    print(f"  {key}")

# 2. 遍历键值对
print("2. 遍历键值对:")
for key, value in d.items():
    print(f"  {key}: {value}")

# 3. 遍历值
print("3. 遍历值:")
for value in d.values():
    print(f"  {value}")
````

## 6.5 增删改

### 添加和修改

```python
字典[键] = 值     # 键存在就改，不存在就加
```

> **跟列表一样**：赋值操作不区分"添加"还是"修改"，键存在就改，不存在就加。

### 📖 例题12：重复赋值是改还是加？

````python
# 📖 例题12：重复赋值是改还是加？

d = {"张三": 98}
print(f"初始: {d}")

# 重复赋值同一个键：是改，不是加
d["张三"] = 88
print(f"改后: {d}")  # {'张三': 88}（改了值）

# 新键：是加
d["李四"] = 70
print(f"加后: {d}")  # {'张三': 88, '李四': 70}
````

### 📖 例题13：字典的增删改

````python
# 📖 例题13：字典的增删改

d = {"张三": 98}
print(f"初始: {d}")

# 增/改：直接赋值
d["李四"] = 88
d["张三"] = 99  # 改张三的成绩
print(f"增删改后: {d}")

# 用update()批量更新
d.update({"王五": 70, "赵六": 85})
print(f"update后: {d}")

# 用|合并（Python 3.9+）
d2 = d | {"钱七": 92}
print(f"|合并后: {d2}")
````

### 删除

```python
del 字典[键]                # 按键删
字典.pop(键)               # 按键删，返回被删的值
字典.popitem()             # 删最后一个
字典.clear()               # 清空整个字典
```

### 📖 例题14：pop()删除并返回值

````python
# 📖 例题14：pop()删除并返回值

d = {"张三": 98, "李四": 88, "王五": 70}

# pop返回被删除的值
removed = d.pop("张三")
print(f"删除了: {removed}")  # 98
print(f"字典: {d}")  # {'李四': 88, '王五': 70}

# 键不存在：设默认值不报错
removed = d.pop("赵六", "不存在")
print(f"删除赵六: {removed}")  # 不存在
````

### 📖 例题15：4种删除方法对比

````python
# 📖 例题15：4种删除方法对比

# del：按键删
d1 = {"张三": 98, "李四": 88}
del d1["张三"]
print(f"del后: {d1}")  # {'李四': 88}

# pop：按键删+返回值
d2 = {"张三": 98, "李四": 88}
removed = d2.pop("张三")
print(f"pop返回: {removed}, 字典: {d2}")  # 98, {'李四': 88}

# popitem：删最后一个
d3 = {"张三": 98, "李四": 88}
removed = d3.popitem()
print(f"popitem返回: {removed}, 字典: {d3}")  # ('李四', 88), {'张三': 98}

# clear：清空
d4 = {"张三": 98}
d4.clear()
print(f"clear后: {d4}")  # {{}}
````

## 6.6 字典推导式

> 跟列表推导式一样，一行生成字典。

```python
{表达式键: 表达式值 for 变量 in 可迭代对象 if 条件}
```

### 📖 例题16：字典推导式

````python
# 📖 例题16：字典推导式

# 生成平方字典
d = {i: i*i for i in range(5)}
print(f"平方字典: {d}")  # {{0:0, 1:1, 2:4, 3:9, 4:16}}

# 筛选：只保留偶数的平方
d2 = {i: i*i for i in range(10) if i % 2 == 0}
print(f"偶数平方: {d2}")  # {{0:0, 2:4, 4:16, 6:36, 8:64}}

# 键值交换
d3 = {"张三": 98, "李四": 88}
d4 = {v: k for k, v in d3.items()}
print(f"键值交换: {d4}")  # {{98: '张三', 88: '李四'}}
````

### 📖 例题17：推导式vs for循环

````python
# 📖 例题17：推导式vs for循环

# for循环
d1 = {}
for i in range(5):
    d1[i] = i*i
print(f"for循环: {d1}")

# 字典推导式（一行搞定）
d2 = {i: i*i for i in range(5)}
print(f"推导式: {d2}")

# 结果一样，推导式更简洁
print(f"结果相同: {d1 == d2}")
````

## 6.7 字典嵌套

> 字典里能装字典——表示"分组"关系。

### 📖 例题18：嵌套字典访问

````python
# 📖 例题18：先找学生，再找学科
data = {
    "张三": {"语文": 98, "数学": 88, "英语": 70},
    "李四": {"语文": 85, "数学": 90, "英语": 88}
}
print("张三语文：", data["张三"]["语文"])
print("李四数学：", data["李四"]["数学"])
print("张三物理：", data["张三"].get("物理", "未登记"))
````

### 📖 例题19：列表里装字典

````python
# 📖 例题19：列表里装字典

economic_data = [
    {"国家": "中国", "GDP": 1434290, "人口": 141200},
    {"国家": "美国", "GDP": 2143320, "人口": 33200},
    {"国家": "日本", "GDP": 508000, "人口": 12400}
]

# 遍历
for country in economic_data:
    print(f"{country['国家']}: GDP {country['GDP']}亿美元, 人口{country['人口']}万")

# 筛选：GDP>100万亿的
big = [c for c in economic_data if c['GDP'] > 1000000]
print(f"\nGDP>100万亿: {len(big)}个")
for c in big:
    print(f"  {c['国家']}: {c['GDP']}")
````

### 📖 例题20：遍历嵌套字典算总分

````python
# 📖 例题20：遍历嵌套字典算总分

data = {
    "张三": {"语文": 98, "数学": 88, "英语": 70},
    "李四": {"语文": 85, "数学": 90, "英语": 88}
}

# 算每个学生总分
for name, scores in data.items():
    total = sum(scores.values())
    print(f"{name}: 总分{total}, 平均分{total/len(scores):.1f}")

# 找最高分学生
max_name = None
max_score = 0
for name, scores in data.items():
    total = sum(scores.values())
    if total > max_score:
        max_score = total
        max_name = name
print(f"最高分: {max_name} ({max_score}分)")
````

## 6.8 集合

> **集合 = 不重复、无序的元素集合**——用花括号 `{}`（空集合用 `set()`）。

```python
set1 = {1, 2, 3, 2, 1}    # 重复的自动去重
print(set1)                # {1, 2, 3}
```

> **大白话**：列表能装重复的，集合**只留一份**。用来去重非常方便。

### 📖 例题21：set()去重

````python
# 📖 例题21：从重复名单中保留不同的名字
names = ["Alice", "Bob", "Alice", "Charlie", "Bob"]
print("原始：", names)
unique = set(names)
print("不同名字的个数：", len(unique))  # 3
print("去重结果：", unique)  # 集合中的顺序不固定
````

### 📖 例题22：集合的基本操作

````python
# 📖 例题22：集合的基本操作

# 创建集合（自动去重）
s1 = {1, 2, 3, 4, 4, 4}
print(f"去重: {s1}")  # {{1, 2, 3, 4}}

# 添加
s1.add(5)
print(f"add后: {s1}")

# 删除
s1.remove(5)  # 不存在会报错
print(f"remove后: {s1}")

s1.discard(6)  # 不存在不报错
print(f"discard后: {s1}")

# 常用方法
print(f"长度: {len(s1)}")
print(f"包含2: {2 in s1}")
````

### 集合的常用操作

### 📖 例题23：并集、交集、差集

````python
# 📖 例题23：并集、交集、差集

A = {"张三", "李四", "王五"}
B = {"李四", "王五", "赵六"}

# 并集：两个集合的所有元素
union = A | B
print(f"并集: {union}")  # {{张三, 李四, 王五, 赵六}}

# 交集：两个集合共有的
intersection = A & B
print(f"交集: {intersection}")  # {{李四, 王五}}

# 差集：A中有但B中没有的
diff = A - B
print(f"A-B: {diff}")  # {{张三}}

# 实际应用：找共同的选修课
course_a = {"Python", "统计学", "计量经济学"}
course_b = {"Python", "Python", "Python", "Python", "计量经济学"}
common = course_a & course_b
print(f"共同课程: {common}")
````

| 字典推导式 | 一行生成字典 |
| 字典嵌套 | 字典里装字典 |
| 集合 | 不重复、无序，去重用 |

> 下章我们学**函数**——把重复的代码打包成一个"工具"，调用就行。
