## 一、选择题

### 1. 要根据学号快速找到对应学生的信息，最合适的组织方式是什么？

A. 用学号作字典的键  
B. 把所有字段拼成一段文字  
C. 只保存成绩  
D. 用布尔值保存整张表  

**答案：A。** 字典把键映射到对应的值。

---

### 2. 同一个字典中再次写入已经存在的键，原来的值会怎样？

A. 保留两份同名键  
B. 被新值覆盖  
C. 字典自动变成列表  
D. 所有键都被删除  

**答案：B。** 字典的键唯一，新赋值会更新该键对应的值。

---

### 3. 查询某个键可能不存在，又不想直接报错，适合使用哪种方法？

A. get  
B. append  
C. strip  
D. sort  

**答案：A。** get 可以为缺失键返回默认值。

---

### 4. 统计报名者姓名时只关心不同姓名有多少个，适合先使用哪种容器去重？

A. 集合  
B. 字符串  
C. 布尔值  
D. 浮点数  

**答案：A。** 集合自动保留不重复的元素。

---

### 5. 遍历字典时，希望同时拿到键和对应的值，应使用哪个方法？

A. keys  
B. values  
C. items  
D. count  

**答案：C。** items 提供键和值的配对。

---

## 二、填空题

### 1. 字典中的“查找标识”称为 ________。

**答案：** 键。键用来找到对应的值。

---

### 2. 同一个字典中，键必须保持 ________。

**答案：** 唯一。重复赋值会覆盖旧值。

---

### 3. 键可能不存在时，常用的安全查询方法名是 ________。

**答案：** get。可同时给出缺失时的默认值。

---

### 4. 遍历字典的键和值，可调用 ________ 方法。

**答案：** items。它提供键值对。

---

### 5. 集合会自动去掉 ________ 的元素。

**答案：** 重复。集合适合去重。

## 三、代码题

### 1. 创建字典存3个学生成绩，打印总分。

````python
# 请在下面填写代码：
d = {"张三":98, "李四":88, "王五":70}
print(sum(d.values()))
````

**答案：** 256

---

### 2. 用字典推导式生成 {1:1, 2:4, 3:9}。

````python
# 请在下面填写代码：
d = {i: i**2 for i in range(1, 4)}
print(d)
````

**答案：** {1: 1, 2: 4, 3: 9}

---

### 3. 用 get() 安全访问可能不存在的键，给出默认值0。

````python
# 请在下面填写代码：
d = {"a": 1, "b": 2}
print(d.get("c", 0))
````

**答案：** 0

---

### 4. 创建嵌套字典存2个学生3科成绩，打印"张三"的"数学"。

````python
# 请在下面填写代码：
d = {"张三":{"语文":98,"数学":88}, "李四":{"语文":85,"数学":92}}
print(d["张三"]["数学"])
````

**答案：** 88

---

### 5. 用集合去重，打印去重前后长度。

````python
# 请在下面填写代码：
names = ["Alice","Bob","Alice","Bob"]
print(len(names), len(set(names)))
````

**答案：** 4 2

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1, "a": 2}
print(len(d))
```

````python
# 请在下面填写改正后的代码：
d = {"a": 2, "b": 1}
print(len(d))
````

**改正说明：** 键重复导致 `len(d)` 为1而不是2。键必须唯一。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1}
print(d[0])
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1}
print(d["a"])
````

**改正说明：** 字典不能用数字索引，要用键名。

---

### 3. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1}
print(d.get(0))
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1}
print(d.get("a"))
````

**改正说明：** get() 用键名，不是索引。

---

### 4. 下面的代码运行后报错，请找出问题并改正

```python
set1 = {}
print(set1)
```

````python
# 请在下面填写改正后的代码：
set1 = set()
print(set1)
````

**改正说明：** 空字典 `{}` 不是空集合，空集合要用 `set()`。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
d = {"a": 1}
d["a"] = 2
print(d)
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1, "b": 2}
d["c"] = 3
print(d)
````

**改正说明：** 原代码覆盖同一个键，看不出添加效果。改成添加新键。
