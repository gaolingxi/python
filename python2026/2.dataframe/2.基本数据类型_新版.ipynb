### 整数的4种写法

| 进制 | 前缀 | 例子 | 十进制值 |
|------|------|------|---------|
| 十进制 | 无前缀 | `123` | 123 |
| 八进制 | `0o` 或 `0O` | `0o127` | 87 |
| 十六进制 | `0x` 或 `0X` | `0x12a` | 298 |
| 二进制 | `0b` 或 `0B` | `0b101` | 5 |

> **十进制不能以0开头**（除了0本身）——Python 3里 `0123` 会报错。
**进位规则（记住就懂了）**：

| 进制 | 进位规则 | 每位能用的数字 |
|------|---------|--------------|
| 二进制 | 逢二进一 | 0, 1 |
| 八进制 | 逢八进一 | 0~7 |
| 十六进制 | 逢十六进一 | 0~9, A~F（A=10, B=11, …, F=15）|

> 💡 手算 `0o127` 转十进制：`1×8² + 2×8¹ + 7×8⁰ = 64+16+7 = 87`
> 💡 手算 `0x12a` 转十进制：`1×16² + 2×16¹ + 10×16⁰ = 256+32+10 = 298`

### 整数的位长度

想知道一个整数用二进制表示有多少位？用 `.bit_length()`。

```python
位数 = 整数.bit_length()
```

> **背景**：计算机底层存储用的是二进制，1 字节（Byte）= 8 位（bit）。

**存储单位换算**：

| 单位 | 等于 | 大约能存多少个整数？|
|------|------|------------------|
| 1 字节（B）| 8 位 | 最大 255（8位全1）|
| 1 KB | 1024 B | 2²⁴ = 16,777,216 |
| 1 MB | 1024 KB | 2³² = 4,294,967,296 |
| 1 GB | 1024 MB | 2⁴⁰ = 约 10 万亿 |

> 💡 Python 的整数不会溢出，上面这些只是"用二进制存它需要多少字节"的参考。


### 📖 例题3：255的二进制要占多少位

````python
# 📖 例题3：255的二进制要占多少位

num = 5
print(num.bit_length())   # 5的二进制是101，占3位
num = 255
print(num.bit_length())   # 255的二进制是11111111，占8位
````

### 运算优先级

从高到低：

| 优先级 | 运算符 |
|--------|--------|
| 1 | `**` 幂 |
| 2 | `*` `/` `%` `//` 算术 |
| 3 | `+` `-` 加减 |
| 4 | `<` `<=` `>` `>=` `==` `!=` 比较 |

同优先级从左到右；括号永远最高。

### 📖 例题6：下面每个表达式先算哪一步？

4 个小题分别对应 4 个优先级：


````python
# 📖 例题6：下面每个表达式先算哪一步？

# 优先级1：幂运算最先
print(2**3 * 4)    # 先算 2**3=8，再 8*4=32

# 优先级2：乘除（从左到右）
print(12 / 4 * 2)  # 先 12/4=3，再 3*2=6（不是 12/(4*2)！）

# 优先级3：加减（从左到右）
print(10 - 3 + 5)  # 先 10-3=7，再 7+5=12（不是 10-(3+5)！）

# 优先级4：比较最后（布尔结果）
print(2 + 3 > 4)   # 先 2+3=5，再 5>4=True

# 括号永远最高（覆盖所有优先级）
print(2 + 3 * (4 - 1))  # 先 (4-1)=3，再 3*3=9，再 2+9=11
````

### 浮点数的精度陷阱（重点！）

> 计算机底层是二进制，很多十进制小数（比如0.1、0.2）在二进制里是**无限循环小数**，只能近似存。
>
> 举例：`0.2` 的二进制展开是 `0.00110011001100...`（`0011` 无限循环），
> 计算机只能存有限位，所以存进去的永远是近似值。
>
> 结果：**看似简单的加法，结果不对**。

<details>
<summary>🤔 猜猜看：<code>0.1 + 0.2</code> 等于多少？</summary>

- A. `0.3`
- B. `0.30000000000000004`  ← 正确答案！
- C. `0.29999999999999999`
- D. 报错

</details>

💡 如果需要精确处理金额，还要选择专门的计算方法。现在先记住：小数计算可能出现近似误差。

### 📖 例题14：0.1 + 0.2 等于 0.3 吗

````python
# 📖 例题14：0.1 + 0.2 等于 0.3 吗

print(0.1 + 0.2)   # 0.30000000000000004 不是0.3！
print(0.1 + 0.3)   # 恰好凑巧精确
print(0.3 - 0.1)   # 0.19999999999999996
````
