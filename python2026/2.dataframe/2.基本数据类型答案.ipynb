### 整数的位长度

想知道一个整数用二进制表示有多少位？用 `.bit_length()`。

```python
位数 = 整数.bit_length()
```

> **背景**：计算机底层存储用的是二进制。1字节 = 8位。

### 第一次看到这种写法：`print()` 和 `.bit_length()` 有什么区别？

这里正好第一次同时看到两种很常见的调用方式：

```python
print(1)
num.bit_length()
```

它们看起来都带括号，但写法不一样。

#### 1. `print(1)`：函数名(参数)

```python
print(1)
int("5")
type(5)
```

这种形式可以先理解为：

> **把括号里的数据交给前面的函数处理。**

例如：

```python
print(1)    # 把 1 交给 print()，让它显示在屏幕上
int("5")    # 把 "5" 交给 int()，得到整数 5
```

#### 2. `num.bit_length()`：对象.方法名()

```python
num = 5
num.bit_length()
```

这里先写 `num`，再写点号 `.`，表示：

> **让 num 这个整数使用它自己的 bit_length() 方法。**

所以：

```python
num = 5
print(num.bit_length())    # 5 的二进制是 101，需要 3 位

num = 255
print(num.bit_length())    # 255 的二进制是 11111111，需要 8 位
```

先记住这两种“长相”就可以：

| 写法 | 含义 | 例子 |
|------|------|------|
| `函数名(数据)` | 把数据交给一个函数 | `print(num)`、`int("5")` |
| `对象.方法名()` | 让某个对象执行自己的方法 | `num.bit_length()` |

以后还会不断看到第二种形式，例如字符串的 `.split()`、列表的 `.append()`。

#### 再看一个容易误解的地方：显示结果 ≠ 返回结果

```python
a = print(1)
print(a)
```

运行时会先看到 `1`，然后看到：

```text
None
```

原因是 `print(1)` 的主要任务是**显示 1**，它本身返回的是 `None`。

但下面两个会真正返回结果：

```python
b = int("5")          # b 得到整数 5

num = 5
c = num.bit_length()  # c 得到整数 3
```

所以以后看到一个函数或方法时，可以多问一句：

> **它只是做了一件事，还是还会返回一个结果给我保存？**

这一点后面学习 `.append()`、`.sort()`、`.split()` 时还会再次遇到。


````python
# 📖 看懂两种调用方式

# 1. 函数名(参数)
a = print(1)          # 先显示 1
print(a)              # print() 的返回值是 None

b = int("5")          # int() 返回整数 5
print(b)

# 2. 对象.方法名()
num = 5
c = num.bit_length()
print(c)              # 3，因为 5 的二进制是 101

num = 255
print(num.bit_length())   # 8，因为 255 的二进制是 11111111
````

### 运算优先级

从高到低：

| 优先级 | 运算符 |
|--------|--------|
| 1 | `**` 幂 |
| 2 | `*` `/` `%` `//` 算术 |
| 3 | `+` `-` 加减 |
| 4 | `<` `<=` `>` `>=` `==` `!=` 比较 |

同优先级从左到右；括号永远最高。

### 📖 例题6：下面每个表达式先算哪一步？

4 个小题分别对应 4 个优先级：


````python
# 📖 例题6：下面每个表达式先算哪一步？

# 优先级1：幂运算最先
print(2**3 * 4)    # 先算 2**3=8，再 8*4=32

# 优先级2：乘除（从左到右）
print(12 / 4 * 2)  # 先 12/4=3，再 3*2=6（不是 12/(4*2)！）

# 优先级3：加减（从左到右）
print(10 - 3 + 5)  # 先 10-3=7，再 7+5=12（不是 10-(3+5)！）

# 优先级4：比较最后（布尔结果）
print(2 + 3 > 4)   # 先 2+3=5，再 5>4=True

# 括号永远最高（覆盖所有优先级）
print(2 + 3 * (4 - 1))  # 先 (4-1)=3，再 3*3=9，再 2+9=11
````

### 浮点数的精度陷阱（重点！）

> 计算机底层是二进制，很多十进制小数（比如0.1、0.2）在二进制里是**无限循环小数**，只能近似存。
>
> 举例：`0.2` 的二进制展开是 `0.00110011001100...`（`0011` 无限循环），
> 计算机只能存有限位，所以存进去的永远是近似值。
>
> 结果：**看似简单的加法，结果不对**。

<details>
<summary>🤔 猜猜看：<code>0.1 + 0.2</code> 等于多少？</summary>

- A. `0.3`
- B. `0.30000000000000004`  ← 正确答案！
- C. `0.29999999999999999`
- D. 报错

</details>

💡 这就是为什么做金融计算时，不能用浮点数！用 `Decimal` 或 `round()` 修正。

### 📖 例题14：0.1 + 0.2 等于 0.3 吗

````python
# 📖 例题14：0.1 + 0.2 等于 0.3 吗

print(0.1 + 0.2)   # 0.30000000000000004 不是0.3！
print(0.1 + 0.3)   # 恰好凑巧精确
print(0.3 - 0.1)   # 0.19999999999999996
````
