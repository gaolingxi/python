## 一、选择题

### 1. 下列哪个是Python的合法变量名？

A. `1age`  
B. `my_age`  
C. `if`  
D. `my-age`

**答案：B**  
Python变量名必须以字母或下划线开头，不能以数字开头，不能用关键字（如if），不能含连字符。

### 2. Python中 `=` 符号的作用是？

A. 判断左右两边是否相等  
B. 把右边的值赋给左边的变量  
C. 引入外部模块  
D. 定义函数

**答案：B**  
`=` 是赋值运算符，方向从右到左。判断相等用 `==`。

### 3. 有如下表达式：`result = (4**2 + 45%2*4) / 5 < 456 // 3`。请用一段语句解释运算符优先级如何影响表达式的求值顺序，并用print输出result的值。

````python
# 请在下面填写代码：
result = (4**2 + 45%2*4) / 5 < 456 // 3
print(result)
````

**答案：** `True`  
解释：先算幂 `4**2=16`，再算取模 `45%2=1`，再算乘法 `1*4=4`，再加法 `16+4=20`，再除法 `20/5=4.0`，再算右边 `456//3=152`，最后比较 `4.0 < 152` 得 `True`。

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
1age = 25
print(1age)
```

````python
# 请在下面填写改正后的代码：
age = 25
print(age)
````

**改正说明：** 变量名不能以数字开头，`1age` 改成 `age`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
name = "小明";
age = 20;
print name
```

````python
# 请在下面填写改正后的代码：
name = "小明"
age = 20
print(name)
````

**改正说明：** 两行错误：①Python不需要分号；②`print` 是函数必须加括号 `print(name)`。

---

### 3. 下面的代码运行后报错，请找出问题并改正

```python
if = 10
print(if)
```

````python
# 请在下面填写改正后的代码：
score = 10
print(score)
````

**改正说明：** `if` 是Python关键字，不能作为变量名。

---

### 4. 下面的代码运行后报错，请找出问题并改正

```python
age = input("你的年龄：")
print(age + 1)
```

````python
# 请在下面填写改正后的代码：
age = int(input("你的年龄："))
print(age + 1)
````

**改正说明：** `input()` 返回字符串，`"25" + 1` 会报错。用 `int()` 转成整数后再加。

---

### 5. 下面的代码运行后报错，请找出问题并改正

```python
x = 5
5 = x
print(x)
```

````python
# 请在下面填写改正后的代码：
x = 5
print(x)
````

**改正说明：** 赋值是"把右边的值放进左边的盒子"，数字5不能当盒子，删掉 `5 = x`。
