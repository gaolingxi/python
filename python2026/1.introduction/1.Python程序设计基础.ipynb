### 补充：exe安装包 vs 快捷方式，别搞混了

装软件时经常会遇到两种东西，很多同学搞混了，尤其是搞不清楚"删了快捷方式"是不是等于"卸载了软件"。用刚才装Anaconda的过程来说明：

---

#### 1. 安装包（.exe）

就是下载下来的那个 `Anaconda3-2025.03-Windows-x86_64.exe`。

- **本质**：一个**安装工具**，双击运行它，它会把程序**装进你的系统**
- **作用**：把软件文件解压、配置、注册到系统里
- **大小**：几百MB，装完就没用了，**可以删掉**
- **和软件的关系**：安装包 ≠ 软件本身，安装包只是"送货的卡车"，软件是"货"

> 类比：安装包就像一盒未拆封的乐高，打开后把零件拼好，盒子就可以扔了。

---

#### 2. 快捷方式（.lnk）

安装完成后，桌面或开始菜单里会出现的图标，比如"Anaconda Prompt"、"Jupyter Notebook"。

- **本质**：只是一个**指向程序的链接**（不到1KB）
- **作用**：点它 = 直接打开软件，不用去文件夹里找
- **和软件的关系**：快捷方式 ≠ 软件本身，它只是"贴在桌上的便利贴"

> 类比：快捷方式就像便利贴，写着"乐高拼好了在客厅沙发上"，便利贴本身不是乐高。

---

#### ⚠️ 重点：删快捷方式 ≠ 卸载软件

| 操作 | 结果 |
|------|------|
| 删安装包(.exe) | 不影响软件，软件还装在系统里 |
| 删快捷方式 | **不影响软件**，软件还装在系统里，只是少了个快捷入口 |
| 卸载软件 | 去控制面板卸载，或运行卸载程序，软件才真正删除 |

所以：
- 觉得桌面图标多？删快捷方式没问题，软件还在
- 想彻底删除软件？去控制面板卸载，光删快捷方式没用

---

#### 三者关系

```
下载 → 安装包(.exe) → 双击安装 → 软件装进系统 → 自动创建快捷方式(.lnk)
                ↓                                        ↓
         几百MB的安装工具                          不到1KB的链接
         装完可以删                                删了软件还在
```

### 变量 vs 数学等号（重点！）

| 场景 | 等号含义 | 方向 |
|------|---------|------|
| 数学：x = 5 | 等式，x等于5 | 无方向，两边相等 |
| 编程：x = 5 | 赋值，把5放进x这个盒子 | 从右到左 |

> **易错点**：`x = 5` 之后写 `5 = x` 会报错——因为5是数字，不能被"装进"。

### 命名风格：三种主流

```python
student_id = "202410312"   # 小写+下划线（Python官方推荐）
StudentId = "202410312"    # 驼峰式（Java风格）
STUDENT_ID = "202410312"   # 全大写（表示常量，不要改）
```

> **本课程统一用**：小写 + 下划线，比如 `avg_age`、`edu_years`。

### 命名规则（记住这6条）

| 规则 | 例子 | 说明 |
|------|------|------|
| ✅ 字母/下划线开头 | `age`, `_private`, `avg_age` | 合法 |
| ❌ 数字开头 | `1age` | 报错 |
| ✅ 可以用中文 | `年龄 = 25` | 合法但不推荐 |
| ❌ 不能用关键字 | `if = 5`, `print = 1` | 报错 |
| ⚠️ 避免 `l`, `O`, `I` | `l` 像 `1`，`O` 像 `0` | 易混淆 |
| ⚠️ 有意义命名 | 用 `avg_age` 而不是 `a` | 可读性 |

## 1.6 输入与输出

> Python里，"和电脑对话"就靠两个函数：`input()` 和 `print()`。

### 输出：`print()`

把内容打印到屏幕上。可以打印文字、数字、表达式。

<details>
<summary>🤔 猜猜看：<code>print(10 * 6)</code> 会打印什么？</summary>

- A. `10 * 6`
- B. `60`  ← 正确答案！print 会先算表达式再打印
- C. `106`
- D. 报错

</details>

### 📖 例题5：下面每一行会打印什么

````python
# 📖 例题5：下面每一行会打印什么

print(123)              # 数字
print("Hello")         # 字符串
print(2 + 3)           # 表达式
print(5 > 3)           # 布尔值
````

### 输出不只是变量

`print()` 可以输出任何内容——变量、数字、文字、表达式都行：

```python
int2 = 10
int3 = 6
print(123)             # 数字
print("我是大一新生")   # 文字
print(int2 * int3)     # 表达式（输出计算结果）
print(int2 > int3)     # 比较结果（True/False）
```

### `print()`背后其实就3步

一行 `print("Hello World")` 看着简单，其实 Python 做了 3 件事：

**① 看括号里写了什么** — 直接是文字（带引号）就原样显示，比如 `print("学号：202410312")`。

**② 是要先算的 → 先算出结果** — 数学式先算，比如 `print(10*6)` 会先算成 60 再显示；变量先查值，比如 `print(avg_age)` 会先把 `avg_age` 存的值拿出来。

**③ 显示到屏幕** — 就是你按 Shift+Enter 后看到的那行字。

> **print 就是 3 步：看括号 → 要算就先算 → 显示。**
>
> **常见错误**：`print(Hello World)` 没加引号 → Python 会把 `Hello World` 当成变量去查值，找不到就报错。想打印文字**一定要加引号**。

### 一个坑：input总是返回字符串


> **🎭 先讲个笑话**

**一段聊天记录：**

```text
A：嘿 #是什么意思啊？

A：你再看看记录...
B：看完了.
A：......所以#是啥？

B：没有啊 你想问什么？
```

A 当场和 B 绝交。

多年之后 A 学了 Python，翻回这段聊天记录，盯着这一行看了半天——
发现连自己都看不懂当年到底想问什么。

> 🤔 为什么这段记录看不懂？先记住这个问号，下面讲完**注释**你就懂了。

> **注释 = 写给人看的说明**，Python不会执行注释里的内容。

### 两种注释方式

| 类型 | 写法 | 用途 |
|------|------|------|
| 单行注释 | `# 说明文字` | 解释一行代码 |
| 多行注释 | 三引号包裹 | 代码开头的"说明卡" |

### 📖 例题7：下面两种注释方式有什么不同

````python
# 📖 例题7：下面两种注释方式有什么不同
# 单行注释

'''
多行注释
'''

print("Hello")
````

### 注释的两个作用

1. **给未来的自己看**：三个月后回头看代码，能想起来当时在干嘛
2. **临时禁用代码**：调试时想跳过某行，前面加个 `#` 就行（比删了再贴回方便）

### 单行注释的两种写法

单行注释用 `#` 开头，有两种放法：

| 放法 | 示例 | 说明 |
|------|------|------|
| 单独一行 | `# 这行代码做了什么` | 注释独占一行，下面跟代码 |
| 行尾 | `代码      # 解释` | 代码后面跟注释 |


````python
# 注释放在代码前面（推荐，更清晰）
# 计算平均年龄
avg_age = 45.3
print(avg_age)

# 注释放在代码后面（行尾注释）
avg_age = 45.3      # 平均年龄
print(avg_age)      # 打印结果
````

### 多行注释的常见错误

多行注释用三引号包裹，但有一个坑：

```python
# 错误：三引号和文字放在同一行，会被当成字符串
'''姓名：张三 学号：12345'''    # 这是字符串，不是注释！

# 正确：三引号独占一行
'''
姓名：张三
学号：12345
'''
```

### 多行注释：用三对双引号也可以

多行注释除了 `'''`，也可以用 `"""`，效果一样。常见写法是在代码开头写作业说明卡：

```python
"""
姓名：高泠溪
学号：XXXXXXXX
学院：经济学院数字经济系
提交日期：2024年9月20日
主题：实验一
"""
print("Hello World")
```
