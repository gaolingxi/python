## 一、选择题

### 1. Python中定义函数使用的关键词是？

A. `func`  
B. `def`  
C. `define`  
D. `function`

**答案：B**  
`def` 是 definition 的缩写。

---

### 2. 下列哪个是正确的函数调用？

A. `print()`  
B. `print`  
C. `print:()`  
D. `print {}`

**答案：A**  
函数调用需要加括号 `()`。

---

### 3. 函数中 `return` 的作用是？

A. 打印结果给人看  
B. 把结果交还给调用者  
C. 结束循环  
D. 导入模块

**答案：B**  
`return` 返回值，可保存可再算。`print` 只是给人看。

---

### 4. `lambda x: x * 2` 相当于哪个def函数？

A. `def f(x): x * 2`  
B. `def f(x): return x * 2`  
C. `def f(x): print(x * 2)`  
D. `def f(): x * 2`

**答案：B**  
lambda必须有参数和返回值，等价于 def + return。

---

### 5. 在函数内修改全局变量，需要什么？

A. `local`  
B. `global`  
C. `var`  
D. `scope`

**答案：B**  
`global` 声明变量为全局变量。

## 三、代码题

### 1. 定义函数 add(a, b) 返回两数之和，调用 add(3, 5)。

````python
# 请在下面填写代码：
def add(a, b): return a + b
print(add(3, 5))
````

**答案：** 8

---

### 2. 定义函数 square(x) 返回 x 的平方，调用 square(7)。

````python
# 请在下面填写代码：
def square(x): return x ** 2
print(square(7))
````

**答案：** 49

---

### 3. 定义带默认值的函数 greet(name, greeting="你好")，调用 greet("张三")。

````python
# 请在下面填写代码：
def greet(name, greeting="你好"): return f"{greeting}, {name}!"
print(greet("张三"))
````

**答案：** 你好, 张三!

---

### 4. 用 lambda 定义函数 double(x)，返回 x*2，计算 double(8)。

````python
# 请在下面填写代码：
double = lambda x: x * 2
print(double(8))
````

**答案：** 16

---

### 5. 定义可变参数函数 total(*nums)，返回所有参数之和。

````python
# 请在下面填写代码：
def total(*nums): return sum(nums)
print(total(1, 2, 3, 4, 5))
````

**答案：** 15

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
def add(a, b)
    return a + b
```

````python
# 请在下面填写改正后的代码：
def add(a, b):
    return a + b
````

**改正说明：** def 行尾必须有冒号 `:`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
def calc(x):
    y = x * 2
print(y)
```

````python
# 请在下面填写改正后的代码：
def calc(x):
    return x * 2
print(calc(5))
````

**改正说明：** 函数内变量是局部变量，外面看不到。要用 return 返回。

---

### 3. 下面的代码输出不正确，请找出问题并改正

```python
def f(x):
    print(x * 2)
r = f(5)
print(r)
```

````python
# 请在下面填写改正后的代码：
def f(x): return x * 2
r = f(5)
print(r)
````

### 4. 下面的代码运行后报错，请找出问题并改正

```python
def calc():
    result = 5 * 2
calc()
print(result)
```

````python
# 请在下面填写改正后的代码：
def calc(): return 5 * 2
print(calc())
````

**改正说明：** 函数内变量是局部变量，外面看不到。要用 return 返回值，调用时用返回值。

---

````python
# 请在下面填写改正后的代码：
def multi(a, b, c):
    return a * b * c
print(multi(2, 3, 4))
````

**改正说明：** 函数需要3个参数，调用时只给了2个。
