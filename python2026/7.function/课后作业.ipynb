## 一、选择题

### 1. 计算函数的结果还要参与下一步运算，函数内部应该怎样把结果交给调用者？

A. 使用 return  
B. 只使用 print  
C. 只写注释  
D. 使用 break  

**答案：A。** return 把计算结果交回调用位置。

---

### 2. 同一个问候函数大多数时候用“你好”，偶尔才改成“早上好”，最适合哪种参数设计？

A. 默认参数  
B. 全局变量  
C. 无限循环  
D. 无参数函数  

**答案：A。** 默认参数让常用值可以省略。

---

### 3. 函数需要接收数量不固定的位置参数，适合使用哪种形式？

A. 星号参数  
B. 固定一个参数  
C. 只用全局变量  
D. 直接删除参数  

**答案：A。** 星号参数可以收集多个位置参数。

---

### 4. 函数内部要修改外部已经定义的全局计数器，需要声明什么？

A. global  
B. lambda  
C. break  
D. continue  

**答案：A。** global 声明要修改全局变量。

---

### 5. 只想写一个很短、临时使用的匿名函数，可以选择什么？

A. lambda  
B. class  
C. while  
D. import  

**答案：A。** lambda 用于简单的匿名函数。

---

## 二、填空题

### 1. 定义函数使用的关键字是 ________。

**答案：** def。函数定义要有名称和参数。

---

### 2. 把函数计算结果交还给调用者使用 ________。

**答案：** return。调用者可以保存该结果。

---

### 3. 数量不固定的位置参数通常由 ________ 收集。

**答案：** *args。收集后可逐项处理。

---

### 4. 数量不固定的关键字参数通常由 ________ 收集。

**答案：** **kwargs。收集后可按键访问。

---

### 5. 函数的说明文档通常写在函数体开头的 ________ 中。

**答案：** 文档字符串。它可以描述用途和参数。

---

## 三、代码题

### 1. 定义函数 add(a, b) 返回两数之和，调用 add(3, 5)。

````python
# 请在下面填写代码：
def add(a, b): return a + b
print(add(3, 5))
````

**答案：** 8

---

### 2. 定义函数 square(x) 返回 x 的平方，调用 square(7)。

````python
# 请在下面填写代码：
def square(x): return x ** 2
print(square(7))
````

**答案：** 49

---

### 3. 定义带默认值的函数 greet(name, greeting="你好")，调用 greet("张三")。

````python
# 请在下面填写代码：
def greet(name, greeting="你好"): return f"{greeting}, {name}!"
print(greet("张三"))
````

**答案：** 你好, 张三!

---

### 4. 用 lambda 定义函数 double(x)，返回 x*2，计算 double(8)。

````python
# 请在下面填写代码：
double = lambda x: x * 2
print(double(8))
````

**答案：** 16

---

### 5. 定义可变参数函数 total(*nums)，返回所有参数之和。

````python
# 请在下面填写代码：
def total(*nums): return sum(nums)
print(total(1, 2, 3, 4, 5))
````

**答案：** 15

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
def add(a, b)
    return a + b
```

````python
# 请在下面填写改正后的代码：
def add(a, b):
    return a + b
````

**改正说明：** def 行尾必须有冒号 `:`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
def calc(x):
    y = x * 2
print(y)
```

````python
# 请在下面填写改正后的代码：
def calc(x):
    return x * 2
print(calc(5))
````

**改正说明：** 函数内变量是局部变量，外面看不到。要用 return 返回。

---

### 3. 下面的代码输出不正确，请找出问题并改正

```python
def f(x):
    print(x * 2)
r = f(5)
print(r)
```

````python
# 请在下面填写改正后的代码：
def f(x): return x * 2
r = f(5)
print(r)
````

### 4. 下面的代码运行后报错，请找出问题并改正

```python
def calc():
    result = 5 * 2
calc()
print(result)
```

````python
# 请在下面填写改正后的代码：
def calc(): return 5 * 2
print(calc())
````

**改正说明：** 函数内变量是局部变量，外面看不到。要用 return 返回值，调用时用返回值。

---

````python
# 请在下面填写改正后的代码：
def multi(a, b, c):
    return a * b * c
print(multi(2, 3, 4))
````
