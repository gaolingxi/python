## 场景引入：50年的GDP增长率

> GDP增长率公式：`(本年GDP - 去年GDP) / 去年GDP x 100%`
>
> 算2020年增长率，写一遍。算2021年，又写一遍。**50年呢？**

不用函数：

```python
# 2020年
gdp_2019 = 21
gdp_2020 = 22.6
rate_2020 = (gdp_2020 - gdp_2019) / gdp_2019 * 100
print(f"2020年增长率: {rate_2020:.2f}%")

# 2021年
gdp_2020 = 22.6
gdp_2021 = 25
rate_2021 = (gdp_2021 - gdp_2020) / gdp_2020 * 100
print(f"2021年增长率: {rate_2021:.2f}%")
```

> 公式变了（比如基期调整），50处都要改。**用函数：改一处，处处生效。**

## 7.1 创建函数

> 用 `def` 关键词创建函数。

```python
def 函数名(参数):
    函数体    # 要执行的代码（缩进4格）
```

> 大白话：函数就是一个**打包好的小工具**——给它数据（参数），它返回结果。

### 📖 例题1：用函数复用增长率公式

````python
# 📖 例题1：用函数复用增长率公式

def growth_rate(current, previous):
    """计算增长率。
    参数: current -- 本期值, previous -- 上期值
    返回: 增长率
    """
    if previous == 0:
        print("上期不能为0")
        return None
    return (current - previous) / previous

gdp_2019 = 21
gdp_2020 = 22.6
gdp_2021 = 25

print(f"2020年: {growth_rate(gdp_2020, gdp_2019):.2%}")
print(f"2021年: {growth_rate(gdp_2021, gdp_2020):.2%}")

# 要改公式？只改函数里一行，50处调用全部生效！
````

### 📖 例题2：不用函数要复制50次

````python
# 📖 例题2：不用函数要复制50次
# 2020年的GDP增长率
gdp_2019 = 21
gdp_2020 = 22.6
gdp_growth_rate_2020 = (gdp_2020 - gdp_2019) / gdp_2019
print(f"2020年增长率: {gdp_growth_rate_2020:.2%}")

# 2021年的GDP增长率（复制粘贴一遍）
gdp_2020 = 22.6
gdp_2021 = 25
gdp_growth_rate_2021 = (gdp_2021 - gdp_2020) / gdp_2020
print(f"2021年增长率: {gdp_growth_rate_2021:.2%}")

# 2022年的GDP增长率（又复制一遍...）
# 50年就要复制50遍！改公式要改50处！
# 这就是为什么需要函数：写一次，用多次。
````

## 7.2 文档字符串

> 函数定义后第一行写**三引号注释**（docstring），说明这个函数干什么、参数是什么、返回什么。

```python
    计算两个数的和。

    参数:
```

## 7.3 参数传递

### 形式参数 vs 实际参数

| 概念 | 时机 | 例子 |
|------|------|------|
| 形式参数 | **定义**函数时 | `def add(a, b)` 里的 a, b |
| 实际参数 | **调用**函数时 | `add(1, 2)` 里的 1, 2 |

> 形式参数是"占位符"，实际参数是"真数据"。

### 默认参数

> 参数有默认值时，调用可以不传，用默认值。

```python
def make_tea(tea_base, sugar="半糖", ice="少冰"):
    return f"{tea_base}，{sugar}，{ice}"
```

> 奶茶例子：只喊"来杯四季春"，其他默认。

### 📖 例题6：默认参数让函数更灵活

````python
# 📖 例题6：默认参数让函数更灵活

def make_tea(tea_base, sugar="半糖", ice="少冰"):
    return f"{tea_base}，{sugar}，{ice}"

print(make_tea("四季春"))                       # 四季春，半糖，少冰
print(make_tea("红茶", sugar="无糖"))            # 红茶，无糖，少冰
print(make_tea("红茶", "全糖", "去冰"))          # 红茶，全糖，去冰
````

### 📖 例题7：默认参数的坑

````python
# 📖 例题7：默认参数的坑
# 金融场景：默认利率
def calc_mpc(delta_income, delta_consumption):
    """计算边际消费倾向MPC。"""
    return delta_consumption / delta_income

income_before = 5000
consumption_before = 3000
income_after = 7000
consumption_after = 4500

# 正确：MPC = 消费变化 / 收入变化
delta_income = income_after - income_before        # 2000
delta_consumption = consumption_after - consumption_before  # 1500

mpc_correct = calc_mpc(delta_income, delta_consumption)
print(f"正确MPC: {mpc_correct:.2%}")

# 错误：顺序反了！
mpc_wrong = calc_mpc(delta_consumption, delta_income)
print(f"错误MPC: {mpc_wrong:.2%}")  # 结果完全不同！
````

### 关键字参数

> 用参数名传值，不依赖顺序。

```python
make_tea(tea_base="红茶", sugar="无糖")    # 顺序随便
```

> 跟 `sorted(list, reverse=True)` 一样——reverse=True 就是关键字参数。

### 📖 例题8：关键字参数可以乱序传

````python
# 📖 例题8：关键字参数可以乱序传

def calc_mpc(delta_income, delta_consumption, as_pct=False):
    mpc = delta_consumption / delta_income
    return mpc * 100 if as_pct else mpc

print(calc_mpc(2000, 1500))                  # 0.75
print(calc_mpc(2000, 1500, True))            # 75.0
print(calc_mpc(as_pct=True, delta_income=2000, delta_consumption=1500))  # 75.0

# sorted() 同理：reverse=True 就是关键字参数
scores = [72, 95, 81, 68, 90]
print(sorted(scores, reverse=True))  # [95, 90, 81, 72, 68]
````

### 📖 例题9：关键字参数让代码更清晰

````python
# 📖 例题9：关键字参数让代码更清晰
# 金融场景：指定参数
def calc_mpc(delta_income, delta_consumption, as_pct=False):
    """计算MPC。
    参数: delta_income -- 收入变化, delta_consumption -- 消费变化
    as_pct: 是否显示为百分比
    """
    mpc = delta_consumption / delta_income
    if as_pct:
        return f"{mpc:.2%}"
    return mpc

delta_income = 2000
delta_consumption = 1500

# 位置参数（顺序不能乱）
mpc1 = calc_mpc(delta_income, delta_consumption)
print(f"位置参数: {mpc1}")

# 关键字参数（顺序可以乱）
mpc2 = calc_mpc(delta_consumption=1500, as_pct=True, delta_income=2000)
print(f"关键字参数: {mpc2}")

# 混合：前两个位置参数，第三个关键字
mpc3 = calc_mpc(delta_income, delta_consumption, as_pct=True)
print(f"混合参数: {mpc3}")
````

> 参数数量不固定时用 `*args`（可变数量位置参数）和 `**kwargs`（可变数量关键字参数）。

```python
*args    -> 把多个参数收集成一个元组
```

### 📖 例题10：*args收集任意参数

````python
# 📖 例题10：*args收集任意参数

def show_scores(*scores):
    print(f"成绩列表: {scores}, 共{len(scores)}门")
    print(f"平均分: {sum(scores)/len(scores):.1f}")

show_scores(98, 88, 70)
show_scores(85, 92, 78, 90)

def show_info(**info):
    print(f"信息: {info}")
    for k, v in info.items():
        print(f"  {k}: {v}")

show_info(name="张三", age=20, major="经济学")
````

## 7.4 返回值

> 函数用 `return` 把结果交还给调用者。

```python
def add(a, b):
    return a + b          # 返回值，可以被接收
```

> **print vs return 的区别**：
> - `print()`：给人看，程序拿不到
> - `return`：给程序用，可以保存、再算

### 📖 例题12：**kwargs 收集任意关键字参数

`**kwargs` 把所有多余的关键字参数收进一个字典，
指标数量不同都能传，不用改函数定义。


````python
# 📖 例题12：**kwargs 收集任意关键字参数
# 场景：不同年份的经济指标数量不同，用 **kwargs 灵活传入

def print_econ_data(year, **metrics):
    """打印某年所有经济指标"""
    print(f"【{year}年】")
    for key, value in metrics.items():
        print(f"  {key}：{value}")
    print()

# 传 3 个指标
print_econ_data("2023", GDP="126万亿", CPI=0.2, 失业率=5.2)
# 传 4 个指标（数量不同，函数不用改）
print_econ_data("2024", GDP="130万亿", CPI=0.3, 贸易顺差="2.1万亿", 增速=5.0)
# 对比 *args：位置传需要记住顺序，**kwargs 自带名称更清晰
````

### 📖 例题13：return和print的区别

````python
# 📖 例题13：return和print的区别

def calc_return(x):
    return x * 2

def calc_print(x):
    print(x * 2)

r1 = calc_return(5)
print(r1)                  # 10（能保存）
r2 = calc_print(5)
print(r2)                  # None（print函数返回None）
````

### 📖 例题14：return能保存print不能

````python
# 📖 例题14：return能保存print不能
# 用print的版本（只能显示，不能保存）
def show_result(x):
    print(x * 2)
    return None  # 没写return就是返回None

# 用return的版本（可以保存、可以再用）
def calc_result(x):
    return x * 2

result_show = show_result(5)    # 显示10，但result_show是None
print(f"show_result返回: {result_show}")

result_calc = calc_result(5)    # 不显示，但result_calc是10
print(f"calc_result返回: {result_calc}")
````

### 多返回值

> 函数可以返回多个值，实际是返回一个元组。

### 📖 例题15：返回多个值怎么接收

````python
# 📖 例题15：返回多个值怎么接收

def calc_avg(nums):
    total = sum(nums)
    avg = total / len(nums)
    return total, avg      # 返回元组

total, avg = calc_avg([98, 88, 70])
print(f"总分: {total}, 平均: {avg:.1f}")
````

## 7.5 变量作用域

> **局部变量**：函数内部定义，函数外面看不到。

````python
# 📖 例题17：局部变量外面用不了

def calc_supply(price):
    par1 = -100      # 局部变量
    par2 = 5         # 局部变量
    return par1 + par2 * price

calc_supply(30)      # 函数正常返回
# print(par1)       # NameError! 局部变量外面用不了
# print(par2)       # NameError!
````

### 在函数里修改全局变量

> 用 `global` 关键词声明，就能在函数里修改全局变量。

### 📖 例题19：函数内不加global修改全局变量

````python
# 📖 例题19：函数内不加global修改全局变量

total = 0

def add_to_total(n):
    total = total + n    # 报错！局部变量total没定义

# add_to_total(5)        # UnboundLocalError!
````

### 📖 例题20：变量作用域对比

````python
# 📖 例题20：变量作用域对比
# 加上 global 就能改全局变量
total = 0

def add_to_total(n):
    global total
    total = total + n

add_to_total(5)
add_to_total(10)
print(total)             # 15
````

## 7.6 lambda函数

> **lambda = 匿名函数**——没有名字的小函数，一行搞定。

```python
lambda 参数: 表达式
```

> 大白话：def 是"有名字的函数"，lambda 是"临时函数"——用完就丢。

### 📖 例题21：lambda和def的区别

````python
# 📖 例题21：lambda和def的区别

def supply(price):
    return -100 + 5 * price

supply_l = lambda p: -100 + 5 * p

print(supply(30))          # 50
print(supply_l(30))        # 50（结果一样）
````

### 📖 例题22：lambda可用作参数

````python
# 📖 例题22：lambda可用作参数
# def版本（完整函数，可以写文档、多行逻辑）
def supply_def(price):
    """线性供给函数 Q = -100 + 5P"""
    return -100 + 5 * price

# lambda版本（一行，适合简单表达式）
supply_lambda = lambda price: -100 + 5 * price

print(f"def版本: {supply_def(30)}")
print(f"lambda版本: {supply_lambda(30)}")

# lambda可以用作参数（关键优势！）
prices = [10, 20, 30, 40, 50]
supplies = list(map(supply_lambda, prices))
print(f"批量算供给: {supplies}")
````

### lambda 在推导式中使用

### 第三关：读懂按钮为什么能调用函数

下面这段是理解本题最关键的代码：

```python
def show_seven():
    print("你按了 7")

button = tk.Button(window, text="7", command=show_seven)
```

`command=show_seven` 表示把函数交给按钮；注意**这里不加括号**！如果写成 `command=show_seven()`，创建按钮时就执行函数，而不是点击后执行。

完整计算器把这件事写成：

```python
command=lambda value=key: press(value)
```

意思是把当前按钮上的文字（如 `7`、`+`、`=`）交给 `press()`。这个写法正好复习前面学过的 `lambda`。

````python
"""第七章综合实践：具有计算器外观的 tkinter 图形界面计算器。"""
import tkinter as tk

# 第一部分：四个运算函数，函数章节的学习重点
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return None
    return a / b

def calculate(a, b, op):
    if op == "+":
        return add(a, b)
    if op == "−":
        return subtract(a, b)
    if op == "×":
        return multiply(a, b)
    if op == "÷":
        return divide(a, b)
    return None

def format_number(number):
    if number == 0:
        return "0"
    return f"{number:.12g}"

# 第二部分：老师提供窗口和按钮框架，学生可以更换配色或布局
def create_calculator():
    root = tk.Tk()
    root.title("迷你计算器")
    root.geometry("360x535")
    root.resizable(False, False)
    root.configure(bg="#151A23")

    display = tk.StringVar(value="0")
    history = tk.StringVar(value="")
    state = {"left": None, "operator": None, "new_number": False}

    def reset():
        display.set("0")
        history.set("")
        state.update(left=None, operator=None, new_number=False)

    def show_error():
        display.set("不能除以 0")
        history.set("按 C 重新开始")
        state.update(left=None, operator=None, new_number=True)

    def press(key):
        current = display.get()
        if key == "C":
            reset()
            return

        if key == "DEL":
            if state["new_number"]:
                return
            display.set(current[:-1] if len(current) > 1 else "0")
            if display.get() in ("", "-"):
                display.set("0")
            return

        if key == "±":
            if state["new_number"]:
                display.set("0")
                state["new_number"] = False
            current = display.get()
            if current != "0":
                display.set(current[1:] if current.startswith("-") else "-" + current)
            return

        if key in ("0", "00", "1", "2", "3", "4", "5", "6", "7", "8", "9", "."):
            if state["new_number"] or current == "不能除以 0":
                current = "0"
                display.set("0")
                state["new_number"] = False
                if state["operator"] is None:
                    history.set("")
            if key == ".":
                if "." not in current:
                    display.set(current + ".")
            elif key == "00":
                if current not in ("0", "-0"):
                    display.set(current + "00")
            elif current in ("0", "-0"):
                display.set(("-" if current.startswith("-") else "") + key)
            else:
                display.set(current + key)
            return

        if key in ("+", "−", "×", "÷"):
            if current == "不能除以 0":
                return
            if state["operator"] is not None and not state["new_number"]:
                result = calculate(state["left"], float(current), state["operator"])
                if result is None:
                    show_error()
                    return
                display.set(format_number(result))
            state["left"] = float(display.get())
            state["operator"] = key
            state["new_number"] = True
            history.set(f"{display.get()} {key}")
            return

        if key == "=":
            if state["operator"] is None or state["new_number"]:
                return
            right = float(display.get())
            result = calculate(state["left"], right, state["operator"])
            if result is None:
                show_error()
                return
            history.set(f"{format_number(state['left'])} {state['operator']} {format_number(right)} =")
            display.set(format_number(result))
            state.update(left=None, operator=None, new_number=True)

    # 显示屏：上方显示上一步算式，下方显示当前数字
    screen = tk.Frame(root, bg="#202735", padx=13, pady=11)
    screen.pack(fill="x", padx=14, pady=(16, 9))
    tk.Label(
        screen, textvariable=history, anchor="e", bg="#202735",
        fg="#AAB5C3", font=("Microsoft YaHei UI", 11), height=2
    ).pack(fill="x")
    tk.Entry(
        screen, textvariable=display, state="readonly", readonlybackground="#202735",
        fg="#F8FAFC", justify="right", relief="flat", borderwidth=0,
        font=("Segoe UI", 32, "bold")
    ).pack(fill="x", ipady=10)

    # 按键区：四列五行，浅色工具键 / 深色数字键 / 橙色运算键 / 绿色等号
    pad = tk.Frame(root, bg="#151A23", padx=10, pady=5)
    pad.pack(fill="both", expand=True)
    for col in range(4):
        pad.columnconfigure(col, weight=1, uniform="button")
    for row in range(5):
        pad.rowconfigure(row, weight=1, uniform="button")

    keys = [
        ["C", "DEL", "±", "÷"],
        ["7", "8", "9", "×"],
        ["4", "5", "6", "−"],
        ["1", "2", "3", "+"],
        ["00", "0", ".", "="],
    ]
    for row, line in enumerate(keys):
        for col, key in enumerate(line):
            if key == "=":
                bg, fg, active = "#4FC3A1", "#10221E", "#70D7B8"
            elif key in ("+", "−", "×", "÷"):
                bg, fg, active = "#EF9A4A", "#231509", "#FFBB77"
            elif key in ("C", "DEL", "±"):
                bg, fg, active = "#424D60", "#F8FAFC", "#596981"
            else:
                bg, fg, active = "#2B3444", "#F8FAFC", "#455269"
            tk.Button(
                pad, text=key, command=lambda value=key: press(value),
                font=("Segoe UI", 20, "bold"), bg=bg, fg=fg,
                activebackground=active, activeforeground=fg,
                relief="flat", bd=0, cursor="hand2", takefocus=0
            ).grid(row=row, column=col, sticky="nsew", padx=4, pady=4)

    # 加分功能：也支持用数字键盘、回车、Esc 和 Backspace 操作
    def on_keyboard(event):
        mapped = {
            "/": "÷", "*": "×", "-": "−", "Return": "=",
            "Escape": "C", "BackSpace": "DEL"
        }
        key = mapped.get(event.keysym, event.char)
        if (key in "0123456789" and len(key) == 1) or key in (".", "+", "−", "×", "÷", "=", "C", "DEL"):
            press(key)

    root.bind("<Key>", on_keyboard)
    return root

def main():
    window = create_calculator()
    window.mainloop()

if __name__ == "__main__":
    main()
````
