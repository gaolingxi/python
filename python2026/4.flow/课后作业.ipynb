## 一、选择题

### 1. 下列哪个是正确的if语句写法？

A. `if x > 0`  
B. `if x > 0:`  
C. `if (x > 0)`  
D. `if x > 0 then:`

**答案：B**  
Python的if语句行尾必须有冒号 `:`，不需要括号，也不需要then。

---

### 2. `range(1, 10, 3)` 生成的数列是？

A. 1, 4, 7, 10  
B. 1, 4, 7  
C. 1, 3, 5, 7, 9  
D. 3, 6, 9

**答案：B**  
range(首项, 尾项, 步长) 是含头不含尾。从1开始，步长3，到10之前停：1, 4, 7。

---

### 3. 下列代码的输出是什么？

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

A. 0 1 2 3 4  
B. 0 1 2 3  
C. 0 1 2  
D. 0 1 2 4

**答案：C**  
i=3时执行break跳出循环，3不会被打印。打印0 1 2后循环结束。

---

### 4. 下列代码的输出是什么？

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

A. 1 2 3 4 5  
B. 1 2 4 5  
C. 1 2  
D. 1 2 4

**答案：B**  
i=3时continue跳过本次，3不打印，但循环继续。打印1 2 4 5。

---

### 5. while循环中，如果忘记写更新条件（如 `i = i + 1`），会发生什么？

A. 循环执行一次就停  
B. 循环正常结束  
C. 死循环，程序停不下来  
D. 报错

**答案：C**  
条件永远满足，循环永远不停，这就是"死循环"。

## 三、代码题

### 1. 输入一个成绩，判断是否及格（≥60打印"及格"，否则打印"不及格"）。

````python
# 请在下面填写代码：
score = int(input("请输入成绩："))
print("及格" if score >= 60 else "不及格")
````

**答案：** 输入55 → `不及格`；输入80 → `及格`

---

### 2. 用for循环和range，打印1到10中所有的偶数。

````python
# 请在下面填写代码：
for i in range(2, 11, 2):
    print(i)
````

**答案：** 2 4 6 8 10

---

### 3. 用while循环计算2+4+6+8+10的结果并打印。

````python
# 请在下面填写代码：
total, i = 0, 2
while i <= 10:
    total += i
    i += 2
print(total)
````

**答案：** `30`

---

### 4. 给定收入income=6000，用if-elif-else分级：≥10000"高收入"，≥5000"中等收入"，≥3000"中低收入"，否则"低收入"。

````python
# 请在下面填写代码：
income = 6000
if income >= 10000: print("高收入")
elif income >= 5000: print("中等收入")
else: print("低收入")
````

**答案：** `中等收入`

---

### 5. 用for循环遍历1到10，遇到3的倍数就跳过（用continue），打印其余数字。

````python
# 请在下面填写代码：
for i in range(1, 11):
    if i % 3 == 0: continue
    print(i)
````

**答案：** 1 2 4 5 7 8 10（跳过3、6、9）

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
score = 75
if score >= 60
    print("及格")
```

````python
# 请在下面填写改正后的代码：
score = 75
if score >= 60:
    print("及格")
````

**改正说明：** if语句行尾必须有冒号 `:`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
for i in range(1, 6)
    print(i)
```

````python
# 请在下面填写改正后的代码：
for i in range(1, 6):
    print(i)
````

**改正说明：** for语句行尾也必须有冒号 `:`。

---

### 3. 下面的代码运行后变成死循环，请找出问题并改正

```python
i = 0
while i < 5:
    print(i)
```

````python
# 请在下面填写改正后的代码：
i = 0
while i < 5:
    print(i)
    i = i + 1
````

**改正说明：** 循环体里缺少 `i = i + 1`，条件永远满足，变成死循环。

---

### 4. 下面的代码输出不正确，请找出问题并改正

```python
for i in range(1, 10):
    if i = 5:
        break
    print(i)
```

````python
# 请在下面填写改正后的代码：
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

**改正说明：** 判断相等用 `==`，不是 `=`（赋值）。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
score = 85
if score >= 60:
    print("及格")
elif score >= 80:
    print("良好")
```

````python
# 请在下面填写改正后的代码：
score = 85
if score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
````

**改正说明：** elif从上到下依次判断，85>=60先满足就走了"及格"，不会再看后面的"良好"。应把更大范围的条件放后面。
