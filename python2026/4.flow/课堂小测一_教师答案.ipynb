## 三、代码题（共4题，每题10分，共40分）

### 1. 从键盘输入两个整数，分别存放在变量 `a` 和 `b` 中，打印交换前的两个数；再交换两个变量的值，并打印交换后的结果。（10分）

**示例输入：** `3` 和 `5`。

**预期输出：**

```text
交换前： 3 5
交换后： 5 3
```

两个数的具体输入可以不同，但输出应体现交换前后的位置变化。

````python
# 第三部分 第1题：从键盘输入两个整数并交换
# 样例运行时分别输入 3 和 5
a = int(input("请输入第一个整数："))
b = int(input("请输入第二个整数："))
print("交换前：", a, b)
a, b = b, a
print("交换后：", a, b)
````

### 2. 已知一笔存款的本金为10000元，存款期限为5年，年利率为3.4%。假设按**单利**计算（不考虑复利），请计算并打印利息和到期总额。（10分）

**预期输出：**

```text
利息： 1700.0
到期总额： 11700.0
```

````python
# 第三部分 第2题：单利计算
principal = 10000
annual_rate = 0.034
years = 5
interest = principal * annual_rate * years
total = principal + interest
print("利息：", interest)
print("到期总额：", total)
````

### 3. 分别使用整除运算符 `//` 计算100除以7的商，使用取模运算符 `%` 计算余数，并计算 `2 ** 10`。请将三个结果按顺序各打印一行。（10分）

**预期输出：**

```text
14
2
1024
```

````python
# 第三部分 第3题：整除、取余与幂运算
print(100 // 7)
print(100 % 7)
print(2 ** 10)
````

### 4. 使用 `format()` 生成三行报告。增长率要显示两位小数。（10分）

```text
国家: 中国
GDP(万亿美元): 14.34
增长率: 6.10%
```

````python
# 第三部分 第4题：format() 格式化报告
country = "中国"
gdp = 14.34
growth = 6.1
print("国家: {}".format(country))
print("GDP(万亿美元): {:.2f}".format(gdp))
print("增长率: {:.2f}%".format(growth))
````

### 1. 下面的代码运行时会报错。请找出原因并修改代码，使程序能够比较成绩是否大于80，并按指定文字输出比较结果。（10分）

### 2. 字符串索引纠错（10分）

下面的程序要求**分两行输出**：第一行输出整个字符串 `Hello`，第二行输出最后一个字符 `o`。但原程序在执行第二个 `print()` 时会报错。请修改代码，使它能够正确输出指定结果。

**预期输出：**

```text
Hello
o
```

**待修改的代码：**

```python
s = "Hello"
print(s[0:6])
print(s[5])
```

````python
# 第四部分 第2题：修正越界的字符串索引
s = "Hello"
print(s[0:6])   # 切片终点超出长度不会报错，仍输出Hello
print(s[-1])    # 最后一个字符o；也可以用s[4]
# 错因：Hello的有效正向索引是0到4，s[5]越界。
````
