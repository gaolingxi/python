> 生活里我们天天做选择：**如果下雨就带伞，不下就不带。**

> 程序也一样——根据条件，走不同的路。

### if语句（一个选择）

```python
if 条件:
    要做的事
```

**重点**：
- `if` 后面跟条件，条件后面有**冒号 `:`**
- 下一行**缩进4格**（Tab或4个空格），表示"属于if的代码块"
- 条件为True就执行，False就跳过

### 📖 例题2：会打印"出门"吗

````python
# 📖 例题2：会打印"出门"吗

weather = "下雨"
if weather == "晴天":
    print("出门")  # 不会打印！条件不满足
````

### Python缩进规则（重点！）

| 规则 | 说明 |
|------|------|
| 冒号 `:` | if / while / for 行尾必须有冒号 |
| 缩进4格 | 属于这个代码块的语句，往右退4格 |
| 取消缩进 | 缩进结束 = 代码块结束 |

> 和数学公式不同，Python靠**缩进**区分"谁属于谁"，不是靠大括号 `{}`。

### if-elif-else语句（多选一）

> 如果不只有两种情况呢？比如成绩分等级：优秀、良好、及格、不及格。

```python
if 条件1:
    语句块1       # 条件1为True
elif 条件2:
    语句块2       # 条件1为False，条件2为True
elif 条件3:
    语句块3       # 前两个False，条件3为True
else:
    语句块n       # 以上都不满足
```

<img src="figure/flow3.png" alt="if-elif-else流程图" width="700" height="500">

> **注意**：elif = else if 的缩写。从上到下依次判断，**只要有一个满足，后面的就不看了**。

### 📖 例题4：score=75 会打印什么等级

````python
# 📖 例题4：score=75 会打印什么等级

score = 75
if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好")
elif score >= 60:
    print("及格")  # 走这条！75 >= 60
else:
    print("不及格")
````

## 4.3 嵌套if语句

> 条件里面**再套条件**——先看大条件，满足后再看小条件。

```python
if 条件1:
    if 条件2:
        语句块1    # 条件1和条件2都满足
    else:
        语句块2    # 条件1满足，条件2不满足
else:
    语句块3        # 条件1不满足
```

<img src="figure/flow4.png" alt="嵌套if流程图" width="450" height="500">

> 比如：**先看年龄≥18，再看收入≥2000，两个都满足才能贷款。**

### 📖 例题5：这组数据能贷款吗

````python
# 📖 例题5：这组数据能贷款吗

age = 25
income = 3000
if age >= 18:
    if income >= 2000:
        print("贷款批准")  # 走这条！两个条件都满足
    else:
        print("收入不够")
else:
    print("年龄不够")
````

## 4.4 循环语句

> 有时候同一件事要做很多遍：100户家庭数据，每户都要算人均收入。
>
> 你不可能写100遍 `print(income / size)`，**循环**就是帮你重复做。

Python有两种循环：
1. **while循环** — 条件满足就一直做（不知道做几次）
2. **for循环** — 数着次数做（知道做几次）

### while循环

```python
while 条件:
    循环体    # 条件为True就一直执行
```

> **大白话**：只要条件还满足，就一直转圈做，**直到条件不满足才停**。

### 📖 例题6：这段循环打印几次

````python
# 📖 例题6：这段循环打印几次

num = 0
while num < 3:
    print(f"第{num+1}次")
    num = num + 1
# 打印3次：第1次、第2次、第3次
````

### while循环的3个要素

| 要素 | 作用 | 本例 |
|------|------|------|
| ① 初始值 | 循环开始前的起点 | `num = 0` |
| ② 条件 | 什么时候继续 | `num < 3` |
| ③ 更新 | 每次循环要变，否则死循环 | `num = num + 1` |

> **最容易踩的坑**：忘了写 `num = num + 1`，条件永远满足，变成**死循环**，程序停不下来。

## 4.5 for循环

> for循环**数着次数做**——知道做几次的场景用它最方便。

```python
for 变量 in 范围:
    循环体
```

<img src="figure/loop2.png" alt="for循环流程图" width="220" height="500">

### range函数：生成一串连续整数

for循环最常搭配 `range()` 使用。

| 写法 | 意思 | 生成的数 |
|------|------|----------|
| `range(5)` | 从0到4 | 0, 1, 2, 3, 4 |
| `range(1, 6)` | 从1到5 | 1, 2, 3, 4, 5 |
| `range(1, 10, 2)` | 从1到9，步长2 | 1, 3, 5, 7, 9 |

> **注意**：range是**含头不含尾**——`range(1, 6)` 生成 1~5，不包含6。

### 📖 例题7：这会打印哪些数

````python
# 📖 例题7：这会打印哪些数

for i in range(1, 6):
    print(i)
# 1 2 3 4 5（不含6！）
````

### 📖 例题8：步长为2呢

````python
# 📖 例题8：步长为2呢

for i in range(0, 10, 2):
    print(i)
# 0 2 4 6 8
````

## 4.6 循环嵌套

> 循环里面**再套一个循环**——外循环每走1次，内循环走完一整轮。

最经典的例子：**九九乘法表**。

```
1×1=1
1×2=2  2×2=4
1×3=3  2×3=6  3×3=9
...
```

### 📖 例题9：外循环控制行，内循环控制列

````python
# 📖 例题9：外循环控制行，内循环控制列

for i in range(1, 10):       # i = 行号 1~9
    for j in range(1, i+1):  # j = 列号 1~i
        print(f"{j}×{i}={i*j}", end="  ")
    print()                  # 换行
````

### 循环嵌套的执行过程

| 外循环 i | 内循环 j 范围 | 打印 |
|----------|---------------|------|
| i=1 | j=1 | 1×1=1 |
| i=2 | j=1,2 | 1×2=2 2×2=4 |
| i=3 | j=1,2,3 | 1×3=3 2×3=6 3×3=9 |
| ... | ... | ... |
| i=9 | j=1~9 | 整行 |

> **关键**：外循环走1次，内循环走**一整轮**。9行乘法表，内循环总共执行 1+2+3+...+9 = 45 次。

## 4.7 break和continue

> 有时候循环走到一半就想停，或者想跳过这一次。

### break：直接跳出循环

> **大白话**：跑步计划跑10圈，第3圈碰见朋友，**不跑了，直接走人**。

### 📖 例题10：循环在第几次停

````python
# 📖 例题10：循环在第几次停

for i in range(1, 10):
    if i == 5:
        print("碰到5了，不干了")
        break
    print(i)
# 打印 1 2 3 4，碰到5就break跳出
````

### continue：跳过这次，继续下一次

> **大白话**：跑步计划跑10圈，第3圈碰见朋友，**这圈不跑了，从第4圈继续**。

### 📖 例题11：哪些数被跳过了

````python
# 📖 例题11：哪些数被跳过了

for i in range(1, 6):
    if i == 3:
        continue       # 跳过3，继续下一次
    print(i)
# 打印 1 2 4 5，3被跳过
````

### break vs continue

| 关键字 | 作用 | 比喻 |
|--------|------|------|
| `break` | 跳出整个循环 | 不跑了，走人 |
| `continue` | 跳过这一次，继续下一次 | 这圈不跑，下一圈继续 |
