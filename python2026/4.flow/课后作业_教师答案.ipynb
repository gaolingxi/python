### 1. 用 while 计算1到n的所有偶数之和

直接设置一个正整数变量 n，**使用 while 循环**计算1到n（含n）的所有偶数之和。

**验收样例：** n=1时得到0；n=10时得到30；n=20时得到110。

**要求：** 必须使用循环累加，不得直接写数学求和公式。

```python
# 编程题1：while循环求1到n的偶数和
n = 10
i = 1
total = 0
while i <= n:
    if i % 2 == 0:
        total += i
    i += 1
print("偶数和：", total)
```

### 2. 使用双重 for 循环打印直角三角形

直接设置一个正整数变量 n，使用**两层嵌套的 for 循环**，输出高度为 n 的左对齐直角三角形。

**当 n=5 时，应该得到：**

~~~text
*
**
***
****
*****
~~~

**要求：**
- 外层 for 循环控制行数，内层 for 循环控制当前行的星号数量。
- 第1行打印1个星号，第2行打印2个，依次递增；每行结束后换行。
- 不能直接写出每行图形，也不能用 `print("*" * i)` 代替内层循环。
- 测试 n=3 与 n=5，均应输出正确图形。

```python
# 编程题2：双重for循环输出左对齐直角三角形
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()
```

### 5. 用循环和 if 打印空心正方形

直接设置整数变量 n（n≥3），使用**嵌套循环和 if**打印 n×n 的空心正方形。上下两行全是星号，中间行只有首尾星号。

**n=5 的图形：**

~~~text
*****
*   *
*   *
*   *
*****
~~~

**要求：** 测试 n=3 和 n=5。必须根据行号、列号判断边框位置，不能直接输出固定的图形。

```python
# 编程题5：嵌套for与if输出空心正方形
n = 5
for i in range(1, n + 1):
    for j in range(1, n + 1):
        if i == 1:
            print("*", end="")
        elif i == n:
            print("*", end="")
        elif j == 1:
            print("*", end="")
        elif j == n:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

### 6. 使用 if-elif-else 判断消费优惠档次

直接设置一个非负整数金额变量 amount，依据消费金额输出一个档次：

- 金额达到500元：输出“满500档”

### 7. 使用嵌套 if 判断活动参加资格

活动资格分两道检查：先确认年龄是否达到18岁，**只有年龄达标后**才判断是否已通过培训。

直接设置年龄变量 age 和培训状态变量 training（1表示通过，0表示未通过）：

- 未满18岁，输出“年龄未达标”
- 已满18岁但未通过培训，输出“请先完成培训”
- 已满18岁且培训通过，输出“可以参加活动”

**自测：** (age=16, training=1)、(age=20, training=0)、(age=20, training=1)。

**要求：** 使用嵌套 if，区分内外两层条件的 else，不要只用一个复合条件完成。

### 8. 计数循环：固定执行4次检查

直接设置 `times = 4`，使用 **`for` 和 `range(times)`** 模拟按次检查：依次输出“第1次检查”“第2次检查”等提示；每完成一次，将 `count` 加1，最后输出总检查次数。

**测试要求：**
- `times = 4`：共检查4次；
- `times = 6`：共检查6次；
- `times = 0`：检查0次。

**要求：** 使用 `for`、`range()` 和计数变量，不得手写多条固定的 `print()`。本题是**预先知道要执行多少次的计数循环**。

```python
# 编程题8：固定次数循环与计数
times = 4
count = 0
for i in range(times):
    print("第", i + 1, "次检查")
    count += 1
print("总检查次数：", count)
```

### 9. 条件停止循环：达到存款目标时结束

直接设置初始存款 `savings = 80`、目标金额 `target = 500`、每周增加的存款 `weekly = 70`。使用 **`while` 循环**，只要存款尚未达到目标，就继续每周存钱；达到目标后自动停止。

输出存款所需的周数，以及最终存款金额。

**测试要求：**
- `savings=80, target=500, weekly=70`：需要6周，最终存款500元；
- `savings=80, target=300, weekly=70`：需要4周，最终存款360元；
- `savings=80, target=80, weekly=70`：需要0周，最终存款80元。

**要求：** 不允许写固定次数的 `range(6)`，必须每轮更新存款并检查是否达到目标。本题是**根据条件决定何时停止的循环**。

```python
# 编程题9：while循环达到存款目标后停止
savings = 80
target = 500
weekly = 70
weeks = 0
while savings < target:
    savings += weekly
    weeks += 1
print("所需周数：", weeks)
print("最终存款：", savings)
```

~~~python

    print("办理一位顾客")
~~~

```python
# 纠错题1：修复while循环——每轮减少剩余人数
remaining = 4
while remaining > 0:
    print("办理一位顾客")
    remaining -= 1
```

### 6. 无效记录应该跳过，却提前停止了全部循环

请修正，并验证第2天步数为-1时，程序仍会继续处理第3天（3000步）的记录。下方用循环变量直接生成测试数据，不需要交互输入。

~~~python

    if steps < 0:
        break
    print("有效步数：", steps)
~~~

```python
# 纠错题6：无效记录仅跳过当前轮，不结束全部循环
for day in range(1, 6):
    steps = day * 1000
    if day == 2:
        steps = -1
    if steps < 0:
        continue
    print("有效步数：", steps)
```
