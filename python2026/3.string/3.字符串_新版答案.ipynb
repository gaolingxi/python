## 3.1 字符串的定义与创建

> **字符串 = 一连串字符**。用引号包起来。

### 三种引号

| 引号 | 写法 | 用途 |
|------|------|------|
| 单引号 | `'...'` | 一般使用 |
| 双引号 | `"..."` | 一般使用 |
| 三引号 | `'''...'''` 或 """...""" | 多行字符串 |

```python
s = "他说："你好""    # A
s = "他说：'你好'"    # B
```

<details>
<summary>🤔 猜猜看：下面哪一行是正确的？</summary>

**答案**：B 正确。字符串里包含双引号时，外层要用单引号包裹；A 的双引号嵌套会报 SyntaxError。

</details>

## 3.3 编码与字节

> **字符串**：人看的文字。
> **字节**：机器存的二进制。
>
> 两者之间需要**编码/解码**。

### 常见编码与长度

| 编码 | 说明 |
|------|------|
| ASCII | 用于表示基本英文字母、数字和符号，共 128 个码位 |
| GB2312 | 常见中文字符编码之一 |
| GBK | 在 GB2312 基础上扩展的中文编码 |
| UTF-8 | 广泛使用的 Unicode 编码方式 |

Python 3 的字符串表示文字；需要字节时再用 `encode()`。

这里先认识 `len()`：它返回数据的长度。`len(文字)` 是字符数；`len(文字.encode("utf-8"))` 是编码后的字节数。

### 例题5：一段中文占多少存储空间？

任务：估算一条中文数据的存储大小。先区分“字符数”和“字节数”，再观察 UTF-8 编码后的结果。

### 📖 例题5："电商发展提升家庭消费" 占多少字节

````python
# 📖 例题5：中文有多少个字符、多少个字节？

s = "电商发展提升家庭消费"
b = s.encode("utf-8")
print("字符数：", len(s))
print("UTF-8 字节数：", len(b))
````

> **本题小结**：字符数描述文字有多少个字符，字节数描述编码后占用多少存储空间。

### 乱码是怎么回事？

用错编码打开文件 = 乱码。比如把 UTF-8 编码的 `.txt` 文件用 GBK 打开：

```python
b = "中文".encode('utf-8')      # 用UTF-8编码
print(b.decode('gbk'))          # 用GBK解码，可能乱码
```

### 例题6：同一段数据为什么会乱码？

## 3.4 索引与切片

> 字符串里的每个字符都有位置编号，从 **0** 开始。

### 索引

```
"Hello World"
 0123456789...
```

```python
s = "Hello"
s[0]    # 'H'
s[1]    # 'e'
s[-1]   # 'o'  倒数第1个
```

<details>
<summary>🤔 猜猜看：<code>s[-1]</code> 是什么？<code>s[5]</code> 会怎样？</summary>

**答案**：`s[-1] = 'o'`（最后一个字符）；`s[5]` 会报错！因为 `"Hello"` 只有 5 个字符（索引 0-4），`s[5]` 越界。

</details>

> **注意**：字符串是**不可变的**——不能改单个字符。

### 例题7：怎样取出地址中的某个字符？

任务：从一段文字中定位指定字符。先观察每个字符的位置编号，再引出字符串索引。

### 📖 例题7：s[6] 是什么

````python
# 📖 例题7：s[6] 是什么

s = "Hello World"
print(s[0])  # H
print(s[6])  # W
````

> **本题小结**：字符串索引从 0 开始，也可以使用负数从末尾定位。

### 切片：取一段字符

```python
s[start:end]   # 从 start 到 end 前一位，含左不含右
```

例如 `"Hello"[1:4]` 得到 `"ell"`。

还可以给出第三个位置表示每次移动几位：`s[start:end:step]`。`step` 为 `-1` 时从右向左取；`s[::-1]` 就会把字符串倒过来。

### 📖 例题10：字符串拼接与类型转换

把成绩信息拼成报告句子，注意**数字要先转字符串**：

```python
报告 = "姓名：" + name + "，成绩：" + str(score)
```

⚠️ 直接 `"成绩：" + score`（score 是数字）会报 `TypeError`。


````python
# 📖 例题10：字符串拼接与类型转换
# 场景：把学生的姓名和成绩拼成一句话

name = "张三"
score = 88
level = "良好"

# 字符串 + 字符串：直接拼
report = "姓名：" + name + "，等级：" + level
print(report)  # 姓名：张三，等级：良好

# 数字不能直接拼，先转成字符串
print("成绩：" + str(score))  # 成绩：88

# 完整报告（注意逗号，避免"姓名：张三成绩：88"）
report_full = "姓名：" + name + "，成绩：" + str(score) + "分，等级：" + level
print(report_full)  # 姓名：张三，成绩：88分，等级：良好
# ✗ 下面的会报 TypeError，取消注释试试：
# print("成绩：" + score)
````

> **本题小结**：字符串不能直接和数字相加，数字需要先转换为字符串。

## 3.5 字符串方法：从一条居民信息开始

```python
info = "姓名=张三,性别=男,学历=本科"
```

先观察：三项内容之间都用英文逗号隔开。我们已经学过按位置取文字；如果能先按逗号把它们分开，就能继续按位置取出各段。

### 先想一想：在哪里切开？

`"姓名=张三,性别=男,学历=本科"` 中有两个英文逗号。我们希望分别打印出：

```text
姓名=张三
性别=男
学历=本科
```

按什么符号处理，才能得到这三段文字？

### 引出 `split()`

`split(",")` 按英文逗号处理字符串。配合前面学过的索引 `[0]`、`[1]`、`[2]`，可以直接取出第 1、2、3 段文字。

```python
文字.split(",")[0]  # 第 1 段
文字.split(",")[1]  # 第 2 段
```

这里沿用前面学过的“位置从 0 开始”的规则。

### 📖 例题12：取出居民信息中的三段文字

先猜每行会打印什么，再运行验证。

````python
info = "姓名=张三,性别=男,学历=本科"
print(info.split(",")[0])
print(info.split(",")[1])
print(info.split(",")[2])
````

运行结果：

```text
姓名=张三
性别=男
学历=本科
```

本题只使用已学过的索引，从每一段文字中按位置取值。

### 替换：`replace()`

`replace()` 会在一段文字中找到指定内容，并换成新内容。

```python
新文字 = 原文字.replace(旧内容, 新内容)
```

它返回新字符串，原字符串不变。

### 📖 例题14：统一标点，替换时要选准范围

先统一记录里的逗号，再看看把“大一”改为“大二”时，为什么只替换“大”会出错。

````python
# 📖 例题14：统一标点，替换时要选准范围
info = "姓名=张三，性别=男,学历=本科"
clean = info.replace("，", ",")
print(clean)

status = "我是大一新生"
print(status.replace("大", "大二"))    # 我是大二一新生：范围太小
print(status.replace("大一", "大二"))  # 我是大二新生
````

> **本题小结**：`replace()` 返回替换后的新字符串，原字符串不会被直接修改。

任务：让一组省份名称显示风格一致。先观察大小写转换的结果，再引出 `upper()` 和 `lower()`。

### 📖 例题16：这三个省份的全大写和全小写

````python
# 📖 例题16：这三个省份的全大写和全小写

s = "Zhejiang Hubei Shanghai"
print(s.upper())  # ZHEJIANG HUBEI SHANGHAI
print(s.lower())  # zhejiang hubei shanghai
````

### 查找：`find()` / `rfind()` / `count()`

```python
s.find('子串')      # 从左往右找，返回位置，找不到返回 -1
s.rfind('子串')     # 从右往左找，返回位置，找不到返回 -1
s.count('子串')     # 返回出现次数
```

> 💡 `find` 和 `rfind` 的区别：`find` 返回**第一次**出现的位置，`rfind` 返回**最后一次**出现的位置。

### 例题17：怎样找到文字第一次和最后一次出现的位置？

任务：检查一条户主记录中关键词出现的位置和次数。先观察同一词出现多次的情况，再区分 `find()`、`rfind()` 和 `count()`。

### 📖 例题17："户主"出现了几次？第一次在什么位置

````python
# 📖 例题17：同一个词出现了几次？
s = "户主是女性户主"
print("第一次位置：", s.find("户主"))   # 0
print("最后一次位置：", s.rfind("户主")) # 5
print("出现次数：", s.count("户主"))   # 2
````

> **本题小结**：`find()` 找第一次，`rfind()` 找最后一次，`count()` 统计出现次数。

### 例题18：这笔消费是网购吗？

观察一条记录 `"网购-淘宝-手机壳-39.9"`，再观察 `"报告.txt"`。怎样只看开头或结尾就作出判断？

### 📖 例题18：判断文字的开头与结尾

`startswith()` 判断指定文字是否出现在开头；`endswith()` 判断是否出现在结尾。两者返回 `True` 或 `False`。

````python
# 📖 例题18：判断一条消费记录
record = "网购-淘宝-手机壳-39.9"
print("是否网购：", record.startswith("网购"))

filename = "户主报告.txt"
print("是否文本文件：", filename.endswith(".txt"))
````

> **本题小结**：`startswith()` 判断开头，`endswith()` 判断结尾，返回 True 或 False。

## 3.6 格式化字符串（重点！）

> 想把变量"插进"字符串里，有3种方法。

### 方法1：`%` 格式化（老式）

```python
"格式串 % 变量"
```

**格式字符速查**：

| 符号 | 含义 | 例子 |
|------|------|------|
| `%s` | 字符串 | `"姓名：%s" % "张三"` |
| `%d` | 十进制整数 | `"年龄：%d" % 25` |
| `%f` | 浮点数 | `"成绩：%.2f" % 99.88` |
| `%c` | 单个字符 | `"首字母：%c" % "Z"` |
| `%e` | 科学计数法 | `"值：%.2e" % 12345` |

**宽度与对齐**：`%09d` = 至少9位，不足补0（整数）；`%-10s` = 至少10位，左对齐（字符串）

### 例题19：姓名和成绩怎样写进报告？

先观察姓名、整数成绩和带小数的平均分；同一条报告为什么需要不同的显示格式？

### 📖 例题19：`%s`、`%d`、`%.2f` 分别做什么？

先用单个变量练习每种格式，再观察编号补零、百分号和科学计数法。

````python
# 📖 例题19：为不同数据选择显示格式
name = "张三"
score = 88
average = 88.5
print("姓名：%s" % name)
print("成绩：%d 分" % score)
print("平均分：%.2f" % average)
print("编号：%05d" % 42)
print("CPI 增长率：%.1f%%" % 0.3)
print("GDP 科学计数：%.2e" % 1260000)
````


## 综合挑战：居民报告

已知：

```python
info = "姓名=张三，性别=男,学历=本科"
```

1. 用 `replace()` 统一中英文逗号；
2. 用 `split()` 和已经学过的索引取出三段文字；
3. 再从每一段中取出等号右侧的内容；
4. 用 f-string 输出：`张三，男，学历为本科。`

先自己尝试，再运行下一格的参考解法。

````python
# 综合挑战参考解法
info = "姓名=张三，性别=男,学历=本科"
clean = info.replace("，", ",")

name_field = clean.split(",")[0]
gender_field = clean.split(",")[1]
education_field = clean.split(",")[2]

name = name_field.split("=")[1]
gender = gender_field.split("=")[1]
education = education_field.split("=")[1]
print(f"{name}，{gender}，学历为{education}。")
````
