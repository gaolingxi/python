## 一、选择题

### 1. 下列哪个字符串表示"她说：'你好'"是正确的？

A. `"她说："你好""`  
B. `"她说：'你好'"`  
C. `'她说：'你好''`  
D. `"她说："你好""`

**答案：B**  
字符串里包含单引号时用双引号包裹：`"她说：'你好'"`。

---

### 2. `"Hello"[1:4]` 的结果是？

A. `"Hel"`  
B. `"ell"`  
C. `"Hello"`  
D. `"ellH"`

**答案：B**  
切片 `[1:4]` 取索引1、2、3（不含4），即 `e`、`l`、`l`。

---

### 3. `"a,b,c".split(",")` 的结果是？

A. `'abc'`  
B. `['a', 'b', 'c']`  
C. `['abc']`  
D. `'a' 'b' 'c'`

**答案：B**  
`split(",")` 按逗号分割成列表 `['a', 'b', 'c']`。

---

### 4. `"  你好  ".strip()` 的结果是？

A. `"  你好  "`  
B. `"你好"`  
C. `"你好  "`  
D. `"  你好"`

**答案：B**  
`strip()` 去掉字符串两端的空格。

---

### 5. `f"2+2={2+2}"` 的结果是？

A. `"2+2={2+2}"`  
B. `"2+2=4"`  
C. `"4"`  
D. 报错

**答案：B**  
f-string 里 `{2+2}` 会被计算成 `4`，最终字符串是 `"2+2=4"`。

### 2. 网站会员名不区分大小写，验证上面六个会员名是否唯一。

**提示：** 本题先只使用字符串方法。把 6 个会员名统一转成小写，再用 `|` 标出每个会员名的边界并合并成一个大字符串，最后使用字符串的 `count()` 统计完整会员名出现的次数。

> 暂时不要用列表的 `count()`。第 5 章学习列表时，我们会用列表方法重新解决同一道题。


````python
# 请在下面填写代码：
name1 = "Alice".lower()
name2 = "mr".lower()
name3 = "MrAdam".lower()
name4 = "HGW".lower()
name5 = "ALICE".lower()
name6 = "Alice".lower()

all_names = (
    "|" + name1 + "|" + name2 + "|" + name3 +
    "|" + name4 + "|" + name5 + "|" + name6 + "|"
)

print("alice出现次数：", all_names.count("|alice|"))
print("mr出现次数：", all_names.count("|mr|"))
print("mradam出现次数：", all_names.count("|mradam|"))
print("hgw出现次数：", all_names.count("|hgw|"))
````

**答案：** 
```
'alice' 重复了 2 次
'alice' 重复了 2 次
验证完成
```

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
s = "她说："你好""
print(s)
```

````python
# 请在下面填写改正后的代码：
s = "她说：'你好'"
print(s)
````

**改正说明：** 字符串里包含双引号时，外层要用单引号包裹，或者转义双引号。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
p = "C:\Users\admin"
print(p)
```

````python
# 请在下面填写改正后的代码：
p = r"C:\Users\admin"
print(p)
````

**改正说明：** `\U`、`\a` 是转义字符（换行、响铃），路径要用原始字符串 `r"..."` 或双反斜杠 `\\`。

---

### 3. 下面的代码输出不正确，请找出问题并改正

```python
s = "Hello"
print(s[0:6])
print(s[5])
```

````python
# 请在下面填写改正后的代码：
s = "Hello"
print(s[0:5])  # 取全部
print(s[4])    # 最后一个字符
````

**改正说明：** `"Hello"` 只有5个字符（索引0-4），`s[5]` 越界。切片 `[0:5]` 取全部，`s[4]` 取最后一个。

---

### 4. 下面的代码输出不正确，请找出问题并改正

```python
s = "我爱经济学我爱Python"
print(s.replace("我", "非常", 1))
```

````python
# 请在下面填写改正后的代码：
s = "我爱经济学我爱Python"
# 第一次出现替换成"非常"，其余保持不变
print(s.replace("我", "非常", 1))
````

**改正说明：** `replace` 的第三个参数是替换次数。`replace("我", "非常", 1)` 只替换第一次出现，结果是 `非常爱经济学我爱Python`。如果题目要求全部替换，去掉第三个参数。

---

### 5. 下面的代码运行后报错，请找出问题并改正

```python
name = "张三"

```

````python
# 请在下面填写改正后的代码：
name = "张三"
age = 20
print(f"我叫{name}，我{age}岁")
````

**改正说明：** 这行代码本身没有错误，能正常运行。如果题目是"输出不正确"，可能是变量没赋值或引号用错。正确答案就是这个。
