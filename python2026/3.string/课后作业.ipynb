### 2. 网站会员名不区分大小写，验证上面六个会员名是否唯一。

**提示：** 本题先只使用字符串方法。把 6 个会员名统一转成小写，再用 `|` 标出每个会员名的边界并合并成一个大字符串，最后使用字符串的 `count()` 统计完整会员名出现的次数。

````python
# 请在下面填写代码：
name1 = "Alice".lower()
name2 = "mr".lower()
name3 = "MrAdam".lower()
name4 = "HGW".lower()
name5 = "ALICE".lower()
name6 = "Alice".lower()

all_names = (
    "|" + name1 + "|" + name2 + "|" + name3 +
    "|" + name4 + "|" + name5 + "|" + name6 + "|"
)

print("alice出现次数：", all_names.count("|alice|"))
print("mr出现次数：", all_names.count("|mr|"))
print("mradam出现次数：", all_names.count("|mradam|"))
print("hgw出现次数：", all_names.count("|hgw|"))
````

**答案：** 完整会员名 Alice 忽略大小写后出现 3 次；mr、mradam、hgw 各出现 1 次。

### 4. 使用 `format()` 生成三行报告。增长率要显示两位小数。

```text
国家: 中国
GDP(万亿美元): 14.34
增长率: 6.10%
```

````python
print("国家: {}\nGDP(万亿美元): {}\n增长率: {:.2f}%".format("中国", 14.34, 6.10))
````

### 5. 给定字符串 s = "  户主模拟数据  "，完成以下操作并打印结果：去两端空格、取长度、取前3个字符、取最后3个字符、反转。

````python
s = "  户主模拟数据  ".strip()
print(s)
print(len(s))
print(s[:3])
print(s[-3:])
print(s[::-1])
````

### 7. 给定字符串 s = "户主是女性户主"，用 find() 查找"户主"和"女性"第一次出现的位置，用 count() 统计"户主"出现的次数。

````python
s = "户主是女性户主"
print(s.find("户主"))
print(s.find("女性"))
print(s.count("户主"))
````

**答案：** “户主”第一次出现在位置 0，“女性”出现在位置 3，“户主”共出现 2 次。

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
s = "她说："你好""
print(s)
```

````python
# 请在下面填写改正后的代码：
s = "她说：'你好'"
print(s)
````

**改正说明：** 字符串里包含双引号时，外层要用单引号包裹，或者转义双引号。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
p = "C:\Users\admin"
print(p)
```

````python
# 请在下面填写改正后的代码：
p = r"C:\Users\admin"
print(p)
````

**改正说明：** 普通字符串中的反斜杠会触发转义；这里的 `\U` 会被当成 Unicode 转义开头，`\a` 表示响铃。路径可用原始字符串写法。

---

### 3. 下面的代码会在第二次取字符时报错，请找出问题并改正

```python
s = "Hello"
print(s[0:6])
print(s[5])
```

````python
# 请在下面填写改正后的代码：
s = "Hello"
print(s[0:5])  # 取全部
print(s[4])    # 最后一个字符
````

**改正说明：** `"Hello"` 只有5个字符（索引0-4），`s[5]` 越界。切片 `[0:5]` 取全部，`s[4]` 取最后一个。

---

### 4. 目标是把句子里所有“我”都换成“非常”。下面的代码只替换了第一个，请改正。

```python
s = "我爱经济学我爱Python"
print(s.replace("我", "非常", 1))
```

````python
s = "我爱经济学我爱Python"
print(s.replace("我", "非常"))
````

**改正说明：** `replace` 的第三个参数限制替换次数；要替换全部，就去掉这个参数。

---

### 5. 下面的报告没有填入姓名和年龄，请找出原因并改正

```python
name = "张三"
age = 20
print("我叫{name}，我{age}岁")
```

````python
name = "张三"
age = 20
print(f"我叫{name}，我{age}岁")
````

**改正说明：** 格式化字符串前缺少 `f`；加上后变量值才会填入。
