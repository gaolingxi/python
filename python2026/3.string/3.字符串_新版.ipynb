## 3.3 编码与字节

> **字符串**：人看的文字。
> **字节**：机器存的二进制。
>
> 两者之间需要**编码/解码**。

### 常见编码

| 编码 | 说明 |
|------|------|
| ASCII | 用于表示基本英文字母、数字和符号，共 128 个码位 |
| GB2312 | 常见中文字符编码之一 |
| GBK | 在 GB2312 基础上扩展的中文编码 |
| UTF-8 | 广泛使用的 Unicode 编码方式 |

Python 3 的字符串表示文字；需要字节时再用 `encode()`。

### 问题引入：一段中文占多少存储空间？

任务：估算一条中文数据的存储大小。先区分“字符数”和“字节数”，再观察 UTF-8 编码后的结果。

### 📖 例题5："电商发展提升家庭消费" 占多少字节

````python
# 📖 例题5：中文有多少个字符、多少个字节？

s = "电商发展提升家庭消费"
b = s.encode("utf-8")
print("字符数：", len(s))
print("UTF-8 字节数：", len(b))
````

> **本题小结**：字符数描述文字有多少个字符，字节数描述编码后占用多少存储空间。

## 3.4 索引与切片

> 字符串里的每个字符都有位置编号，从 **0** 开始。

### 索引

```
"Hello World"
 0123456789...
```

```python
s = "Hello"
s[0]    # 'H'
s[1]    # 'e'
s[-1]   # 'o'  倒数第1个
```

<details>
<summary>🤔 猜猜看：<code>s[-1]</code> 是什么？<code>s[5]</code> 会怎样？</summary>

**答案**：`s[-1] = 'o'`（最后一个字符）；`s[5]` 会报错！因为 `"Hello"` 只有 5 个字符（索引 0-4），`s[5]` 越界。

</details>

> **注意**：字符串是**不可变的**——不能改单个字符。

> **本题小结**：字符串索引从 0 开始，也可以使用负数从末尾定位。

### 📖 例题10：字符串拼接与类型转换

把成绩信息拼成报告句子，注意**数字要先转字符串**：

```python
报告 = "姓名：" + name + "，成绩：" + str(score)
```

⚠️ 直接 `"成绩：" + score`（score 是数字）会报 `TypeError`。


````python
# 📖 例题10：字符串拼接与类型转换
# 场景：把学生的姓名和成绩拼成一句话

name = "张三"
score = 88
level = "良好"

# 字符串 + 字符串：直接拼
report = "姓名：" + name + "，等级：" + level
print(report)  # 姓名：张三，等级：良好

# 数字不能直接拼，先转成字符串
print("成绩：" + str(score))  # 成绩：88

# 完整报告（注意逗号，避免"姓名：张三成绩：88"）
report_full = "姓名：" + name + "，成绩：" + str(score) + "分，等级：" + level
print(report_full)  # 姓名：张三，成绩：88分，等级：良好
# ✗ 下面的会报 TypeError，取消注释试试：
# print("成绩：" + score)
````

> **本题小结**：字符串不能直接和数字相加，数字需要先转换为字符串。

`split(",")` 按英文逗号处理字符串。配合前面学过的索引 `[0]`、`[1]`、`[2]`，可以直接取出第 1、2、3 段文字。

```python

```

这里沿用前面学过的“位置从 0 开始”的规则。

运行结果：

````python
# 📖 例题16：这三个省份的全大写和全小写

s = "Zhejiang Hubei Shanghai"
print(s.upper())  # ZHEJIANG HUBEI SHANGHAI
print(s.lower())  # zhejiang hubei shanghai
````

> **本题小结**：`upper()` 转为大写，`lower()` 转为小写。

### 查找：`find()` / `rfind()` / `count()`

```python
s.find('子串')      # 从左往右找，返回位置，找不到返回 -1
s.rfind('子串')     # 从右往左找，返回位置，找不到返回 -1
s.count('子串')     # 返回出现次数
```

> 💡 `find` 和 `rfind` 的区别：`find` 返回**第一次**出现的位置，`rfind` 返回**最后一次**出现的位置。

### 问题引入：怎样找到文字第一次和最后一次出现的位置？

任务：检查一条户主记录中关键词出现的位置和次数。先观察同一词出现多次的情况，再区分 `find()`、`rfind()` 和 `count()`。

### 📖 例题17："户主"出现了几次？第一次在什么位置

````python
# 📖 例题17：同一个词出现了几次？
s = "户主是女性户主"
print("第一次位置：", s.find("户主"))   # 0
print("最后一次位置：", s.rfind("户主")) # 5
print("出现次数：", s.count("户主"))   # 2
````

> **本题小结**：`find()` 找第一次，`rfind()` 找最后一次，`count()` 统计出现次数。

运行结果：

```text
出现次数： 2
```

### 📖 例题18：判断文字的开头与结尾

`startswith()` 判断指定文字是否出现在开头；`endswith()` 判断是否出现在结尾。两者返回 `True` 或 `False`。

````python
# 📖 例题18：只判断一条记录，不需要循环
record = "网购-淘宝-手机壳-39.9"
print("是否网购：", record.startswith("网购"))

filename = "户主报告.txt"
print("是否文本文件：", filename.endswith(".txt"))
````

## 3.6 格式化字符串：把数据填进句子

前面我们学过用 `+` 拼接字符串，也学过用 `str()` 把数字变成文字。

现在再学一种更方便的方法：**先写好一句话，再把变化的数据填进去。**

这一节先理解“填空”的思路，再学习 `format()` 和更简单的 f-string。

### 场景引入：给不同学生生成成绩报告

现在要输出两位学生的成绩：

- 张三，88分
- 李四，95分

如果每来一位学生，就重新写一遍整句话，学生多了会很麻烦。其实，两句话的结构完全一样，变化的只是**姓名和成绩**。

````python
# 每个学生都要重新写一句话
print("张三的成绩是88分")
print("李四的成绩是95分")
````

### 先把句子变成一张“填空题”

我们真正想写的是：

~~~text
______ 的成绩是 ______ 分
~~~

句子中固定的文字不变，只有两个空需要填写。

在 Python 中，可以用一对大括号 `{}` 表示一个空位。这个空位也叫**占位符**。

~~~python
"{}的成绩是{}分"
~~~

先不用记复杂规则，只要把 `{}` 理解成“这里等着放一个数据”就可以。

### 📖 例题19：先填一个空

句子是“欢迎______同学”。

如果要把“张三”填进去，可以这样写：

~~~python
"欢迎{}同学".format("张三")
~~~

- `{}`：句子中等待填写的位置；
- `"张三"`：要填进去的内容；
- `.format()`：负责把内容填进空位。

````python
# 📖 例题19：先填一个空
print("欢迎{}同学".format("张三"))
````

**运行结果：**

~~~text
欢迎张三同学
~~~

### 📖 例题20：再填两个空

现在句子中有两个空：

~~~python
"{}的成绩是{}分".format("张三", 88)
~~~

**填写顺序从左到右：**

| 句子中的位置 | 填进去的数据 |
|---|---|
| 第一个 `{}` | `"张三"` |
| 第二个 `{}` | `88` |

记住：**第一个空对应第一个数据，第二个空对应第二个数据。**

````python
# 📖 例题20：按顺序填写两个空
print("{}的成绩是{}分".format("张三", 88))
````

**运行结果：**

~~~text
张三的成绩是88分
~~~

空位有几个，就按顺序提供几个数据。三个空也是同样的道理：

~~~python
"{}选修了{}，期末成绩是{}分".format("李四", "经济学", 92)
~~~

三个空依次填入：李四、经济学、92。

### 📖 例题21：把具体数据换成变量

刚才我们直接把“张三”和“88”写在 `.format()` 里。

实际编程时，数据通常保存在变量中。现在把具体数据换成变量：

````python
# 📖 例题21：用变量填写报告
name = "张三"
score = 88

report = "{}的成绩是{}分".format(name, score)
print(report)
````

**运行结果：**

~~~text
张三的成绩是88分
~~~

现在只需要修改变量，句子模板不用改：

~~~python
name = "李四"
score = 95
~~~

再次运行同一条 `format()` 语句，就会得到“李四的成绩是95分”。

**这就是格式化字符串的好处：固定句子，替换数据。**

### 🤔 想一想：如果把数据顺序写反了，会怎样？

~~~python
name = "张三"
score = 88

print("{}的成绩是{}分".format(score, name))
~~~

<details>
<summary>点击展开答案</summary>

运行结果是：

~~~text
88的成绩是张三分
~~~

Python 不知道第一个空应该放姓名、第二个空应该放成绩。它只按照顺序填写：第一个数据放第一个空，第二个数据放第二个空。

所以，**数据的顺序必须和空位的顺序对应。**

</details>

### 为什么 `format()` 前面有一个点号？

回忆前面学过的字符串方法：

| 写法 | 作用 |
|---|---|
| `" ABC ".strip()` | 让这段文字去掉两端空格 |
| `"abc".upper()` | 让这段文字变成大写 |
| `"欢迎{}同学".format("张三")` | 让这段文字把空位填上 |

这些写法都有一个共同点：**先写字符串，再用点号调用它的功能。**

所以，`.format()` 不是单独写在前面的函数，而是字符串自带的一个方法。现在先把它理解成“让这句话完成填空”。

### 同一句话，还有更直接的写法：f-string

`format()` 是先写句子，再把数据放在后面。

还有一种写法，可以直接把变量放进句子里：**f-string**。

只需要在字符串最前面加一个字母 `f`，然后把变量名写在大括号中：

~~~python
f"{name}的成绩是{score}分"
~~~

对比一下：

- `"{}的成绩是{}分".format(name, score)`：空位写在句子里，数据写在后面；
- `f"{name}的成绩是{score}分"`：直接把变量名写在句子里。

两种写法都能得到同样的结果。**本课程优先使用 f-string。**

````python
# 同样的报告，两种写法得到相同结果
name = "张三"
score = 88

print("{}的成绩是{}分".format(name, score))
print(f"{name}的成绩是{score}分")
````

### f-string：怎样让小数保留两位？

生成报告时，有时不希望小数显示太多位。

~~~python
average = 45.3219
print(f"平均分：{average:.2f}")
~~~

这里的 `:.2f` 可以先整体记忆为：**显示这个小数时，保留小数点后两位。**

````python
average = 45.3219
print(f"平均分：{average:.2f}")  # 平均分：45.32
````

### 补充：旧式的 `%` 写法（认识即可）

以前的 Python 程序中，也经常用 `%` 把数据放进句子：

~~~python
name = "张三"
score = 88

print("姓名：%s，成绩：%d分" % (name, score))
~~~

这里的 `%s` 用来放字符串，`%d` 用来放整数。

这种写法比较旧，**本课程不要求重点掌握，看到时能够认出来即可。**

````python
# 旧式写法：认识即可
name = "张三"
score = 88
print("姓名：%s，成绩：%d分" % (name, score))
````
