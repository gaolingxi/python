## 3.1 字符串的定义与创建

> **字符串 = 一连串字符**。用引号包起来。

### 三种引号

| 引号 | 写法 | 用途 |
|------|------|------|
| 单引号 | `'...'` | 一般使用 |
| 双引号 | `"..."` | 一般使用 |
| 三引号 | `'''...'''` 或 """...""" | 多行字符串 |

```python
s = "他说："你好""    # A
s = "他说：'你好'"    # B
```

<details>
<summary>🤔 猜猜看：下面哪一行是正确的？</summary>

**答案**：B 正确。字符串里包含双引号时，外层要用单引号包裹；A 的双引号嵌套会报 SyntaxError。

</details>

## 3.3 编码与字节

> **字符串**：人看的文字。
> **字节**：机器存的二进制。
>
> 两者之间需要**编码/解码**。

### 常见编码

| 编码 | 说明 |
|------|------|
| ASCII | 只能表示256个字符，只支持英文 |
| GBK | 中国中文编码 |
| UTF-8 | 全球通用，Python 3默认 |

> **Python 3的字符串是Unicode**，可以表示任何语言的字符。

### 📖 例题5："电商发展提升家庭消费" 占多少字节

````python
# 📖 例题5："电商发展提升家庭消费" 占多少字节

s = "电商发展提升家庭消费"
b = s.encode('utf-8')
print(f"字符数：{len(s)}")
print(f"字节数：{len(b)}")  # UTF-8 中一个中文占3字节
````

### 乱码是怎么回事？

用错编码打开文件 = 乱码。比如把 UTF-8 编码的 `.txt` 文件用 GBK 打开：

```python
b = "中文".encode('utf-8')      # 用UTF-8编码
print(b.decode('gbk'))          # 用GBK解码，可能乱码
```

### 📖 例题10：字符串拼接与类型转换

把成绩信息拼成报告句子，注意**数字要先转字符串**：

```python
报告 = "姓名：" + name + "，成绩：" + str(score)
```

⚠️ 直接 `"成绩：" + score`（score 是数字）会报 `TypeError`。


````python
# 📖 例题10：字符串拼接与类型转换
# 场景：把学生的姓名和成绩拼成一句话

name = "张三"
score = 88
level = "良好"

# 字符串 + 字符串：直接拼
report = "姓名：" + name + "，等级：" + level
print(report)  # 姓名：张三，等级：良好

# 数字不能直接拼，先转成字符串
print("成绩：" + str(score))  # 成绩：88

# 完整报告（注意逗号，避免"姓名：张三成绩：88"）
report_full = "姓名：" + name + "，成绩：" + str(score) + "分，等级：" + level
print(report_full)  # 姓名：张三，成绩：88分，等级：良好
# ✗ 下面的会报 TypeError，取消注释试试：
# print("成绩：" + score)
````

### 📖 例题11：从学号取出生年份，算年级

学号第 7~10 位（索引 `6:10`）是入学年份。
用切片 + `int()` 提取，再算现在是几年级：


````python
# 📖 例题11：从学号取出生年份，算年级
# 场景：学号前6位地区代码，第7-10位入学年份，如"3301002023010101"

student_id = "3301002023010101"

# 取入学年份（索引6到9，共4位）
year_in = int(student_id[6:10])
print(f"入学年份：{year_in}")  # 2023

# 算 2024-2025 学年是几年级（入学当年 = 一年级）
grade = 2024 - year_in + 1
print(f"2024-2025 学年是第 {grade} 年级")  # 第 2 年级

# 取地区代码（前6位）
region = student_id[0:6]
print(f"地区代码：{region}")  # 330100（浙江杭州）
````

### ⭐ 综合例题：网站会员名不区分大小写，怎样检查是否重复？

网站已有 6 个会员名：

```text
Alice、mr、MrAdam、HGW、ALICE、Alice
```

网站规定：**会员名不区分大小写**。

也就是说，下面三个名字在网站看来其实是同一个名字：

```text
Alice
ALICE
alice
```

现在要判断：**这 6 个会员名是否全部唯一？**

这一题很适合把本章几个字符串方法串起来：

```text
lower()  →  统一大小写
   ↓
用 | 分隔  →  明确每个会员名的边界
   ↓
count()  →  统计完整会员名出现几次
```

#### 第一步：分别保存 6 个会员名

```python
name1 = "Alice"
name2 = "mr"
name3 = "MrAdam"
name4 = "HGW"
name5 = "ALICE"
name6 = "Alice"
```

#### 第二步：先统一成小写

因为网站不区分大小写，所以比较之前必须先把大小写统一：

```python
name1 = name1.lower()
name2 = name2.lower()
name3 = name3.lower()
name4 = name4.lower()
name5 = name5.lower()
name6 = name6.lower()
```

现在得到：

```text
alice、mr、mradam、hgw、alice、alice
```

#### 第三步：用 `|` 把它们拼成一个大字符串

```python
all_names = (
    "|" + name1 +
    "|" + name2 +
    "|" + name3 +
    "|" + name4 +
    "|" + name5 +
    "|" + name6 + "|"
)

print(all_names)
```

结果：

```text
|alice|mr|mradam|hgw|alice|alice|
```

这里为什么特意加 `|`？

如果直接在一个大字符串里搜索 `"ann"`，它可能会误匹配到 `"anna"`。写成 `"|ann|"` 后，我们查找的是一个**完整会员名**，而不是会员名中的一部分。

两端也加 `|`，是为了让第一个和最后一个会员名拥有和中间会员名完全一样的边界。

#### 第四步：用字符串的 `count()` 检查

```python
print("alice：", all_names.count("|alice|"))
print("mr：", all_names.count("|mr|"))
print("mradam：", all_names.count("|mradam|"))
print("hgw：", all_names.count("|hgw|"))
```

结果：

```text
alice： 3
mr： 1
mradam： 1
hgw： 1
```

因此：

- 出现 **1 次** → 会员名唯一；
- 出现 **大于 1 次** → 会员名重复；
- `alice` 出现了 **3 次**，所以这 6 个会员名**并不唯一**。

> 这里我们故意只用字符串的思路解决：`lower()`、字符串拼接和 `str.count()`。到第 5 章学习列表后，我们会用 **list 的方法重新解决同一道题**。

#### 一个重要区别：这里为什么暂时不用 `split()`？

`split()` 虽然是字符串方法，但它运行后会产生一个**列表**：

```python
"|alice|mr|".split("|")
```

后面如果继续对拆出的结果写 `.count("alice")`，这里的 `count()` 已经是 **list.count()**，不是字符串的 `str.count()` 了。

所以这一章先保持思路纯粹：

> **把 6 个名字看成一整段字符串，用字符串方法解决。**

第 5 章再把它们真正当成“6 个列表元素”来处理。

````python
# ⭐ 综合例题：只用字符串的方法检查会员名是否重复

name1 = "Alice"
name2 = "mr"
name3 = "MrAdam"
name4 = "HGW"
name5 = "ALICE"
name6 = "Alice"

# 1. 网站不区分大小写：全部统一成小写
name1 = name1.lower()
name2 = name2.lower()
name3 = name3.lower()
name4 = name4.lower()
name5 = name5.lower()
name6 = name6.lower()

# 2. 用 | 标出每个会员名的边界
all_names = (
    "|" + name1 +
    "|" + name2 +
    "|" + name3 +
    "|" + name4 +
    "|" + name5 +
    "|" + name6 + "|"
)

print(all_names)

# 3. 用字符串的 count() 统计完整会员名
print("alice：", all_names.count("|alice|"))    # 3
print("mr：", all_names.count("|mr|"))          # 1
print("mradam：", all_names.count("|mradam|"))  # 1
print("hgw：", all_names.count("|hgw|"))        # 1
````

### 常用方法速查

| 方法 | 作用 | 例子 |
|------|------|------|
| `split(sep)` | 按sep切分 | `"a,b".split(",")` -> `['a','b']` |
| `join(iter)` | 用分隔符合并 | `",".join(['a','b'])` -> `"a,b"` |
| `replace(a,b)` | 替换 | `"a".replace("a","b")` -> `"b"` |
| `strip()` | 去空格 | `" hi ".strip()` -> `"hi"` |
| `upper()` | 大写 | `"abc".upper()` -> `"ABC"` |
| `lower()` | 小写 | `"ABC".lower()` -> `"abc"` |
| `find(sub)` | 查找位置 | `"abc".find("b")` -> `1` |
| `count(sub)` | 出现次数 | `"abcabc".count("abc")` -> `2` |
| `startswith(prefix)` | 是否以某字符开头 | `"abc".startswith("a")` -> `True` |
| `endswith(suffix)` | 是否以某字符结尾 | `"abc".endswith("c")` -> `True` |
| `isdigit()` | 是否全是数字 | `"123".isdigit()` -> `True` |
| `isalpha()` | 是否全是字母 | `"abc".isalpha()` -> `True` |

### 📖 例题18：用 startswith / endswith 判断消费渠道

```python
s.startswith(前缀)   # 是 -> True，不是 -> False
s.endswith(后缀)     # 是 -> True，不是 -> False
```


````python
# 📖 例题18：用 startswith / endswith 判断消费渠道
# 场景：家庭 12 个月消费记录，判断哪些是线上消费

records = [
    "网购-淘宝-手机壳-39.9",
    "线下-超市-牛奶-15.8",
    "网购-京东-教材-89.0",
    "线下-餐馆-聚餐-120.5",
]

# 统计线上消费（以"网购"开头的）
online = [r for r in records if r.startswith("网购")]
print(f"线上消费共 {len(online)} 笔：")
for r in online:
    print(f"  {r}")

# 判断某条记录是否以".5"结尾（正好整数+5毛）
check = "线下-餐馆-聚餐-120.5"
print(f"'{check}' 末尾是 5 毛？", check.endswith(".5"))  # True

# startswith 第二个参数可以指定起始索引（从第几位开始匹配）
# 例：从索引4开始看是否以"淘宝"开头
test = "01-网购-淘宝-手机壳"
print(test.startswith("淘宝", 4))  # True
````

## 3.6 格式化字符串（重点！）

> 想把变量"插进"字符串里，有3种方法。

### 方法1：`%` 格式化（老式）

```python
"格式串 % 变量"
```

### 📖 例题19：%s 和 %d 有什么区别

````python
# 📖 例题19：% 格式化各种写法

name = "张三"
age = 25
gpa = 3.8
print("姓名：%s，年龄：%d" % (name, age))
print("GPA：%.2f" % gpa)          # 3.80
print("编号：%05d" % 42)         # 00042（补0到5位）
print("浮点科学计数：%.2e" % 12345.678)  # 1.23e+04
````

### 方法2：`format()` 方法

```python
"{}".format(变量)     # 或 "第1个{}，第2个{}".format(a, b)
```

### 📖 例题20：format() 和 % 哪个更灵活

````python
# 📖 例题20：format() 和 % 哪个更灵活

name = "高泠溪"
age = 25
print("姓名：{}，年龄：{}".format(name, age))
````

### 方法3：f-string（Python 3.6+，推荐！）

```python
f"变量：{变量}"
```

**直接把变量名写进字符串里，最简洁。**

### 📖 例题21：f-string 能直接写表达式吗

````python
# 📖 例题21：f-string 能直接写表达式吗

name = "高泠溪"
age = 25
edu = 16
print(f"{name}今年{age}岁，学历{edu}年")
````

### 三种方法对比

| 方法 | 写法 | 优点 |
|------|------|------|
| `%` | `"..." % (a, b)` | 老式，稳定 |
| `format()` | `"{}".format(a, b)` | 灵活，可读性好 |
| `f-string` | `f"...{a}..."` | **最简洁，本课程推荐** |
