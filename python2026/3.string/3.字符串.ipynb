>
> 结果发现——有的字段是 `"性别=男,教育=本科"`，有的是 `"性别=男，教育=本科"`。中文逗号vs英文逗号，split时搞混了。

## 3.1 字符串的定义与创建

> **字符串（string）就是一串文字或符号，在 Python 中用 `str` 表示。**

初学时先记住三个重点：

1. **只要被引号包起来，就是字符串。** `"123"` 是字符串，`123` 才是整数。
2. **引号只是告诉 Python“文字从这里开始、到这里结束”**，引号本身通常不是字符串内容的一部分。
3. 字符串里可以放中文、英文、数字、空格和标点，例如 `"杭州"`、`"A001"`、`"2026年9月"`。

```python
name = "张三"
student_id = "20260001"
age = 18

print(type(name))        # <class 'str'>
print(type(student_id))  # <class 'str'>
print(type(age))         # <class 'int'>
```

💡 **为什么学号常常要保存成字符串？**  
因为学号虽然“看起来像数字”，但我们通常不会拿它做加减乘除，而且它还可能以 0 开头。


### 先看懂一行最普通的字符串代码

```python
school = "杭州师范大学"
```

这行代码可以拆成三部分理解：

- `school`：变量名，相当于给数据贴了一个标签；
- `=`：赋值，把右边的数据交给左边的变量；
- `"杭州师范大学"`：字符串字面量，也就是直接写在程序里的文字。

运行后，变量 `school` 保存的是 **杭州师范大学** 这 6 个字符，而不是带双引号的 `"杭州师范大学"`。

```python
print(school)       # 杭州师范大学
print(len(school))  # 6
```

> **容易混淆：** `print("school")` 打印的是单词 `school`；`print(school)` 打印的是变量里保存的内容。


### 三种引号

Python 常见的字符串写法有三种：

| 写法 | 示例 | 初学者怎么理解 |
|------|------|------|
| 单引号 | `'hello'` | 普通字符串 |
| 双引号 | `"hello"` | 普通字符串 |
| 三引号 | `'''...'''` 或 `"""..."""` | 可以直接写多行文字 |

**单引号和双引号没有“等级差别”**，下面两个字符串的值完全相同：

```python
s1 = 'student'
s2 = "student"
print(s1 == s2)    # True
```

选择哪一种，主要看字符串内部有没有引号：

```python
a = '他说："你好"'      # 外面单引号，里面可以直接写双引号
b = "It's Python"      # 外面双引号，里面可以直接写单引号
```

如果内外都使用同一种引号，Python 会误以为字符串提前结束：

```python
# s = "他说："你好""     # SyntaxError
```

三引号适合多行文字：

```python
message = """第一行
第二行
第三行"""
print(message)
```

> 现阶段可以简单记成：**一行文字用单引号或双引号，多行文字可以用三引号。**

## 3.2 转义字符

有时我们希望字符串里出现一些“特殊效果”，例如换行、制表符，或者在双引号字符串里再显示一个双引号。

Python 使用反斜杠 `\` 来表示这类特殊写法，这叫**转义字符**。

```python
print("第一行\n第二行")
```

这里的 `\n` 不是两个普通字符“反斜杠+n”，Python 会把它解释成一个**换行符**。

> 初学者可以把反斜杠理解成一句提示：**“后面的字符不要按普通文字理解。”**


### 常用转义字符

| 写法 | 含义 | 例子 |
|------|------|------|
| `\n` | 换行 | `"A\nB"` |
| `\t` | 制表符（横向对齐） | `"姓名\t成绩"` |
| `\"` | 在双引号字符串中显示双引号 | `"他说：\"你好\""` |
| `\'` | 在单引号字符串中显示单引号 | `'It\'s OK'` |
| `\\` | 显示一个反斜杠 | `"C:\\data"` |
| `\r` | 回车 | 了解即可 |
| `\f` | 换页 | 了解即可 |

最需要掌握的是前 5 个，尤其是 **`\n`、`\t` 和 `\\`**。

### `print()` 和 `repr()` 看起来为什么不一样？

```python
s = "你好\n世界"
print(s)
print(repr(s))
```

`print(s)` 会真正执行换行；`repr(s)` 更像是在告诉我们“字符串内部实际保存了什么”，所以会看到 `\n`。

这在检查空格、换行和制表符时非常有用。

print(s1)

print(s2)

### 原始字符串：`r"..."`

Windows 路径里经常出现反斜杠，例如：

```text
C:\Users\admin\data
```

如果按普通字符串写，某些组合可能被 Python 当成转义字符。可以在字符串前加 `r`，告诉 Python：**里面的反斜杠尽量按普通字符处理。**

```python
path1 = r"C:\Users\admin\data\户主数据.xlsx"
path2 = "C:\\Users\\admin\\data\\户主数据.xlsx"

print(path1)
print(path2)
print(path1 == path2)   # True
```

这种写法叫 **raw string（原始字符串）**。

> 对初学者最常见的用途：**写 Windows 路径。**

⚠️ 一个小限制：原始字符串不能以“单独一个反斜杠”结尾，例如 `r"C:\data\"` 会有语法问题。平时写完整文件路径时通常不会遇到。

## 3.3 编码与字节

这一节解决一个常见问题：**明明文件里是中文，为什么有时打开会乱码？**

先区分两个概念：

- **字符串 `str`**：我们看到和处理的文字，例如 `"杭州"`；
- **字节 `bytes`**：计算机保存到文件、通过网络传输时使用的一串二进制数据。

两者之间需要转换：

```text
字符串 str  -- encode(编码方式) -->  字节 bytes
字节 bytes  -- decode(编码方式) -->  字符串 str
```

例如：

```python
text = "杭州"
data = text.encode("utf-8")

print(text)   # 杭州
print(data)   # b'...' 形式的字节
print(data.decode("utf-8"))  # 杭州
```

> **encode：文字 → 字节；decode：字节 → 文字。**


### 常见编码

| 编码 | 初学者需要知道什么 |
|------|------|
| ASCII | 最早期的英文字符编码，标准 ASCII 一共 128 个字符 |
| GBK | 中国常见的中文编码，老文件中比较常见 |
| UTF-8 | 目前非常常用的 Unicode 编码，能表示世界各地文字 |

Python 3 的 `str` 使用 Unicode 来表示字符，因此可以直接写中文、日文、英文等。

```python
s = "Python中文日本語"
print(s)
```

但要注意：

> **“Python 的字符串是 Unicode”不等于“所有文件默认都是 UTF-8”。**

读取外部文本文件时，仍然要知道文件本身采用什么编码。

另外，**字符数和字节数不是一回事**：

```python
s = "A中"
print(len(s))                 # 2 个字符
print(len(s.encode("utf-8"))) # 4 个字节：A 1字节 + 中 3字节
```

### 乱码是怎么回事？

乱码的本质通常是：

> **写入时用一种编码，读取时却按另一种编码解释。**

可以把编码方式想成“密码本”。同一串字节，用错密码本，就可能得到奇怪字符，甚至直接报错。

```python
text = "中文"
data = text.encode("utf-8")      # 按 UTF-8 变成字节

print(data.decode("utf-8"))     # 正确：中文
```

如果改成错误的编码：

```python
# print(data.decode("gbk"))
```

有些内容会显示乱码，有些内容会直接产生 `UnicodeDecodeError`。这两种情况都说明：**编码和解码方式没有匹配。**

💡 以后读取文本文件时经常会看到：

```python
open("data.txt", encoding="utf-8")
```

这里的 `encoding="utf-8"` 就是在明确告诉 Python 应该使用哪一种编码。

## 3.4 索引与切片

字符串是一串有顺序的字符，所以我们可以按“位置”取字符。

Python 的位置编号叫**索引（index）**，并且从 **0** 开始，而不是从 1 开始。

```text
字符： H  e  l  l  o
正索引 0  1  2  3  4
负索引-5 -4 -3 -2 -1
```

因此：

- `s[0]`：第一个字符；
- `s[1]`：第二个字符；
- `s[-1]`：最后一个字符；
- `s[-2]`：倒数第二个字符。

> **正索引适合从左边数，负索引适合从右边数。**


### 索引：一次取一个字符

```python
s = "Hello"

print(s[0])   # H
print(s[1])   # e
print(s[-1])  # o
```

字符串长度是 5，所以合法的正索引只有 `0, 1, 2, 3, 4`。

```python
# print(s[5])   # IndexError: string index out of range
```

> **为什么 `s[5]` 不行？**  
> 因为“第 5 个字符”的索引其实是 4；索引从 0 开始。

### 字符串不能直接修改某一个位置

```python
s = "cat"
# s[0] = "b"     # TypeError
```

字符串是**不可变对象（immutable）**。如果要得到 `"bat"`，通常是生成一个新的字符串，而不是在原字符串上直接改。


### 📖 例题7：s[6] 是什么

````python
# 📖 例题7：s[6] 是什么

s = "Hello World"
print(s[0])  # H
print(s[6])  # W
````

### 切片：一次取一段字符

索引一次只能取一个字符，切片可以取一段：

```python
s[start:stop]
```

规则是：

> **从 `start` 开始取，一直取到 `stop` 前一个位置。左边包含，右边不包含。**

例如：

```python
s = "Hello"
print(s[1:4])   # ell
```

实际取的是索引 1、2、3。

为什么右边不包含？因为这样长度很好算：

```text
s[1:4] 的长度 = 4 - 1 = 3
```

常见省略写法：

```python
s[:3]    # 从开头取到索引2
s[2:]    # 从索引2取到结尾
s[:]     # 整个字符串
s[-3:]   # 最后3个字符
```

完整切片还可以写成：

```python
s[start:stop:step]
```

`step` 表示“每隔几步取一次”：

```python
s = "abcdef"
print(s[::2])   # ace
print(s[::-1])  # fedcba，字符串反转
```

💡 **索引越界会报错，但切片超出范围通常不会报错。**

```python
print("abc"[:100])   # abc
```

把成绩信息拼成报告句子，注意**数字要先转字符串**：

````python
# 📖 例题10：字符串拼接与类型转换
# 场景：把学生的姓名和成绩拼成一句话

name = "张三"
score = 88
level = "良好"

# 字符串 + 字符串：直接拼
report = "姓名：" + name + "，等级：" + level
print(report)  # 姓名：张三，等级：良好

# 数字不能直接拼，先转成字符串
print("成绩：" + str(score))  # 成绩：88

# 完整报告（注意逗号，避免"姓名：张三成绩：88"）
report_full = "姓名：" + name + "，成绩：" + str(score) + "分，等级：" + level
print(report_full)  # 姓名：张三，成绩：88分，等级：良好
# ✗ 下面的会报 TypeError，取消注释试试：
# print("成绩：" + score)
````

## 3.5 字符串方法（重点！）

字符串后面经常会出现一个点 `.`，例如：

```python
s.strip()
s.split(",")
s.replace("旧", "新")
```

这种写法叫调用字符串的**方法（method）**。

可以先把它理解成：

> **“让这个字符串执行一个专门的操作。”**

非常重要的一点：

> **大多数字符串方法不会直接修改原字符串，而是返回一个新的字符串或其他结果。**

```python
s = " hello "
s.strip()

print(repr(s))          # ' hello '，原来的 s 没变

new_s = s.strip()
print(repr(new_s))      # 'hello'
```

所以初学时经常要写：

```python
s = s.strip()
```

也就是把处理后的新结果重新赋值给 `s`。


### 先把两种常见写法彻底看懂：`函数(...)` 和 `对象.方法(...)`

前面已经见过：

```python
print(s)
len(s)
```

现在又开始大量出现：

```python
s.strip()
s.lower()
s.split(",")
```

初学者很容易问：

> **为什么有时直接写 `len(...)`，有时却要先写 `s.`？**

Python 里经常会看到两种调用形式。

#### 第一种：`函数名(参数)`

例如：

```python
print(s)
len(s)
int("123")
```

可以先这样理解：

> **把括号里的数据交给一个独立的函数处理。**

例如：

```python
s = "Python"

print(s)      # 把 s 交给 print()，显示出来
n = len(s)    # 把 s 交给 len()，得到长度 6
```

这里：

- `print`、`len` 是函数名；
- `()` 表示“现在调用它”；
- 括号里的 `s` 是传给函数的数据，也叫**参数**。

#### 第二种：`对象.方法名(参数)`

例如：

```python
s.lower()
s.strip()
s.split(",")
```

可以读成：

> **让前面的这个对象，使用它自己拥有的某个方法。**

例如：

```python
s = "  Python  "

s.strip()     # 让字符串 s 使用 strip 方法
s.lower()     # 让字符串 s 使用 lower 方法
```

这里：

- `s` 是对象；
- `.` 可以先理解为“的”；
- `strip` 是这个字符串对象拥有的方法；
- `()` 仍然表示“调用这个方法”。

所以：

```python
s.lower()
```

可以很自然地读成：

> **s 的 lower 方法。**

#### 为什么不是所有功能都写成同一种形式？

因为有些功能是比较通用的函数：

```python
len(s)
print(s)
```

而有些功能属于某一类对象本身：

```python
s.lower()
s.split(",")
```

例如“转小写”是字符串才有意义的操作，所以写成字符串的方法：

```python
"HELLO".lower()
```

整数就没有这个方法：

```python
# 123.lower()   # 会报错
```

以后学习列表时也会看到：

```python
scores.append(90)
scores.count(80)
```

这仍然是同一种 **对象.方法()** 的写法。

### 括号里有没有内容，表示什么？

方法后面的括号可能是空的：

```python
s.lower()
s.strip()
```

因为这些操作不需要你再提供额外信息。

有些方法需要额外信息：

```python
s.split(",")
s.replace("旧", "新")
```

这里的 `","`、`"旧"`、`"新"` 就是传给方法的参数。

### 一行里同时出现函数和方法，怎么读？

看这一行：

```python
print(len(s.strip()))
```

不要从左往右硬读，**从最里面开始**：

1. `s.strip()`：先去掉字符串两端空格；
2. `len(...)`：再计算处理后字符串的长度；
3. `print(...)`：最后把长度显示出来。

例如：

```python
s = "  Python  "
print(len(s.strip()))
```

运行结果：

```text
6
```

可以把它想成一层一层套盒子：

```text
s.strip()
   ↓
"Python"
   ↓
len("Python")
   ↓
6
   ↓
print(6)
```

> **看到多层括号时，优先从最里面开始计算。**

### 最后再记一个很重要的区别：调用不等于打印

```python
s.lower()
```

这行代码会得到一个新的小写字符串，但如果没有 `print()`，你不一定会把它作为正式输出显示出来，也没有把结果保存下来。

更常见的写法是：

```python
new_s = s.lower()
print(new_s)
```

所以看到一个函数或方法时，可以连续问三个问题：

1. **谁在执行？** 是独立函数，还是某个对象的方法？
2. **括号里给了什么参数？**
3. **它执行后返回什么结果？这个结果有没有保存？**


````python
# 📖 看懂：函数调用 vs 对象的方法调用

s = "  Python  "

# 1. 函数名(参数)
print(s)
print(len(s))

# 2. 对象.方法名()
clean = s.strip()
lower = s.lower()

print(clean)
print(lower)

# 3. 一行中嵌套调用：从最里面往外算
print(len(s.strip()))   # 6
````

### 分割：`split()`

`split()` 的作用可以记成一个字：

> **拆。**

它把一个字符串按照指定分隔符拆成若干部分，结果是一个列表。

```python
text = "张三,李四,王五"
names = text.split(",")

print(names)   # ['张三', '李四', '王五']
```

这里最容易出错的是：**分隔符必须和原文字完全一致。**

```python
"男,本科".split(",")   # ['男', '本科']
"男，本科".split(",")  # ['男，本科']，没有拆开
```

因为中文逗号 `，` 和英文逗号 `,` 是两个不同字符。

#### 不写分隔符时

```python
"Python   is  easy".split()
# ['Python', 'is', 'easy']
```

不写参数时，`split()` 会按连续的空白字符切分。

#### 第二个参数 `maxsplit`

```python
"a,b,c,d".split(",", 2)
# ['a', 'b', 'c,d']
```

`2` 表示“**最多切 2 次**”，不是“最后得到 2 个元素”。

> **合。**

```python
names = ["张三", "李四", "王五"]
text = "、".join(names)

print(text)   # 张三、李四、王五
```

`join()` 的写法第一次看通常有点反直觉：

```python
"分隔符".join(一组字符串)
```

可以把 `"、".join(names)` 读成：

> “用顿号，把 names 里的内容连接起来。”

⚠️ `join()` 要连接的元素必须是字符串：

```python
numbers = [1, 2, 3]
# ",".join(numbers)     # TypeError
```

现阶段只要记住：**split 把一个字符串拆开；join 把多个字符串合起来。**

### ⭐ 综合例题：网站会员名不区分大小写，怎样检查是否重复？

网站已有 6 个会员名：

```text
Alice、mr、MrAdam、HGW、ALICE、Alice
```

网站规定：**会员名不区分大小写**。

也就是说，下面三个名字在网站看来其实是同一个名字：

```text
Alice
ALICE
alice
```

现在要判断：**这 6 个会员名是否全部唯一？**

这一题很适合把本章几个字符串方法串起来：

```text
lower()  →  统一大小写
   ↓
用 | 分隔  →  明确每个会员名的边界
   ↓
count()  →  统计完整会员名出现几次
```

#### 第一步：分别保存 6 个会员名

```python
name1 = "Alice"
name2 = "mr"
name3 = "MrAdam"
name4 = "HGW"
name5 = "ALICE"
name6 = "Alice"
```

#### 第二步：先统一成小写

因为网站不区分大小写，所以比较之前必须先把大小写统一：

```python
name1 = name1.lower()
name2 = name2.lower()
name3 = name3.lower()
name4 = name4.lower()
name5 = name5.lower()
name6 = name6.lower()
```

现在得到：

```text
alice、mr、mradam、hgw、alice、alice
```

#### 第三步：用 `|` 把它们拼成一个大字符串

```python
all_names = (
    "|" + name1 +
    "|" + name2 +
    "|" + name3 +
    "|" + name4 +
    "|" + name5 +
    "|" + name6 + "|"
)

print(all_names)
```

结果：

```text
|alice|mr|mradam|hgw|alice|alice|
```

这里为什么特意加 `|`？

如果直接在一个大字符串里搜索 `"ann"`，它可能会误匹配到 `"anna"`。写成 `"|ann|"` 后，我们查找的是一个**完整会员名**，而不是会员名中的一部分。

两端也加 `|`，是为了让第一个和最后一个会员名拥有和中间会员名完全一样的边界。

#### 第四步：用字符串的 `count()` 检查

```python
print("alice：", all_names.count("|alice|"))
print("mr：", all_names.count("|mr|"))
print("mradam：", all_names.count("|mradam|"))
print("hgw：", all_names.count("|hgw|"))
```

结果：

```text
alice： 3
mr： 1
mradam： 1
hgw： 1
```

因此：

- 出现 **1 次** → 会员名唯一；
- 出现 **大于 1 次** → 会员名重复；
- `alice` 出现了 **3 次**，所以这 6 个会员名**并不唯一**。

> 这里我们故意只用字符串的思路解决：`lower()`、字符串拼接和 `str.count()`。到第 5 章学习列表后，我们会用 **list 的方法重新解决同一道题**。

#### 一个重要区别：这里为什么暂时不用 `split()`？

`split()` 虽然是字符串方法，但它运行后会产生一个**列表**：

```python
"|alice|mr|".split("|")
```

后面如果继续对拆出的结果写 `.count("alice")`，这里的 `count()` 已经是 **list.count()**，不是字符串的 `str.count()` 了。

所以这一章先保持思路纯粹：

> **把 6 个名字看成一整段字符串，用字符串方法解决。**

第 5 章再把它们真正当成“6 个列表元素”来处理。

````python
# ⭐ 综合例题：只用字符串的方法检查会员名是否重复

name1 = "Alice"
name2 = "mr"
name3 = "MrAdam"
name4 = "HGW"
name5 = "ALICE"
name6 = "Alice"

# 1. 网站不区分大小写：全部统一成小写
name1 = name1.lower()
name2 = name2.lower()
name3 = name3.lower()
name4 = name4.lower()
name5 = name5.lower()
name6 = name6.lower()

# 2. 用 | 标出每个会员名的边界
all_names = (
    "|" + name1 +
    "|" + name2 +
    "|" + name3 +
    "|" + name4 +
    "|" + name5 +
    "|" + name6 + "|"
)

print(all_names)

# 3. 用字符串的 count() 统计完整会员名
print("alice：", all_names.count("|alice|"))    # 3
print("mr：", all_names.count("|mr|"))          # 1
print("mradam：", all_names.count("|mradam|"))  # 1
print("hgw：", all_names.count("|hgw|"))        # 1
````

### 常用方法速查

| 方法 | 作用 | 例子 |
|------|------|------|
| `split(sep)` | 按sep切分 | `"a,b".split(",")` -> `['a','b']` |
| `join(iter)` | 用分隔符合并 | `",".join(['a','b'])` -> `"a,b"` |
| `replace(a,b)` | 替换 | `"a".replace("a","b")` -> `"b"` |
| `strip()` | 去空格 | `" hi ".strip()` -> `"hi"` |
| `upper()` | 大写 | `"abc".upper()` -> `"ABC"` |
| `lower()` | 小写 | `"ABC".lower()` -> `"abc"` |
| `find(sub)` | 查找位置 | `"abc".find("b")` -> `1` |
| `count(sub)` | 出现次数 | `"abcabc".count("abc")` -> `2` |
| `startswith(prefix)` | 是否以某字符开头 | `"abc".startswith("a")` -> `True` |
| `endswith(suffix)` | 是否以某字符结尾 | `"abc".endswith("c")` -> `True` |
| `isdigit()` | 是否全是数字 | `"123".isdigit()` -> `True` |
| `isalpha()` | 是否全是字母 | `"abc".isalpha()` -> `True` |

### 📖 例题18：用 `startswith()` / `endswith()` 判断开头和结尾

这两个方法返回布尔值 `True` 或 `False`：

```python
s.startswith("前缀")
s.endswith("后缀")
```

它们特别适合判断：

- 某条记录是不是以固定标签开头；
- 文件名是不是某种扩展名；
- 编号是不是某个地区或年份开头。


````python
# 📖 例题18：startswith / endswith

record = "网购-淘宝-手机壳-39.9"
print(record.startswith("网购"))    # True
print(record.startswith("线下"))    # False

filename = "household_data.csv"
print(filename.endswith(".csv"))    # True

student_id = "20260001"
print(student_id.startswith("2026")) # True

# 也可以从指定索引开始检查
test = "01-网购-淘宝-手机壳"
print(test.startswith("淘宝", 6))    # True
````

## 3.6 格式化字符串（重点！）

格式化字符串要解决的问题是：

> **一句话里既有固定文字，又有会变化的变量，怎么方便地拼在一起？**

例如姓名和成绩会变化：

```python
name = "张三"
score = 88
```

我们希望得到：

```text
张三的成绩是88分
```

当然可以用 `+` 拼接，但数字还要先 `str()` 转换，变量多了以后很难读。

Python 常见有三种写法：

1. `%`：老式写法，能看懂即可；
2. `.format()`：较新的写法；
3. **f-string：最直观，本课程重点掌握。**


### 方法1：`%` 格式化（老式）

```python
"格式串 % 变量"
```

**格式字符速查**：

| 符号 | 含义 | 例子 |
|------|------|------|
| `%s` | 字符串 | `"姓名：%s" % "张三"` |
| `%d` | 十进制整数 | `"年龄：%d" % 25` |
| `%f` | 浮点数 | `"成绩：%.2f" % 99.88` |
| `%c` | 单个字符 | `"首字母：%c" % "Z"` |
| `%e` | 科学计数法 | `"值：%.2e" % 12345` |

**宽度与对齐**：`%09d` = 至少9位，不足补0（整数）；`%-10s` = 至少10位，左对齐（字符串）

### 📖 例题19：%s 和 %d 有什么区别

````python
# 📖 例题19：% 格式化各种写法

name, total, gpa = "张三", 256, 3.53

print("%s 总分：%d，GPA：%.2f" % (name, total, gpa))
print("编号：%05d" % 42)             # 00042
print("CPI 增长率：%.1f%%" % 0.3)    # 0.3%
print("GDP 科学计数：%.2e" % 1260000)
````

````python
# 输出：

张三 总分：256.0，GPA：3.53
编号：00042
CPI 增长率：0.3%
GDP 科学计数：1.26e+06
````

### 方法2：`format()` 方法

```python
"{}".format(变量)
# 或带占位符编号：
"第{0}个：{1}".format(1, "值")
# 填充与宽度：
"{0:0>6}".format(42)   # '000042'（右对齐，补0到6位）
```

### 📖 例题20：format() 和 % 哪个更灵活

````python
# 📖 例题20：format() 和 % 哪个更灵活

name = "张三"
age = 25
print("姓名：{}，年龄：{}".format(name, age))
````

### 方法3：f-string（Python 3.6+，推荐！）

在普通字符串前加字母 `f`，就可以在 `{}` 中直接写变量或表达式：

```python
name = "张三"
score = 88

print(f"{name}的成绩是{score}分")
```

可以这样理解：

- 引号里的普通文字：原样输出；
- `{name}`：把变量 `name` 的值放进来；
- `{score}`：把变量 `score` 的值放进来。

`{}` 里还可以放简单表达式：

```python
price = 12.5
count = 3
print(f"总价：{price * count}元")   # 总价：37.5元
```

### 控制小数位数

```python
mean = 45.3267
print(f"均值：{mean:.2f}")   # 均值：45.33
```

`:.2f` 表示“按浮点数显示，并保留 2 位小数”。

### 显示百分比

```python
rate = 0.1234
print(f"增长率：{rate:.1%}")  # 增长率：12.3%
```

> 本课程写报告句子、结果输出时，优先使用 f-string。

## 3.7 本章小结

这一章最重要的不是记住所有方法，而是建立下面这条思路：

```text
创建字符串
   ↓
按位置取内容：索引 / 切片
   ↓
清洗和加工：strip / replace / split / join
   ↓
查找和判断：find / count / startswith / endswith
   ↓
把结果输出成句子：f-string
```

### 必须掌握

| 知识点 | 一句话记忆 |
|--------|------------|
| 引号 | 单引号、双引号都能创建字符串 |
| 转义字符 | `\n` 换行，`\t` 制表，`\\` 反斜杠 |
| `len()` | 字符串有多少个字符 |
| 索引 | `s[0]` 第一个，`s[-1]` 最后一个 |
| 切片 | `s[start:stop]` 左含右不含 |
| `split()` | 一个字符串拆成多个部分 |
| `join()` | 多个字符串合成一个 |
| `replace()` | 替换指定内容 |
| `strip()` | 清理两端空白 |
| f-string | 把变量自然地放进一句话 |

### 三个最容易错的地方

1. **字符串索引从 0 开始。**
2. **字符串方法通常不会修改原字符串，要保存返回值。**
3. **中文标点和英文标点不是同一个字符。**
