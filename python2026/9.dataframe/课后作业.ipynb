## 一、选择题

### 1. pandas的导入方式通常是？

A. `import pandas`
B. `import pandas as pd`
C. `from pandas import *`
D. `include pandas`

**答案：B**
常规做法是导入时取别名 pd，简洁常用。

---

### 2. 读取Excel文件用？

A. `pd.read_file()`
B. `pd.read_excel()`
C. `pd.read_csv()`
D. `pd.load()`

**答案：B**
read_excel() 读Excel，read_csv() 读CSV。

---

### 3. 查看DataFrame前5行用？

A. `df.first(5)`
B. `df.head(5)`
C. `df.top(5)`
D. `df.show(5)`

**答案：B**
head() 看前5行，tail() 看后5行。

---

### 4. 删除缺失值用？

A. `df.fillna()`
B. `df.dropna()`
C. `df.delete_na()`
D. `df.remove_na()`

**答案：B**
dropna() 删除缺失行，fillna() 填充缺失值。

---

### 5. 分组统计用？

A. `df.group()`
B. `df.groupby()`
C. `df.groupbyall()`
D. `df.groupby_mean()`

**答案：B**
groupby() 分组，然后接统计方法如 .mean()。

## 三、代码题

### 1. 创建DataFrame，打印shape。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", "李四"], "成绩": [98, 88]})
print(df.shape)
````

**答案：** (2, 2)

---

### 2. 用字典创建DataFrame，筛选成绩>80。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", "李四", "王五"], "成绩": [98, 88, 70]})
print(df[df["成绩"] > 80])
````

**答案：** 张三、李四

---

### 3. 计算总分列并打印。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"语文": [98, 88], "数学": [85, 92]})
df["总分"] = df["语文"] + df["数学"]
print(df["总分"])
````

**答案：** 183, 180

---

### 4. 分组算平均分。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"班级": ["一班", "一班", "二班"], "成绩": [98, 88, 90]})
print(df.groupby("班级")["成绩"].mean())
````

**答案：** 一班: 93.0, 二班: 90.0

---

### 5. 删除缺失值并打印。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", None, "王五"], "成绩": [98, None, 70]})
print(df.dropna())
````

**答案：** 只剩张三和王五两行
