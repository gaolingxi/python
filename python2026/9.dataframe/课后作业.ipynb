## 三、代码题

### 1. 创建DataFrame，打印shape。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", "李四"], "成绩": [98, 88]})
print(df.shape)
````

**答案：** (2, 2)

---

### 2. 用字典创建DataFrame，筛选成绩>80。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", "李四", "王五"], "成绩": [98, 88, 70]})
print(df[df["成绩"] > 80])
````

**答案：** 张三、李四

---

### 3. 计算总分列并打印。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"语文": [98, 88], "数学": [85, 92]})
df["总分"] = df["语文"] + df["数学"]
print(df["总分"])
````

**答案：** 183, 180

---

### 4. 分组算平均分。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"班级": ["一班", "一班", "二班"], "成绩": [98, 88, 90]})
print(df.groupby("班级")["成绩"].mean())
````

**答案：** 一班: 93.0, 二班: 90.0

---

### 5. 删除缺失值并打印。

````python
# 请在下面填写代码：
import pandas as pd
df = pd.DataFrame({"姓名": ["张三", None, "王五"], "成绩": [98, None, 70]})
print(df.dropna())
````

**答案：** 只剩张三和王五两行
