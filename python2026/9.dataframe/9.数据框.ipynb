## 9.2 什么是DataFrame

> **DataFrame = 带列名的表格**，像Excel的工作表，但能写代码操作。

| 特征 | 说明 |
|------|------|
| 行 | 每条记录（如一个学生） |
| 列 | 每个字段（如姓名、成绩） |
| 列名 | 用列名取数据，不用索引 |
| 类型 | 每列有自己的数据类型 |

### 📖 例题3：DataFrame的基本结构

````python
# 📖 例题3：DataFrame的基本结构
# DataFrame = 行 + 列 + 索引

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '语文': [98, 88, 70],
    '数学': [85, 90, 88]
})

print(df)
print(f"\n形状: {df.shape}")  # (3行, 2列)
print(f"列名: {list(df.columns)}")
print(f"索引: {list(df.index)}")
````

### 📖 例题4：用字典列表创建DataFrame

````python
# 📖 例题4：用字典列表创建DataFrame
# 每个字典代表一行数据

import pandas as pd

# 方式1：字典的列表（每个字典是一行）
data = [
    {'姓名': '张三', '成绩': 98},
    {'姓名': '李四', '成绩': 88},
    {'姓名': '王五', '成绩': 70}
]
df1 = pd.DataFrame(data)
print(f"方式1:")
print(df1)

# 方式2：列表的字典（每个键是一列）
data2 = {
    '姓名': ['张三', '李四', '王五'],
    '成绩': [98, 88, 70]
}
df2 = pd.DataFrame(data2)
print(f"\n方式2:")
print(df2)

# 两种方式结果一样，但数据组织方式不同
````

## 9.3 创建DataFrame

> 三种常见方式：字典、列表、读文件。

### 📖 例题5：字典创建 vs 列表创建

````python
# 📖 例题5：字典创建 vs 列表创建
# 两种创建方式对比

import pandas as pd

# 方式1：字典创建（每个键是一列）
df_dict = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '成绩': [98, 88, 70]
})
print(f"字典创建:")
print(df_dict)

# 方式2：列表创建（需要指定columns参数）
df_list = pd.DataFrame(
    [['张三', 98], ['李四', 88], ['王五', 70]],
    columns=['姓名', '成绩']
)
print(f"\n列表创建:")
print(df_list)

# 方式3：不指定columns（自动编号0,1,2...）
df_auto = pd.DataFrame([['张三', 98], ['李四', 88]])
print(f"\n自动编号:")
print(df_auto)
````

## 9.6 数据筛选

> 按条件筛选行 - 数学不及格的学生

### 📖 例题9：条件筛选——只看及格的学生

````python
# 📖 例题9：条件筛选——只看及格的学生
# DataFrame的筛选功能

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '成绩': [98, 88, 70, 55]
})

# 筛选：成绩>=60
pass_students = df[df['成绩'] >= 60]
print(f"及格学生:")
print(pass_students)

# 筛选：成绩<60
fail_students = df[df['成绩'] < 60]
print(f"\n不及格学生:")
print(fail_students)

# 筛选：成绩在60-80之间
mid_students = df[(df['成绩'] >= 60) & (df['成绩'] <= 80)]
print(f"\n中等学生:")
print(mid_students)
````

### 📖 例题10：多条件筛选——组合条件

````python
# 📖 例题10：多条件筛选——组合条件
# & 表示"且"，| 表示"或"

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '语文': [98, 88, 70, 85, 92],
    '数学': [85, 90, 88, 55, 76]
})

# 条件1：语文>=80 且 数学>=80
both_good = df[(df['语文'] >= 80) & (df['数学'] >= 80)]
print(f"双科优秀:")
print(both_good)

# 条件2：语文>=80 或 数学>=80
either_good = df[(df['语文'] >= 80) | (df['数学'] >= 80)]
print(f"\n单科优秀:")
print(either_good)

# 条件3：语文<80 且 数学<80
both_bad = df[(df['语文'] < 80) & (df['数学'] < 80)]
print(f"\n双科需要努力:")
print(both_bad)
````

### loc 和 iloc

| 方法 | 索引方式 | 例子 |
|------|----------|------|
| loc | 按列名/索引标签 | df.loc[0, 语文] |
| iloc | 按位置索引 | df.iloc[0, 0] |

### 📖 例题11：loc和iloc的区别

````python
# 📖 例题11：loc和iloc的区别
# 按标签索引 vs 按位置索引

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '成绩': [98, 88, 70]
}, index=[2, 0, 1])  # 自定义索引

print(f"原始数据:")
print(df)

# loc：按标签（索引值）取值
print(f"\nloc按标签:")
print(df.loc[0])  # 取索引为0的行

# iloc：按位置（位置号）取值
print(f"\niloc按位置:")
print(df.iloc[1])  # 取第1行（从0开始）
````

### 📖 例题12：只看某一列的3种方法

````python
# 📖 例题12：只看某一列的3种方法
# DataFrame选列

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '语文': [98, 88, 70],
    '数学': [85, 90, 88]
})

# 方式1：df['列名']
print(f"方式1:")
print(df['语文'])

# 方式2：df.列名（列名没有特殊字符时）
print(f"\n方式2:")
print(df.语文)

# 方式3：df.loc[:, '列名']
print(f"\n方式3:")
print(df.loc[:, '语文'])

# 选多列
print(f"\n选多列:")
print(df[['姓名', '语文']])
````

### 📖 例题13：loc和iloc的配合使用

````python
# 📖 例题13：loc和iloc的配合使用
# 按条件+位置筛选

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '成绩': [98, 88, 70, 85]
}, index=[3, 1, 2, 0])

print(f"原始数据:")
print(df)

# loc：先按条件筛选，再选列
print(f"\nloc按条件:")
print(df.loc[df['成绩'] >= 80, ['姓名', '成绩']])

# iloc：按位置筛选
print(f"\niloc按位置:")
print(df.iloc[1:3, :2])  # 第1-2行，前2列

# 组合：先loc筛选，再iloc选位置
print(f"\n组合使用:")
filtered = df.loc[df['成绩'] >= 80]
print(filtered.iloc[0])  # 筛选结果的第1行
````

## 9.7 数据清洗

> 真实数据总有问题：缺失值、重复值、类型不对。先清洗才能分析。

### 缺失值处理

### 📖 例题14：处理缺失值

````python
# 📖 例题14：处理缺失值
# 真实数据中经常有缺失

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '成绩': [98, 88, None, 85]  # None表示缺失
})

# 检查缺失值
print(f"原始数据:")
print(df)
print(f"\n缺失值:")
print(df.isnull())  # True表示缺失

# 统计缺失值
print(f"\n各列缺失数:")
print(df.isnull().sum())

# 删除缺失行
df_clean = df.dropna()
print(f"\n删除缺失后:")
print(df_clean)

# 填充缺失值
df_fill = df.fillna(0)  # 用0填充
print(f"\n填充0后:")
print(df_fill)
````

### 重复值处理

### 📖 例题15：处理重复值

````python
# 📖 例题15：处理重复值
# 数据去重

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '张三', '王五'],
    '成绩': [98, 88, 98, 70]  # 张三重了
})

print(f"原始数据:")
print(df)

# 检查重复值
print(f"\n重复值:")
print(df.duplicated())  # True表示重复

# 删除重复行
df_clean = df.drop_duplicates()
print(f"\n去重后:")
print(df_clean)

# 只保留第一次出现的
df_first = df.drop_duplicates(keep='first')
print(f"\n保留第一次:")
print(df_first)

# 按姓名去重
df_by_name = df.drop_duplicates(subset='姓名')
print(f"\n按姓名去重:")
print(df_by_name)
````

### 类型转换

### 📖 例题16：数据类型转换

````python
# 📖 例题16：数据类型转换
# 字符串转数字、日期转时间

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '成绩': ['98', '88', '70']  # 字符串格式
})

print(f"原始数据:")
print(df)
print(f"成绩类型: {df['成绩'].dtype}")  # object（字符串）

# 转换为数字
df['成绩'] = df['成绩'].astype(float)
print(f"\n转换后:")
print(df)
print(f"成绩类型: {df['成绩'].dtype}")  # float64

# 现在可以算平均分了
print(f"平均分: {df['成绩'].mean():.1f}")
````

## 9.8 数据统计

### 📖 例题18：describe()快速统计

````python
# 📖 例题18：describe()快速统计
# 一键查看数据的统计摘要

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '成绩': [98, 88, 70, 85, 92]
})

# describe()：快速统计
print(f"成绩统计:")
print(df['成绩'].describe())

# 包括：count, mean, std, min, 25%, 50%, 75%, max
# count: 有效值个数
# mean: 平均值
# std: 标准差
# min/max: 最小/最大值
# 25%/50%/75%: 四分位数
````

### 分组统计 groupby

### 📖 例题19：分组统计——按班级算平均分

````python
# 📖 例题19：分组统计——按班级算平均分
# groupby + 聚合函数

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '班级': ['一班', '一班', '二班', '二班'],
    '成绩': [98, 88, 70, 85]
})

# 按班级分组，算平均分
avg_by_class = df.groupby('班级')['成绩'].mean()
print(f"各班平均分:")
print(avg_by_class)

# 按班级分组，算总分
sum_by_class = df.groupby('班级')['成绩'].sum()
print(f"\n各班总分:")
print(sum_by_class)

# 按班级分组，算人数
count_by_class = df.groupby('班级')['成绩'].count()
print(f"\n各班人数:")
print(count_by_class)
````

### 📖 例题20：算总分、平均分、最高分

````python
# 📖 例题20：算总分、平均分、最高分
# 常用统计组合

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六'],
    '语文': [98, 88, 70, 85],
    '数学': [85, 90, 88, 55]
})

# 算总分
df['总分'] = df['语文'] + df['数学']
print(f"总分:")
print(df)

# 算平均分
df['平均分'] = df['总分'] / 2
print(f"\n平均分:")
print(df)

# 找最高分学生
max_idx = df['总分'].idxmax()
print(f"\n最高分: {df.loc[max_idx, '总分']}")
print(f"学生: {df.loc[max_idx, '姓名']}")

# 找最低分学生
min_idx = df['总分'].idxmin()
print(f"\n最低分: {df.loc[min_idx, '总分']}")
print(f"学生: {df.loc[min_idx, '姓名']}")
````

### 📖 例题21：agg一次算多个统计量

````python
# 📖 例题21：agg一次算多个统计量
# 聚合函数的高级用法

import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '成绩': [98, 88, 70, 85, 92]
})

# 方式1：单个统计量
print(f"平均分: {df['成绩'].mean():.2f}")
print(f"最高分: {df['成绩'].max()}")
print(f"最低分: {df['成绩'].min()}")

# 方式2：agg一次算多个
stats = df['成绩'].agg(['mean', 'max', 'min', 'std'])
print(f"\nagg统计:")
print(stats)

# 方式3：自定义聚合函数
custom_stats = df['成绩'].agg({
    '成绩': ['mean', 'max', 'min', 'count']
})
print(f"\n自定义聚合:")
print(custom_stats)
````
