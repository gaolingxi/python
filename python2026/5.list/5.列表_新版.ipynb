## 5.2 访问元素

> 列表里每个位置都有编号，从 **0** 开始（跟字符串一样）。

### 📖 例题3：访问列表元素

````python
# 例题3：索引从0开始，切片左闭右开
scores = [90, 80, 70, 60, 50]
print(scores[0])       # 第1个
print(scores[-1])      # 最后1个
print(scores[-2])      # 倒数第2个
print(scores[1:3])     # 第2到第3个
print(scores[:3])      # 前3个
print(scores[-2:])     # 最后2个
print(scores[::2])     # 步长2：每隔一个取一个
````

</details>

## 5.3 遍历列表

> 遍历 = 一个一个拿出来，做同样的操作。30个学生，挨个算总分。

### 📖 例题5：遍历列表——最基础的操作

````python
# 例题5：列表遍历的两种方式
scores = [90, 80, 70]
for score in scores:
    print(score)
for i in range(len(scores)):
    print(i, scores[i])
````

<details>
<summary>🤔 思考3：直接写 `for x in scores`，与按索引 `for i in range(len(scores))` 有什么不同？（点击展开答案）</summary>

第一种直接拿到**元素值**，适合逐个处理成绩；第二种先拿到**位置**，适合既要知道第几个元素、又要读取该元素的情况。只需要编号和值时，还可用 `enumerate()`。

</details>

### 📖 例题6：多项式求值（原版进阶例子，简化数字）

````python
# 例题6：保留原版多项式求值，但只有3项
# P(x) = 1 + 2x + 3x²
coef = [1, 2, 3]
x = 2
p = 0
for i in range(len(coef)):
    p = p + coef[i] * x ** i
print(p)  # 17
````

### 遍历的进阶用法：多项式求值

原版系数列表的思路不变，只用三个系数计算1 + 2×2 + 3×2² = 17。这是进阶阅读，不作为需要反复练习的基础题。

## 5.4 列表的独有操作

> 列表是**可变**的——能加、能改、能删。这是它跟字符串、元组最大的区别。

### 添加元素：append()

```python
列表.append(元素)     # 在末尾加一个
```

> ⚠️ **注意**：`append()` 返回 `None`，不能这样写 `x = list1.append(...)`。

### 📖 例题8：append的坑——返回None

````python
# 例题8：append会修改原列表，但返回None
nums = [1, 2]
result = nums.append(3)
print(nums)     # [1, 2, 3]
print(result)   # None
````

<details>
<summary>🤔 思考4：为什么 `result = nums.append(3)` 不能当作新列表使用？（点击展开答案）</summary>

`append()` 会**直接修改原列表**，但返回值是 `None`。应先写 `nums.append(3)`，之后再读取 `nums`；不要把返回值当成列表。

</details>

### 📖 例题9：append的正确用法

````python
# 例题9：append修改原列表，+ 创建新列表
nums = []
for x in [1, 2, 3]:
    nums.append(x)
print(nums)
new_nums = nums + [4]
print(nums)       # 原列表不变
print(new_nums)   # 新列表
````

### ⚠️ append的坑：返回None

> `append()` 会修改原列表，但返回 `None`。别写错成 `list2 = list1.append(x)`。

### 修改元素：通过索引

```python
列表[索引] = 新值
```

### 📖 例题10：修改列表元素

````python
# 例题10：通过索引修改元素
scores = [80, 70, 60]
scores[1] = 90
print(scores)

# 原版保留：用循环批量修改
scores = [80, 50, 90]
for i in range(len(scores)):
    if scores[i] < 60:
        scores[i] = 0
print(scores)  # [80, 0, 90]
````

### 删除元素

```python
del 列表[索引]           # 按索引删
列表.remove(值)          # 按值删（删第一个匹配）
```

### 📖 例题11：del vs remove——两种删除方法

````python
# 例题11：del按位置，remove按值
a = [10, 20, 30]
del a[0]
print(a)

# 原版强调：remove只删除第一个匹配值
b = [10, 20, 20, 30]
b.remove(20)
print(b)  # [10, 20, 30]
````

### 📖 例题12：pop()返回被删除的值（对比del与remove）

````python
# 例题12：pop既按位置删除，又返回被删元素
nums = [10, 20, 30]
deleted = nums.pop(1)
print(deleted)  # 20
print(nums)     # [10, 30]
````

<details>
<summary>🤔 思考5：知道要删除的位置，也想知道删掉的值，为什么选 `pop()` 而不是 `del` 或 `remove()`？（点击展开答案）</summary>

`del` 按位置删除，但不返回被删除值；`remove()` 按**值**删除；`pop(位置)` 按**位置**删除，并返回这个元素。三者解决的需求不同。

</details>

### 删除元素：3种方法对比

| 方法 | 语法 | 删除什么 |
|------|------|----------|
| `del` | `del list[索引]` | 按位置删 |
| `pop` | `list.pop(索引)` | 按位置删+返回值 |
| `remove` | `list.remove(值)` | 按值删（第一个匹配） |

### ⭐ 旧题新做（必讲）：第3章会员名重复检查 → 第5章列表方法

**同一道题，沿用第3章原版的6个会员名：**

```text
Alice、mr、MrAdam、HGW、ALICE、Alice
```

网站规定会员名**不区分大小写**。我们要判断这6个会员名是不是唯一，尤其注意：`mr` 不能因为出现在 `MrAdam` 里就被误判为重复。

**第3章的旧方法：字符串。** 分别保存会员名，统一为小写，再用 `|` 标记每个名字的边界，最后用 `str.count("|alice|")` 统计完整名字。第3章新版也已恢复这道原题。

**第5章的新方法：列表。** 会员名本来就是6个独立元素，不需要先拼成一大段文字：

```python
names = ["Alice", "mr", "MrAdam", "HGW", "ALICE", "Alice"]
```

1. 用 `lower()` 统一大小写，并用 `append()` 保存新列表。
2. 用 `list.count()` 统计**完整列表元素**出现的次数。
3. 用 `index()` 找第一次出现的位置，再识别重复登记的位置。

| 比较 | 第3章字符串方法 | 第5章列表方法 |
|---|---|---|
| 数据表示 | 拼成一段文字 | 每个人名是一个列表元素 |
| 防止误匹配 | 必须增加 `|` 边界 | `count()` 天然按完整元素匹配 |
| 重复个数 | `all_names.count("|alice|")` | `lower_names.count("alice")` |

**结论：** 两种方法都能完成同一道题，但列表更符合“一组会员名”的数据结构。要学会根据数据本身选方法，而不仅是记住语法。

````python
# ⭐ 旧题新做：第3章字符串 → 第5章列表
names = ["Alice", "mr", "MrAdam", "HGW", "ALICE", "Alice"]

# 第一步：统一大小写
lower_names = []
for name in names:
    lower_names.append(name.lower())

# 第二步：数某个完整会员名出现几次
print("alice出现：", lower_names.count("alice"))  # 3
print("mr出现：", lower_names.count("mr"))        # 1

# 第三步：找出重复登记的具体位置
for i, name in enumerate(lower_names):
    if lower_names.index(name) != i:
        first = lower_names.index(name)
        print("第", i + 1, "个名字", names[i], "重复，第一次在第", first + 1, "个位置")
````

### 排序：sort() 和 sorted()

| 方法 | 作用 | 原列表变不变 |
|------|------|-------------|
| `列表.sort()` | 原地排序（从小到大） | **变了** |
| `列表.sort(reverse=True)` | 原地降序 | **变了** |
| `sorted(列表)` | 返回新列表（原列表不变） | **没变** |
| `sorted(列表, reverse=True)` | 返回降序新列表 | **没变** |

> 大白话：`sort()` 是**就地改**，`sorted()` 是**另开一个新列表**。

### 📖 例题15：sort vs sorted——两种排序方法

````python
# 例题15：sort会修改原列表，sorted创建新列表
a = [3, 1, 2]
result = a.sort()
print(a)       # [1, 2, 3]
print(result)  # None

b = [3, 1, 2]
c = sorted(b, reverse=True)
print(b)  # b不变
print(c)  # [3, 2, 1]
````

<details>
<summary>🤔 思考7：`a.sort()` 和 `sorted(a)` 都能排序，为什么还要保留两种写法？（点击展开答案）</summary>

`a.sort()` **直接修改原列表**；`sorted(a)` 返回新的排序结果，**原列表不变**。如果后面还需要原来的录入顺序，优先用 `sorted(a)`。

</details>

### 📖 例题18：列表推导式基础

## 5.7 二维列表

> 列表里还能装列表——这就是**二维列表**，相当于一个表格。

酒店有很多楼层，每层有很多房间 → 二维列表就是"表格"。

```python
二维列表 = [[元素11, 元素12, ...],
            [元素21, 元素22, ...],
            ...]
```

> 访问方式：`二维列表[行][列]`——先选行，再选列。

### 📖 例题21：二维列表访问

````python
# 例题21：二维列表——先取行再取列
table = [[1, 2, 3], [4, 5, 6]]
print(table[0])      # 第1行
print(table[1][2])   # 第2行第3列：6

# 原版保留：从每一行取同一列
for row in table:
    print(row[1])   # 第2列的2和5
````

### 📖 例题22：创建二维列表的多种方式

````python
# 例题22：原版的三种二维列表创建方式，简化为2行3列
a = []
for i in range(2):
    row = []
    for j in range(3):
        row.append(0)
    a.append(row)
b = [[0 for j in range(3)] for i in range(2)]
c = [[0, 0, 0], [0, 0, 0]]
print(a, b, c)
````

<details>
<summary>🤔 思考9：为什么 `[[0] * 3] * 2` 不适合创建两行可独立修改的表格？（点击展开答案）</summary>

这种写法重复引用**同一行列表**。改第一行的某个元素，第二行同一列也会变化。正确做法之一是 `[[0] * 3 for _ in range(2)]`，每次循环生成新的子列表。

</details>

### 创建二维列表的多种方式

保留原版三种方式：嵌套循环、列表推导式、逐行直接创建。

注意：如果需要独立修改每一行，不要写成[[0] * 3] * 2，因为两行指向同一个子列表。可以改用[[0] * 3 for _ in range(2)]。

### 📖 例题23：二维列表遍历

````python
# 例题23：二维列表遍历的两种写法
table = [[1, 2], [3, 4]]

# 原版保留：双重for按行和列访问
for i in range(len(table)):
    for j in range(len(table[i])):
        print(i, j, table[i][j])

# 简单写法：每行做一次统计
for row in table:
    print("行总和", sum(row))
````

### 用列表推导式创建二维列表

示例：[[0] * 3 for _ in range(2)]。外层控制行数，内层控制每行元素数，确保两行互不共享。

## 5.8 元组

> **元组 = 不能改的列表**——用圆括号 `()` 创建。

```python
list1 = [1, 2, 3]      # 列表：能加能改能删
tuple1 = (1, 2, 3)     # 元组：定了就不能动
```

> **大白话**：列表像**铅笔写的字**，能擦能改；元组像**钢笔写的字**，擦不掉。
>
> 数据不该被改的时候用元组——比如GDP、省份名这些**固定的东西**。

### 📖 例题24：元组不可变——不能修改

````python
# 例题24：元组能访问，不能修改
nums = (1, 2, 3)
print(nums[0])
# nums[0] = 9   # TypeError
# nums.append(4)  # AttributeError
````

<details>
<summary>🤔 思考10：元组不能改变元素，为什么 `a + b` 或 `sorted(a)` 仍然能运行？（点击展开答案）</summary>

不能修改是指**原元组的元素不能原地改变**。`a + b` 会生成新元组，`sorted(a)` 会生成新列表，都不修改原元组。

</details>

### 📖 例题25：元组的拼接、重复、排序

````python
# 例题25：元组的拼接、重复、排序
a = (1, 2)
b = (3,)
print(a + b)
print(a * 2)
print(sorted((3, 1, 2)))  # sorted产生列表
````

### 📖 例题26：遍历成对的元组数据（保留原版GDP情境）

````python
# 例题26：遍历成对的元组，并求总值
# 以下是假设的地区GDP，用来演示，不代表真实统计
gdp = (("甲地", 100), ("乙地", 150))
total = 0
for area, value in gdp:
    print(area, value)
    total = total + value
print("合计", total)  # 250
````

### 列表 vs 元组

| 特性 | 列表 `[]` | 元组 `()` |
|------|----------|----------|
| 可变？ | 能加能改能删 | 不能改 |
| 方法多 | sort/append/remove... | 只有 count/index |
| 用途 | 要操作的数据 | 固定的数据（如坐标、GDP） |
