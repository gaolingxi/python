## 5.2 访问元素

> 列表里每个位置都有编号，从 **0** 开始（跟字符串一样）。

### 📖 例题3：访问列表元素

````python
# 例题4：长度与成员检查
scores = [90, 80, 70]
print(len(scores))   # 3
print(80 in scores)  # True
print(60 in scores)  # False
````

### 长度和成员检查

```python
len(列表)              # 列表长度
元素 in 列表            # True 或 False
```

> `in` 用来判断某个值在不在列表里，做筛选很好用。

### 📖 例题4：长度与成员检查

````python
# 📖 例题4：长度与成员检查
# 判断某个值在不在列表里，做筛选很好用

scores = [98, 88, 70, 85, 90]

# 取长度
print(f"成绩个数: {len(scores)}")   # 5

# 检查成员：用 in 判断
print(f"70在列表里吗: {70 in scores}")  # True
print(f"60在列表里吗: {60 in scores}")  # False

# 实际应用：筛选合格成绩
pass_threshold = 60
for score in scores:
    if score >= pass_threshold:
        print(f"  {score} 分 - 合格")
    else:
        print(f"  {score} 分 - 不合格")
````

## 5.3 遍历列表

> 遍历 = 一个一个拿出来，做同样的操作。30个学生，挨个算总分。

### 📖 例题5：遍历列表——最基础的操作

````python
# 📖 例题5：遍历列表——最基础的操作
# 列表最常用的操作：逐个取出每个元素

students = ["张三", "李四", "王五"]

# 方式1：直接遍历（最常用）
print("方式1：直接遍历")
for name in students:
    print(f"  {name}")

# 方式2：用索引遍历（需要知道位置时用）
print("方式2：用索引遍历")
for i in range(len(students)):
    print(f"  第{i+1}名: {students[i]}")
````

### 📖 例题6：多项式求值（原版进阶例子，简化数字）

````python
# 例题6：保留原版多项式求值，但只有3项
# P(x) = 1 + 2x + 3x²
coef = [1, 2, 3]
x = 2
p = 0
for i in range(len(coef)):
    p = p + coef[i] * x ** i
print(p)  # 17
````

### 遍历的进阶用法：多项式求值

原版系数列表的思路不变，只用三个系数计算1 + 2×2 + 3×2² = 17。这是进阶阅读，不作为需要反复练习的基础题。

## 5.4 列表的独有操作

> 列表是**可变**的——能加、能改、能删。这是它跟字符串、元组最大的区别。

### 添加元素：append()

```python
列表.append(元素)     # 在末尾加一个
```

> ⚠️ **注意**：`append()` 返回 `None`，不能这样写 `x = list1.append(...)`。

### 📖 例题8：append的坑——返回None

````python
# 例题8：append会修改原列表，但返回None
nums = [1, 2]
result = nums.append(3)
print(nums)     # [1, 2, 3]
print(result)   # None
````

### 📖 例题9：append的正确用法

````python
# 例题9：用append逐步建立列表
nums = []
for x in [1, 2, 3]:
    nums.append(x)
print(nums)
````

### ⚠️ append的坑：返回None

> `append()` 会修改原列表，但返回 `None`。别写错成 `list2 = list1.append(x)`。

### 修改元素：通过索引

```python
列表[索引] = 新值
```

### 📖 例题10：修改列表元素

````python
# 例题10：按索引修改一个元素
scores = [80, 70, 60]
scores[1] = 90
print(scores)
````

### 删除元素

```python
del 列表[索引]           # 按索引删
列表.remove(值)          # 按值删（删第一个匹配）
```

### 📖 例题11：del vs remove——两种删除方法

````python
# 例题11：del按位置，remove按值
a = [10, 20, 30]
del a[0]
print(a)
b = [10, 20, 30]
b.remove(20)
print(b)
````

### 📖 例题12：pop()返回被删除的值（对比del与remove）

````python
# 例题12：pop既按位置删除，又返回被删元素
nums = [10, 20, 30]
deleted = nums.pop(1)
print(deleted)  # 20
print(nums)     # [10, 30]
````

### 删除元素：3种方法对比

| 方法 | 语法 | 删除什么 |
|------|------|----------|
| `del` | `del list[索引]` | 按位置删 |
| `pop` | `list.pop(索引)` | 按位置删+返回值 |
| `remove` | `list.remove(值)` | 按值删（第一个匹配） |

> 列表能算东西：数次数、找位置、求和、排序。

### count 和 index

```python
列表.count(值)     # 数某值出现几次
列表.index(值)     # 找某值第一次出现的索引
```

### 📖 例题13：count和index——统计方法

### 第3章与第5章的count()比较（保留原版知识衔接）

字符串count()数的是子字符串；列表count()数的是完整元素。例如："aba".count("a") 与 ["a", "b", "a"].count("a") 都等于2，但对象不同。原版较长的会员重复案例在这里简化为概念对比。

### 排序：sort() 和 sorted()

| 方法 | 作用 | 原列表变不变 |
|------|------|-------------|
| `列表.sort()` | 原地排序（从小到大） | **变了** |
| `列表.sort(reverse=True)` | 原地降序 | **变了** |
| `sorted(列表)` | 返回新列表（原列表不变） | **没变** |
| `sorted(列表, reverse=True)` | 返回降序新列表 | **没变** |

> 大白话：`sort()` 是**就地改**，`sorted()` 是**另开一个新列表**。

### 📖 例题15：sort vs sorted——两种排序方法

````python
# 例题15：sort改变原列表，sorted创建新列表
a = [3, 1, 2]
a.sort()
print(a)
b = [3, 1, 2]
c = sorted(b)
print(b, c)  # b没变
````

### 📖 例题16：用max()和index()找最高分学生

````python
# 例题16：找最高分与对应同学
names = ["小王", "小李", "小张"]
scores = [85, 92, 80]
position = scores.index(max(scores))
print(names[position], scores[position])
````

### 📖 例题17：排名应用——取前两名，不改变原列表

````python
# 例题17：从高到低排名，不改变原顺序
scores = [85, 92, 80]
ranking = sorted(scores, reverse=True)
print(ranking[:2])  # 前两名
print(scores)       # 原顺序不变
````

### 排序进阶：sort() vs sorted() 完整对比

> `sort()` 原地排序（改原列表），`sorted()` 返回新列表（不改原列表）。

## 5.6 列表推导式

> **列表推导式 = 一行代码生成列表**，比写for循环简洁得多。

```python
[表达式 for 变量 in 可迭代对象 if 条件]
```

> 大白话：`if` 不是必须的，加了就筛选，不加就全选。

### 📖 例题18：列表推导式基础

````python
# 例题18：for和推导式完成同一个任务
squares = []
for n in [1, 2, 3]:
    squares.append(n * n)
print(squares)
print([n * n for n in [1, 2, 3]])
````

### 📖 例题19：给每个成绩加 5 分

````python
# 例题19：每个成绩加5
scores = [70, 80, 90]
print([s + 5 for s in scores])
````

### 📖 例题20：只保留价格超过 5000 元的商品

````python
# 例题20：筛选价格
prices = [5, 18, 30, 8]
print([p for p in prices if p > 10])
````

### 列表推导式进阶

`[表达式 for 变量 in 原数据 if 条件]` 中的 `if` 可以选择保留哪些值；不需要筛选时可以省略。

## 5.7 二维列表

> 列表里还能装列表——这就是**二维列表**，相当于一个表格。

酒店有很多楼层，每层有很多房间 → 二维列表就是"表格"。

```python
二维列表 = [[元素11, 元素12, ...],
            [元素21, 元素22, ...],
            ...]
```

> 访问方式：`二维列表[行][列]`——先选行，再选列。

### 📖 例题21：二维列表访问

````python
# 例题21：二维索引
table = [[1, 2], [3, 4]]
print(table[0])
print(table[1][0])  # 3
````

### 📖 例题22：创建二维列表的多种方式

````python
# 例题22：原版的三种二维列表创建方式，简化为2行3列
a = []
for i in range(2):
    row = []
    for j in range(3):
        row.append(0)
    a.append(row)
b = [[0 for j in range(3)] for i in range(2)]
c = [[0, 0, 0], [0, 0, 0]]
print(a, b, c)
````

### 创建二维列表的多种方式

保留原版三种方式：嵌套循环、列表推导式、逐行直接创建。

注意：如果需要独立修改每一行，不要写成[[0] * 3] * 2，因为两行指向同一个子列表。可以改用[[0] * 3 for _ in range(2)]。

### 📖 例题23：二维列表遍历

````python
# 例题23：逐行遍历二维列表
table = [[1, 2, 3], [4, 5, 6]]
for i, row in enumerate(table, start=1):
    print("第", i, "行，总和", sum(row))
````

### 用列表推导式创建二维列表

示例：[[0] * 3 for _ in range(2)]。外层控制行数，内层控制每行元素数，确保两行互不共享。

## 5.8 元组

> **元组 = 不能改的列表**——用圆括号 `()` 创建。

```python
list1 = [1, 2, 3]      # 列表：能加能改能删
tuple1 = (1, 2, 3)     # 元组：定了就不能动
```

> **大白话**：列表像**铅笔写的字**，能擦能改；元组像**钢笔写的字**，擦不掉。
>
> 数据不该被改的时候用元组——比如GDP、省份名这些**固定的东西**。

### 📖 例题24：元组不可变——不能修改

````python
# 例题24：元组能访问，不能修改
nums = (1, 2, 3)
print(nums[0])
# nums[0] = 9   # TypeError
# nums.append(4)  # AttributeError
````

### 📖 例题25：元组的拼接、重复、排序

````python
# 例题25：元组的拼接、重复、排序
a = (1, 2)
b = (3,)
print(a + b)
print(a * 2)
print(sorted((3, 1, 2)))  # sorted产生列表
````

### 📖 例题26：遍历成对的元组数据（保留原版GDP情境）

````python
# 例题26：保留原版“地区GDP”情境，改为假设数据
gdp = (("甲地", 100), ("乙地", 150))
for area, value in gdp:
    print(area, value)
````

### 列表 vs 元组

| 特性 | 列表 `[]` | 元组 `()` |
|------|----------|----------|
| 可变？ | 能加能改能删 | 不能改 |
| 方法多 | sort/append/remove... | 只有 count/index |
| 用途 | 要操作的数据 | 固定的数据（如坐标、GDP） |
