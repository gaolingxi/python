**学完这份作业，重点掌握三类循环：**

1. `for x in nums`：读取原列表、计算新值并 `append()`；
2. `for i in range(len(nums))`：知道索引，修改原列表或关联另一列表；
3. `for row in matrix`：逐行读取二维列表，必要时在内部再写一层循环。
