## 5.2 访问元素

> 列表里每个位置都有编号，从 **0** 开始（跟字符串一样）。

### 📖 例题3：访问列表元素

````python
# 📖 例题3：访问列表元素
# 列表的每个位置都有编号，从0开始

scores = [98, 88, 70, 85, 90]

# 单个元素：用索引 [0] 开始
print(f"第1个成绩: {scores[0]}")   # 98
print(f"最后一个: {scores[-1]}")    # 90（-1表示倒数第1个）
print(f"倒数第2个: {scores[-2]}")   # 85

# 切片：取一段 [起始:结束]（不包含结束位置）
print(f"第2到第4个: {scores[1:4]}")  # [88, 70, 85]
print(f"前3个: {scores[:3]}")        # [98, 88, 70]
print(f"最后2个: {scores[-2:]}")     # [85, 90]

# 步长切片：每隔n个取一个
print(f"每隔1个: {scores[::2]}")     # [98, 70, 90]
````

### 长度和成员检查

```python
len(列表)              # 列表长度
元素 in 列表            # True 或 False
```

> `in` 用来判断某个值在不在列表里，做筛选很好用。

### 📖 例题4：长度与成员检查

````python
# 📖 例题4：长度与成员检查
# 判断某个值在不在列表里，做筛选很好用

scores = [98, 88, 70, 85, 90]

# 取长度
print(f"成绩个数: {len(scores)}")   # 5

# 检查成员：用 in 判断
print(f"70在列表里吗: {70 in scores}")  # True
print(f"60在列表里吗: {60 in scores}")  # False

# 实际应用：筛选合格成绩
pass_threshold = 60
for score in scores:
    if score >= pass_threshold:
        print(f"  {score} 分 - 合格")
    else:
        print(f"  {score} 分 - 不合格")
````

## 5.3 遍历列表

> 遍历 = 一个一个拿出来，做同样的操作。30个学生，挨个算总分。

### 📖 例题5：遍历列表——最基础的操作

````python
# 📖 例题5：遍历列表——最基础的操作
# 列表最常用的操作：逐个取出每个元素

students = ["张三", "李四", "王五"]

# 方式1：直接遍历（最常用）
print("方式1：直接遍历")
for name in students:
    print(f"  {name}")

# 方式2：用索引遍历（需要知道位置时用）
print("方式2：用索引遍历")
for i in range(len(students)):
    print(f"  第{i+1}名: {students[i]}")
````

### 📖 例题6：多项式求值——列表的进阶用法

````python
# 📖 例题6：根据位置取系数，计算多项式
coef = [1, 2, 4, 5, 6]
x = 23
p = 0
for i in range(len(coef)):
    p = p + coef[i] * x ** i
print(f"P({x}) = {p}")
````

### 遍历的进阶用法：多项式求值

> 经济学里经常要算 `y = a0 + a1*x + a2*x^2`，用列表存系数，遍历算结果。

### enumerate：同时拿编号和值

> 有时候不仅要知道"是谁"，还要知道"排第几"。`enumerate()` 帮你同时拿到编号和值。

### 📖 例题7：enumerate——同时拿编号和值

````python
# 📖 例题7：enumerate——同时拿编号和值
# 遍历时需要知道位置，用enumerate

students = ["张三", "李四", "王五"]

# start=0：从0开始编号（默认）
print("start=0:")
for idx, name in enumerate(students):
    print(f"  编号{idx}: {name}")

# start=1：从1开始编号（更自然）
print("start=1:")
for idx, name in enumerate(students, start=1):
    print(f"  第{idx}名: {name}")
````

## 5.4 列表的独有操作

> 列表是**可变**的——能加、能改、能删。这是它跟字符串、元组最大的区别。

### 添加元素：append()

```python
列表.append(元素)     # 在末尾加一个
```

> ⚠️ **注意**：`append()` 返回 `None`，不能这样写 `x = list1.append(...)`。

### 📖 例题8：append的坑——返回None

````python
# 📖 例题8：append的坑——返回None
# append会修改原列表，但返回None

scores = [98, 88, 70]

# 正确用法：直接调用，不接返回值
scores.append(85)
print(f"append后: {scores}")  # [98, 88, 70, 85]

# 错误用法：把返回值赋给变量
wrong = scores.append(90)  # 返回None！
print(f"返回值: {wrong}")    # None
print(f"原列表: {scores}")   # [98, 88, 70, 85, 90]（还是被改了）

# 记住：append是"原地修改"，不返回新列表
````

### 📖 例题9：append的正确用法

````python
# 📖 例题9：append的正确用法
# 不要写成 list2 = list1.append(x)

scores = [98, 88, 70]

# 错误写法：
# wrong = scores.append(85)  # wrong会变成None！
# 正确写法：
scores.append(85)
print(f"append后: {scores}")  # [98, 88, 70, 85]

# 如果确实要得到"原列表+新元素"，用 + 创建新列表
new_scores = scores + [90]
print(f"原列表: {scores}")    # [98, 88, 70, 85]（没变）
print(f"新列表: {new_scores}") # [98, 88, 70, 85, 90]
````

### ⚠️ append的坑：返回None

> `append()` 会修改原列表，但返回 `None`。别写错成 `list2 = list1.append(x)`。

### 修改元素：通过索引

```python
列表[索引] = 新值
```

### 📖 例题10：修改列表元素

````python
# 📖 例题10：修改列表元素
# 通过索引直接修改，原地生效

scores = [98, 88, 70]

# 把第3个成绩从70改成90
scores[2] = 90
print(f"修改后: {scores}")  # [98, 88, 90]

# 批量修改：把所有低于60的改成0（模拟"不及格清零"）
for i in range(len(scores)):
    if scores[i] < 60:
        scores[i] = 0

print(f"清零后: {scores}")
````

### 删除元素

```python
del 列表[索引]           # 按索引删
列表.remove(值)          # 按值删（删第一个匹配）
```

### 📖 例题11：del vs remove——两种删除方法

````python
# 📖 例题11：del vs remove——两种删除方法
# 按位置删 vs 按值删

scores = [98, 88, 70, 85, 90]

# del：按索引删（删掉位置0的元素）
del scores[0]
print(f"del后: {scores}")  # [88, 70, 85, 90]

# remove：按值删（删掉第一个匹配的）
scores2 = [98, 88, 70, 85, 70]
scores2.remove(70)
print(f"remove后: {scores2}")  # [98, 88, 85, 70]（只删第一个70）
````

### 📖 例题12：3种删除方法对比

````python
# 📖 例题12：3种删除方法对比
# del、pop、remove 的区别

# 方式1：del 按位置删（不返回值）
list1 = [98, 88, 70, 85, 90]
del list1[0]
print(f"del后: {list1}")  # [88, 70, 85, 90]

# 方式2：pop 按位置删（返回值）
list2 = [98, 88, 70, 85, 90]
removed = list2.pop(2)  # 删掉索引2的元素，返回它
print(f"pop返回值: {removed}")  # 70
print(f"pop后: {list2}")        # [98, 88, 85, 90]

# 方式3：remove 按值删（不返回值）
list3 = [98, 88, 70, 85, 90]
list3.remove(70)
print(f"remove后: {list3}")  # [98, 88, 85, 90]

# 总结：
# del：按位置，无返回值
# pop：按位置，有返回值（适合"取走"的场景）
# remove：按值，无返回值（适合"删掉某值"的场景）
````

### 删除元素：3种方法对比

| 方法 | 语法 | 删除什么 |
|------|------|----------|
| `del` | `del list[索引]` | 按位置删 |
| `pop` | `list.pop(索引)` | 按位置删+返回值 |
| `remove` | `list.remove(值)` | 按值删（第一个匹配） |

### 排序：sort() 和 sorted()

| 方法 | 作用 | 原列表变不变 |
|------|------|-------------|
| `列表.sort()` | 原地排序（从小到大） | **变了** |
| `列表.sort(reverse=True)` | 原地降序 | **变了** |
| `sorted(列表)` | 返回新列表（原列表不变） | **没变** |
| `sorted(列表, reverse=True)` | 返回降序新列表 | **没变** |

> 大白话：`sort()` 是**就地改**，`sorted()` 是**另开一个新列表**。

### 📖 例题15：sort vs sorted——两种排序方法

````python
# 📖 例题15：sort vs sorted——两种排序方法
# 一个原地排序，一个返回新列表

scores = [85, 98, 70]

# sort：原地排序（列表本身变了，返回None）
scores.sort()
print(f"sort后: {scores}")  # [70, 85, 98]
print(f"sort返回值: {scores.sort()}")  # None

# sorted：返回新列表（原列表不变）
scores2 = [85, 98, 70]
sorted_list = sorted(scores2)
print(f"原列表: {scores2}")      # [85, 98, 70]（没变）
print(f"sorted结果: {sorted_list}")  # [70, 85, 98]

# 降序：reverse=True
print(f"降序: {sorted(scores2, reverse=True)}")  # [98, 85, 70]
````

### 📖 例题17：已算出的总分怎样排序？

````python
# 📖 例题17：一维成绩排序
totals = [256, 232, 266]
print("排序前：", totals)
totals.sort(reverse=True)
print("sort 后：", totals)

original = [256, 232, 266]
ranked = sorted(original, reverse=True)
print("原列表：", original)
print("sorted 结果：", ranked)
````

### 排序进阶：sort() vs sorted() 完整对比

> `sort()` 原地排序（改原列表），`sorted()` 返回新列表（不改原列表）。

## 5.6 列表推导式

> **列表推导式 = 一行代码生成列表**，比写for循环简洁得多。

```python
[表达式 for 变量 in 可迭代对象 if 条件]
```

> 大白话：`if` 不是必须的，加了就筛选，不加就全选。

### 📖 例题18：列表推导式基础

````python
# 📖 例题18：列表推导式基础
# 一行代码生成列表，比for循环简洁

# 1. 生成1~5的数字
numbers = [i for i in range(1, 6)]
print(f"1~5: {numbers}")  # [1, 2, 3, 4, 5]

# 2. 每个数平方
squares = [i**2 for i in range(1, 6)]
print(f"平方: {squares}")  # [1, 4, 9, 16, 25]

# 3. 筛选：只留大于3的数
big = [i for i in range(1, 10) if i > 3]
print(f">3的数: {big}")  # [4, 5, 6, 7, 8, 9]
````

### 列表推导式进阶

`[表达式 for 变量 in 原数据 if 条件]` 中的 `if` 可以选择保留哪些值；不需要筛选时可以省略。

## 5.7 二维列表

> 列表里还能装列表——这就是**二维列表**，相当于一个表格。

酒店有很多楼层，每层有很多房间 → 二维列表就是"表格"。

```python
二维列表 = [[元素11, 元素12, ...],
            [元素21, 元素22, ...],
            ...]
```

> 访问方式：`二维列表[行][列]`——先选行，再选列。

### 📖 例题21：二维列表访问

````python
# 📖 例题21：二维列表访问
# 列表的列表——成绩表的标准结构

list2D = [
    [1, 2, 3],   # 第0行
    [4, 5, 6],   # 第1行
    [7, 8, 9]    # 第2行
]

# 访问单个元素：先选行，再选列
print(f"第1行第2列: {list2D[1][2]}")  # 6

# 访问整行
print(f"第0行: {list2D[0]}")  # [1, 2, 3]

# 访问第3列（需要遍历每行）
print("第2列:")
for row in list2D:
    print(f"  {row[2]}")  # 3, 6, 9
````

### 📖 例题22：创建二维列表的多种方式

````python
# 📖 例题22：创建二维列表的多种方式
# 创建4×5的矩阵

# 方式1：嵌套循环（最直观）
mat1 = []
for i in range(4):
    row = []
    for j in range(5):
        row.append(1)
    mat1.append(row)
print(f"方式1: {mat1}")

# 方式2：列表推导式（最简洁）
mat2 = [[1 for j in range(5)] for i in range(4)]
print(f"方式2: {mat2}")

# 方式3：成绩表方式
grade1 = [98, 88, 70]
grade2 = [85, 90, 88]
grade3 = [76, 82, 95]
list_grade = [grade1, grade2, grade3]
print(f"方式3 成绩表: {list_grade}")
````

### 创建二维列表的多种方式

### 📖 例题23：二维列表遍历

````python
# 📖 例题23：二维列表遍历
# 遍历成绩表，算总分

list_grade = [
    [98, 88, 70],   # 学生1
    [88, 64, 80],   # 学生2
    [80, 90, 96]    # 学生3
]

# 方式1：嵌套for
for i in range(len(list_grade)):
    for j in range(len(list_grade[i])):
        print(f"  学生{i+1}第{j+1}科: {list_grade[i][j]}")

# 方式2：enumerate + sum（更简洁）
print("用enumerate+sum:")
for i, grade in enumerate(list_grade, start=1):
    print(f"  学生{i}: {grade}, 总分: {sum(grade)}")
````

## 5.8 元组

> **元组 = 不能改的列表**——用圆括号 `()` 创建。

```python
list1 = [1, 2, 3]      # 列表：能加能改能删
tuple1 = (1, 2, 3)     # 元组：定了就不能动
```

> **大白话**：列表像**铅笔写的字**，能擦能改；元组像**钢笔写的字**，擦不掉。
>
> 数据不该被改的时候用元组——比如GDP、省份名这些**固定的东西**。

### 📖 例题24：元组不可变——不能修改

````python
# 📖 例题24：元组不可变——不能修改
# 元组创建后不能改，但可以读

tuple1 = (1, 2, 3, 4)

# 可以做的操作
print(f"元组: {tuple1}")
print(f"长度: {len(tuple1)}")     # 4
print(f"成员检查: {2 in tuple1}")  # True
print(f"访问: {tuple1[0]}")        # 1

# 不能做的操作（会报错）
# tuple1[0] = 10     # TypeError: 不能修改
# tuple1.append(5)   # AttributeError: 没有append方法
# tuple1.sort()      # AttributeError: 没有sort方法
````

### 📖 例题25：元组的拼接、重复、排序

````python
# 📖 例题25：元组的拼接、重复、排序
# 元组不可变，但可以做"不可变操作"

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

# 拼接：生成新元组
tuple3 = tuple1 + tuple2
print(f"拼接: {tuple3}")  # (1, 2, 3, 4, 5, 6)

# 重复
tuple4 = tuple1 * 2
print(f"重复: {tuple4}")  # (1, 2, 3, 1, 2, 3)

# 排序：sorted返回新列表，原元组不变
print(f"原元组: {tuple1}")         # (1, 2, 3)
print(f"sorted: {sorted(tuple1)}")  # [1, 2, 3]
````

### 📖 例题26：元组遍历——GDP数据

````python
# 📖 例题26：元组遍历——GDP数据
# 元组适合存"固定数据"，比如GDP统计

gdp = (
    ("中国", 1434290),
    ("美国", 2143320),
    ("日本", 508000),
)

# 遍历元组，解包每对数据
for country, value in gdp:
    print(f"{country}: {value}亿美元")

# 算总值
total = sum(v for _, v in gdp)
print(f"三国GDP总和: {total}亿美元")
````

### 列表 vs 元组

| 特性 | 列表 `[]` | 元组 `()` |
|------|----------|----------|
| 可变？ | 能加能改能删 | 不能改 |
| 方法多 | sort/append/remove... | 只有 count/index |
| 用途 | 要操作的数据 | 固定的数据（如坐标、GDP） |
