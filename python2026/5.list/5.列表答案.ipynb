## 5.1 列表的创建

> **列表 = 用方括号 `[ ]` 装一串东西**，东西之间用逗号分隔。

```python
list1 = [98, 88, 70]          # 数字列表
list2 = ["张三", "李四", "王五"]   # 字符串列表
list3 = []                      # 空列表
list4 = list(range(1, 6))       # 用range生成 [1,2,3,4,5]
```

> **列表能装不同类型**——数字、字符串、甚至别的列表都行。

### 📖 例题1：列表的拼接与重复

````python
# 📖 例题1：列表的拼接与重复
# 在经济学研究中，经常需要合并两组数据
# 比如：把两个班级的成绩合并统计

# 甲班成绩
class_a = [98, 88, 70]

# 乙班成绩
class_b = [85, 90, 88]

# 用 + 拼接两个列表（生成新列表）
all_scores = class_a + class_b
print(f"合并后: {all_scores}")    # [98, 88, 70, 85, 90, 88]
print(f"长度: {len(all_scores)}")  # 6个成绩

# 用 * 重复列表（注意：不是数值乘法！）
print(f"甲班成绩重复2次: {class_a * 2}")  # [98, 88, 70, 98, 88, 70]
````

## 5.2 访问元素

> 列表里每个位置都有编号，从 **0** 开始（跟字符串一样）。

### 📖 例题2：列表的5种创建方式

````python
# 📖 例题2：列表的5种创建方式
# 不同的场景，用不同的方式创建列表

# 方式1：直接写字（最简单，适合少量数据）
scores = [98, 88, 70]
print(f"方式1 直接写字: {scores}")

# 方式2：空列表 + append（适合数据逐步增加）
empty = []
empty.append(98)
empty.append(88)
empty.append(70)
print(f"方式2 空列表+append: {empty}")

# 方式3：用list()把别的序列变列表（元组、字符串都可以）
tuple_data = (98, 88, 70)
from_tuple = list(tuple_data)
print(f"方式3 list()转换: {from_tuple}")

# 方式4：用list(range())生成数字列表（适合连续数字）
numbers = list(range(1, 6))
print(f"方式4 list(range): {numbers}")

# 方式5：用split()把字符串拆成列表（适合处理文本数据）
text = "语文:98,数学:88,英语:70"
parts = text.split(",")
print(f"方式5 split拆分: {parts}")
````

### 长度和成员检查

```python
len(列表)              # 列表长度
元素 in 列表            # True 或 False
```

> `in` 用来判断某个值在不在列表里，做筛选很好用。

### 📖 例题3：访问列表元素

````python
# 📖 例题3：访问列表元素
# 列表的每个位置都有编号，从0开始

scores = [98, 88, 70, 85, 90]

# 单个元素：用索引 [0] 开始
print(f"第1个成绩: {scores[0]}")   # 98
print(f"最后一个: {scores[-1]}")    # 90（-1表示倒数第1个）
print(f"倒数第2个: {scores[-2]}")   # 85

# 切片：取一段 [起始:结束]（不包含结束位置）
print(f"第2到第4个: {scores[1:4]}")  # [88, 70, 85]
print(f"前3个: {scores[:3]}")        # [98, 88, 70]
print(f"最后2个: {scores[-2:]}")     # [85, 90]

# 步长切片：每隔n个取一个
print(f"每隔1个: {scores[::2]}")     # [98, 70, 90]
````

## 5.3 遍历列表

> 遍历 = 一个一个拿出来，做同样的操作。30个学生，挨个算总分。

### 📖 例题4：长度与成员检查

````python
# 📖 例题4：长度与成员检查
# 判断某个值在不在列表里，做筛选很好用

scores = [98, 88, 70, 85, 90]

# 取长度
print(f"成绩个数: {len(scores)}")   # 5

# 检查成员：用 in 判断
print(f"70在列表里吗: {70 in scores}")  # True
print(f"60在列表里吗: {60 in scores}")  # False

# 实际应用：筛选合格成绩
pass_threshold = 60
for score in scores:
    if score >= pass_threshold:
        print(f"  {score} 分 - 合格")
    else:
        print(f"  {score} 分 - 不合格")
````

### enumerate：同时拿编号和值

> 有时候不仅要知道"是谁"，还要知道"排第几"。`enumerate()` 帮你同时拿到编号和值。

### 📖 例题5：遍历列表——最基础的操作

````python
# 📖 例题5：遍历列表——最基础的操作
# 列表最常用的操作：逐个取出每个元素

students = ["张三", "李四", "王五"]

# 方式1：直接遍历（最常用）
print("方式1：直接遍历")
for name in students:
    print(f"  {name}")

# 方式2：用索引遍历（需要知道位置时用）
print("方式2：用索引遍历")
for i in range(len(students)):
    print(f"  第{i+1}名: {students[i]}")
````

## 5.4 列表的独有操作

> 列表是**可变**的——能加、能改、能删。这是它跟字符串、元组最大的区别。

### 添加元素：append()

```python

```

> ⚠️ **注意**：`append()` 返回 `None`，不能这样写 `x = list1.append(...)`。

### 📖 例题6：多项式求值——列表的进阶用法

````python
# 📖 例题6：多项式求值——列表的进阶用法
# 经济学中常用多项式：P(x) = a0 + a1*x + a2*x² + ... + an*x^n
# 系数存成列表，用enumerate同时拿位置和值

coef = [1, 2, 4, 5, 6]  # a0, a1, a2, a3, a4
x = 23
p = 0

# enumerate同时返回索引和值
for i, a in enumerate(coef):
    p += a * x ** i  # 每次加上 ai * x^i

print(f"P({x}) = {p}")
# 计算过程：1 + 2*23 + 4*23² + 5*23³ + 6*23⁴ = 228184064
````

### 📖 例题7：enumerate——同时拿编号和值

````python
# 📖 例题7：enumerate——同时拿编号和值
# 遍历时需要知道位置，用enumerate

students = ["张三", "李四", "王五"]

# start=0：从0开始编号（默认）
print("start=0:")
for idx, name in enumerate(students):
    print(f"  编号{idx}: {name}")

# start=1：从1开始编号（更自然）
print("start=1:")
for idx, name in enumerate(students, start=1):
    print(f"  第{idx}名: {name}")
````

### 排序：sort() 和 sorted()

| 方法 | 作用 | 原列表变不变 |
|------|------|-------------|
| `列表.sort()` | 原地排序（从小到大） | **变了** |
| `列表.sort(reverse=True)` | 原地降序 | **变了** |
| `sorted(列表)` | 返回新列表（原列表不变） | **没变** |
| `sorted(列表, reverse=True)` | 返回降序新列表 | **没变** |

> 大白话：`sort()` 是**就地改**，`sorted()` 是**另开一个新列表**。

### 📖 例题11：del vs remove——两种删除方法

````python
# 📖 例题11：del vs remove——两种删除方法
# 按位置删 vs 按值删

scores = [98, 88, 70, 85, 90]

# del：按索引删（删掉位置0的元素）
del scores[0]
print(f"del后: {scores}")  # [88, 70, 85, 90]

# remove：按值删（删掉第一个匹配的）
scores2 = [98, 88, 70, 85, 70]
scores2.remove(70)
print(f"remove后: {scores2}")  # [98, 88, 85, 70]（只删第一个70）
````

### 📖 例题12：3种删除方法对比

````python
# 📖 例题12：3种删除方法对比
# del、pop、remove 的区别

# 方式1：del 按位置删（不返回值）
list1 = [98, 88, 70, 85, 90]
del list1[0]
print(f"del后: {list1}")  # [88, 70, 85, 90]

# 方式2：pop 按位置删（返回值）
list2 = [98, 88, 70, 85, 90]
removed = list2.pop(2)  # 删掉索引2的元素，返回它
print(f"pop返回值: {removed}")  # 70
print(f"pop后: {list2}")        # [98, 88, 85, 90]

# 方式3：remove 按值删（不返回值）
list3 = [98, 88, 70, 85, 90]
list3.remove(70)
print(f"remove后: {list3}")  # [98, 88, 85, 90]

# 总结：
# del：按位置，无返回值
# pop：按位置，有返回值（适合"取走"的场景）
# remove：按值，无返回值（适合"删掉某值"的场景）
````

### 📖 例题13：count和index——统计方法

````python
# 📖 例题13：count和index——统计方法
# 统计某个值出现几次，第一次在哪个位置

scores = [98, 88, 70, 85, 70]

# count：统计出现次数
print(f"70出现次数: {scores.count(70)}")  # 2次
print(f"98出现次数: {scores.count(98)}")  # 1次
print(f"60出现次数: {scores.count(60)}")  # 0次

# index：找第一次出现的位置
print(f"70第一次位置: {scores.index(70)}")  # 索引2
print(f"98位置: {scores.index(98)}")         # 索引0

# 注意：index找不到会报错，用try-except处理
try:
    pos = scores.index(60)
except ValueError:
    print("60不在列表里")
````

### 📖 例题15：sort vs sorted——两种排序方法

````python
# 📖 例题15：sort vs sorted——两种排序方法
# 一个原地排序，一个返回新列表

scores = [85, 98, 70]

# sort：原地排序（列表本身变了，返回None）
scores.sort()
print(f"sort后: {scores}")  # [70, 85, 98]
print(f"sort返回值: {scores.sort()}")  # None

# sorted：返回新列表（原列表不变）
scores2 = [85, 98, 70]
sorted_list = sorted(scores2)
print(f"原列表: {scores2}")      # [85, 98, 70]（没变）
print(f"sorted结果: {sorted_list}")  # [70, 85, 98]

# 降序：reverse=True
print(f"降序: {sorted(scores2, reverse=True)}")  # [98, 85, 70]
````

## 5.7 二维列表

> 列表里还能装列表——这就是**二维列表**，相当于一个表格。

酒店有很多楼层，每层有很多房间 → 二维列表就是"表格"。

```python
二维列表 = [[元素11, 元素12, ...],
            [元素21, 元素22, ...],
            ...]
```

> 访问方式：`二维列表[行][列]`——先选行，再选列。

### 📖 例题16：找最高分学生

````python
# 📖 例题16：找最高分学生
# 二维列表 + sum + index 的组合应用

list_id = [1, 2, 3]
list_grade = [[98, 88, 70], [88, 64, 80], [80, 90, 96]]

# 算每个学生的总分
list_sum = [sum(g) for g in list_grade]
print(f"各学生总分: {list_sum}")  # [256, 232, 266]

# 找最高分
max_score = max(list_sum)
max_index = list_sum.index(max_score)
print(f"最高分: {max_score}")      # 266
print(f"学生编号: {list_id[max_index]}")  # 3
````

### 📖 例题17：按总分排序学生

````python
# 📖 例题17：按总分排序学生
# sort和sorted的区别在实际场景中的应用

list_grade = [[98, 88, 70], [88, 64, 80], [80, 90, 96]]
list_sum = [sum(g) for g in list_grade]
print(f"排序前: {list_sum}")  # [256, 232, 266]

# sort：原地排序（列表本身变了）
list_sum.sort(reverse=True)
print(f"sort后: {list_sum}")   # [266, 256, 232]

# sorted：返回新列表（原列表不变）
list_grade2 = [[98, 88, 70], [88, 64, 80], [80, 90, 96]]
list_sum2 = [sum(g) for g in list_grade2]
sorted_sum = sorted(list_sum2, reverse=True)
print(f"原列表: {list_sum2}")       # [256, 232, 266]（没变）
print(f"sorted结果: {sorted_sum}")   # [266, 256, 232]
````

## 5.8 元组

> **元组 = 不能改的列表**——用圆括号 `()` 创建。

```python
list1 = [1, 2, 3]      # 列表：能加能改能删
tuple1 = (1, 2, 3)     # 元组：定了就不能动
```

> **大白话**：列表像**铅笔写的字**，能擦能改；元组像**钢笔写的字**，擦不掉。
>
> 数据不该被改的时候用元组——比如GDP、省份名这些**固定的东西**。

### 📖 例题21：二维列表访问

````python
# 📖 例题21：二维列表访问
# 列表的列表——成绩表的标准结构

list2D = [
    [1, 2, 3],   # 第0行
    [4, 5, 6],   # 第1行
    [7, 8, 9]    # 第2行
]

# 访问单个元素：先选行，再选列
print(f"第1行第2列: {list2D[1][2]}")  # 6

# 访问整行
print(f"第0行: {list2D[0]}")  # [1, 2, 3]

# 访问第3列（需要遍历每行）
print("第2列:")
for row in list2D:
    print(f"  {row[2]}")  # 3, 6, 9
````

### 📖 例题22：创建二维列表的多种方式

````python
# 📖 例题22：创建二维列表的多种方式
# 创建4×5的矩阵

# 方式1：嵌套循环（最直观）
mat1 = []
for i in range(4):
    row = []
    for j in range(5):
        row.append(1)
    mat1.append(row)
print(f"方式1: {mat1}")

# 方式2：列表推导式（最简洁）
mat2 = [[1 for j in range(5)] for i in range(4)]
print(f"方式2: {mat2}")

# 方式3：成绩表方式
grade1 = [98, 88, 70]
grade2 = [85, 90, 88]
grade3 = [76, 82, 95]
list_grade = [grade1, grade2, grade3]
print(f"方式3 成绩表: {list_grade}")
````

### 📖 例题23：二维列表遍历

````python
# 📖 例题23：二维列表遍历
# 遍历成绩表，算总分

list_grade = [
    [98, 88, 70],   # 学生1
    [88, 64, 80],   # 学生2
    [80, 90, 96]    # 学生3
]

# 方式1：嵌套for
for i in range(len(list_grade)):
    for j in range(len(list_grade[i])):
        print(f"  学生{i+1}第{j+1}科: {list_grade[i][j]}")

# 方式2：enumerate + sum（更简洁）
print("用enumerate+sum:")
for i, grade in enumerate(list_grade, start=1):
    print(f"  学生{i}: {grade}, 总分: {sum(grade)}")
````

### 📖 例题24：元组不可变——不能修改

````python
# 📖 例题24：元组不可变——不能修改
# 元组创建后不能改，但可以读

tuple1 = (1, 2, 3, 4)

# 可以做的操作
print(f"元组: {tuple1}")
print(f"长度: {len(tuple1)}")     # 4
print(f"成员检查: {2 in tuple1}")  # True
print(f"访问: {tuple1[0]}")        # 1

# 不能做的操作（会报错）
# tuple1[0] = 10     # TypeError: 不能修改
# tuple1.append(5)   # AttributeError: 没有append方法
# tuple1.sort()      # AttributeError: 没有sort方法
````

### 📖 例题25：元组的拼接、重复、排序

````python
# 📖 例题25：元组的拼接、重复、排序
# 元组不可变，但可以做"不可变操作"

tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)

# 拼接：生成新元组
tuple3 = tuple1 + tuple2
print(f"拼接: {tuple3}")  # (1, 2, 3, 4, 5, 6)

# 重复
tuple4 = tuple1 * 2
print(f"重复: {tuple4}")  # (1, 2, 3, 1, 2, 3)

# 排序：sorted返回新列表，原元组不变
print(f"原元组: {tuple1}")         # (1, 2, 3)
print(f"sorted: {sorted(tuple1)}")  # [1, 2, 3]
````

### 📖 例题26：元组遍历——GDP数据

````python
# 📖 例题26：元组遍历——GDP数据
# 元组适合存"固定数据"，比如GDP统计

gdp = (
    ("中国", 1434290),
    ("美国", 2143320),
    ("日本", 508000),
)

# 遍历元组，解包每对数据
for country, value in gdp:
    print(f"{country}: {value}亿美元")

# 算总值
total = sum(v for _, v in gdp)
print(f"三国GDP总和: {total}亿美元")
````

### 列表 vs 元组

| 特性 | 列表 `[]` | 元组 `()` |
|------|----------|----------|
| 可变？ | 能加能改能删 | 不能改 |
| 方法多 | sort/append/remove... | 只有 count/index |
| 用途 | 要操作的数据 | 固定的数据（如坐标、GDP） |
