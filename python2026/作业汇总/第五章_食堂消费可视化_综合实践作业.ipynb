## ③ 做一张食堂消费图（30分）

下方提供了一段**老师给的绘图模板**，你不需要自己学习新的绘图库。先运行①的数据，再运行下面的代码，就能看到食堂消费柱状图。

**你只要做两件事：** 检查图上的金额与你的列表一致；然后借助AI至少**修改一处外观**（如标题、颜色、背景或柱子样式）。运行遇到错误也可以问AI。

```python
# 老师提供的画图模板：只要运行，就能得到一张柱状图
# 可以让AI修改颜色、标题和样式；绘图语法不是本课考点
import matplotlib.pyplot as plt

plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei", "PingFang SC", "Noto Sans CJK SC", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(9, 4.5), dpi=120)
fig.patch.set_facecolor("#F7F9FC")
ax.set_facecolor("#F7F9FC")
bars = ax.bar(days, money, color="#45B79D", width=0.65)

for bar, value in zip(bars, money):
    ax.text(bar.get_x()+bar.get_width()/2, value+0.3, str(value),
            ha="center", va="bottom", fontsize=10, color="#23485D")

ax.set_title("我的食堂一周午饭消费", fontsize=17, weight="bold", pad=16)
ax.set_ylabel("元", fontsize=11)
ax.set_ylim(0, max(money)+5)
ax.grid(axis="y", alpha=0.18)
ax.set_axisbelow(True)
for edge in ["top", "right", "left"]:
    ax.spines[edge].set_visible(False)
plt.tight_layout()
plt.show()
```
