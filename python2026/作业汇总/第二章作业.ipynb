## 一、选择题

### 1. 下列哪个是浮点数？

A. `42`  
B. `3.14`  
C. `True`  
D. `"42"`

**答案：B**  
`3.14` 有小数点，是浮点数（float）。A是整数，C是布尔值，D是字符串。

---

### 2. `0.1 + 0.2 == 0.3` 的结果是？

A. True  
B. False  
C. 报错  
D. 0.3

**答案：B**  
浮点精度问题：`0.1 + 0.2 = 0.30000000000000004`，不等于0.3。

---

### 3. `int("123")` 的结果是？

A. 123  
B. "123"  
C. 报错  
D. 1.23

**答案：A**  
`int()` 把字符串 `"123"` 转成整数 123。

---

### 4. `bool(0)` 的结果是？

A. True  
B. False  
C. 0  
D. "0"

**答案：B**  
`bool(0)` 返回 `False`，`bool(非零数)` 返回 `True`。

---

### 5. 下列哪个表达式会报错？

A. `int("123")`  
B. `int("12.5")`  
C. `int(3.7)`  
D. `int(True)`

**答案：B**  
`int("12.5")` 报错，字符串里有小数点不能直接转int。需要先 `int(float("12.5"))`。

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
print("3.14".float())
```

````python
# 请在下面填写改正后的代码：
print(float("3.14"))
````

**改正说明：** `float()` 是函数不是方法，用 `float("3.14")` 而不是 `"3.14".float()`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
age = int("3.5")
print(age)
```

````python
# 请在下面填写改正后的代码：
age = int(float("3.5"))
print(age)
````

**改正说明：** `"3.5"` 有小数点，不能直接 `int()`。要先 `float()` 转成浮点数，再 `int()` 转成整数。

---

### 3. 下面的代码运行后报错，请找出问题并改正

```python
x = 0.1 + 0.2
print("结果等于0.3" if x == 0.3 else "结果不等于0.3")
```

````python
# 请在下面填写改正后的代码：
x = 0.1 + 0.2
print(f"结果：{x}")
print("结果不等于0.3，因为浮点精度问题")
````

**改正说明：** 浮点精度问题，`0.1 + 0.2 ≠ 0.3`，应该打印实际值而不是用 `== 0.3` 判断。

---

### 4. 下面的代码运行后报错，请找出问题并改正

```python
score = "85"
print("成绩大于80：" + score > 80)
```

````python
# 请在下面填写改正后的代码：
score = int("85")
print("成绩大于80：" + str(score > 80))
````

**改正说明：** 两行错误：①`score`是字符串不能和整数比较，用`int()`转换；②字符串不能和布尔值拼接，用`str()`转换。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
x = 10
y = "10"
print(x == y)
```

````python
# 请在下面填写改正后的代码：
x = 10
y = int("10")
print(x == y)
````

**改正说明：** `10`（int）和 `"10"`（str）类型不同，永远不相等。用 `int("10")` 统一类型后比较。
