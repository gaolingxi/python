## 一、选择题

### 1. 下列哪个是合法的字典创建方式？

A. `d = (1, 2)`  
B. `d = {1: 2}`  
C. `d = 1: 2`  
D. `d = [1, 2]`

**答案：B**  
字典用花括号 `{}`，键值之间用冒号 `:`，键值对之间用逗号分隔。

---

### 2. 字典的键必须满足什么条件？

A. 必须是字符串  
B. 必须是数字  
C. 必须唯一  
D. 必须有序

**答案：C**  
字典的键必须唯一，同名键后者覆盖前者。键可以是字符串、数字、元组等不可变类型。

---

### 3. `d = {"a": 1, "b": 2}`，`d.get("c")` 的结果是？

A. `None`  
B. `0`  
C. 报错  
D. `"c"`

**答案：A**  
`get()` 在键不存在时返回 `None`，不报错。`d["c"]` 才会报错。

---

### 4. `set([1, 2, 2, 3])` 的结果是？

A. `{1, 2, 2, 3}`  
B. `{1, 2, 3}`  
C. `[1, 2, 3]`  
D. 报错

**答案：B**  
集合自动去重，`2` 只保留一个。

---

### 5. 遍历字典时，同时获取键和值的方法是？

A. `d.keys()`  
B. `d.values()`  
C. `d.items()`  
D. `d.pairs()`

**答案：C**  
`items()` 返回键值对，可以解包为 `for key, value in d.items()`。

## 三、代码题

### 1. 创建字典存3个学生成绩，打印总分。

````python
# 请在下面填写代码：
d = {"张三":98, "李四":88, "王五":70}
print(sum(d.values()))
````

**答案：** 256

---

### 2. 用字典推导式生成 {1:1, 2:4, 3:9}。

````python
# 请在下面填写代码：
d = {i: i**2 for i in range(1, 4)}
print(d)
````

**答案：** {1: 1, 2: 4, 3: 9}

---

### 3. 用 get() 安全访问可能不存在的键，给出默认值0。

````python
# 请在下面填写代码：
d = {"a": 1, "b": 2}
print(d.get("c", 0))
````

**答案：** 0

---

### 4. 创建嵌套字典存2个学生3科成绩，打印"张三"的"数学"。

````python
# 请在下面填写代码：
d = {"张三":{"语文":98,"数学":88}, "李四":{"语文":85,"数学":92}}
print(d["张三"]["数学"])
````

**答案：** 88

---

### 5. 用集合去重，打印去重前后长度。

````python
# 请在下面填写代码：
names = ["Alice","Bob","Alice","Bob"]
print(len(names), len(set(names)))
````

**答案：** 4 2

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1, "a": 2}
print(len(d))
```

````python
# 请在下面填写改正后的代码：
d = {"a": 2, "b": 1}
print(len(d))
````

**改正说明：** 键重复导致 `len(d)` 为1而不是2。键必须唯一。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1}
print(d[0])
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1}
print(d["a"])
````

**改正说明：** 字典不能用数字索引，要用键名。

---

### 3. 下面的代码运行后报错，请找出问题并改正

```python
d = {"a": 1}
print(d.get(0))
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1}
print(d.get("a"))
````

**改正说明：** get() 用键名，不是索引。

---

### 4. 下面的代码运行后报错，请找出问题并改正

```python
set1 = {}
print(set1)
```

````python
# 请在下面填写改正后的代码：
set1 = set()
print(set1)
````

**改正说明：** 空字典 `{}` 不是空集合，空集合要用 `set()`。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
d = {"a": 1}
d["a"] = 2
print(d)
```

````python
# 请在下面填写改正后的代码：
d = {"a": 1, "b": 2}
d["c"] = 3
print(d)
````

**改正说明：** 原代码覆盖同一个键，看不出添加效果。改成添加新键。
