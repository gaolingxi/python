## 一、选择题

### 1. 奖学金标准按“优秀、合格、其他”依次判断。写多分支条件时，哪种安排最合适？

A. 先判断覆盖范围最大的条件  
B. 先判断要求最严格的条件  
C. 任意调换顺序都一样  
D. 只保留最后一个条件  

**答案：B。** 多分支判断从上到下执行；先写严格条件才能保留后续分支。

---

### 2. 逐条检查记录，一旦找到目标就停止整个查找，应使用什么？

A. continue  
B. break  
C. else  
D. elif  

**答案：B。** break 结束当前循环。

---

### 3. 清洗记录时只想跳过当前这条，后面的记录还要继续检查，应使用什么？

A. break  
B. continue  
C. return  
D. pass  

**答案：B。** continue 跳过本次循环的剩余内容，继续下一次。

---

### 4. 设计 while 循环时，怎样降低忘记结束循环的风险？

A. 保证循环体会更新影响条件的变量  
B. 删掉循环条件  
C. 只保留打印语句  
D. 把循环变量写成字符串  

**答案：A。** 条件相关的状态需要逐步变化，才能在合适时结束。

---

### 5. 用 range 指定起点和终点，终点在生成范围中通常如何处理？

A. 始终包含  
B. 不包含  
C. 只生成终点  
D. 忽略起点  

**答案：B。** range 包含起点，不包含终点。

## 二、填空题

### 1. Python 用 ________ 表示条件或循环内部的代码层级。

**答案：** 缩进。同一代码块应保持一致的缩进。

---

### 2. 多个条件依次判断时，“否则如果”对应的关键字是 ________。

**答案：** continue。它不会结束整个循环。

## 三、代码题

### 1. 输入一个成绩，判断是否及格（≥60打印"及格"，否则打印"不及格"）。

````python
# 请在下面填写代码：
score = int(input("请输入成绩："))
print("及格" if score >= 60 else "不及格")
````

**答案：** 输入55 → `不及格`；输入80 → `及格`

---

### 2. 用for循环和range，打印1到10中所有的偶数。

````python
# 请在下面填写代码：
for i in range(2, 11, 2):
    print(i)
````

**答案：** 2 4 6 8 10

---

### 3. 用while循环计算2+4+6+8+10的结果并打印。

````python
# 请在下面填写代码：
total, i = 0, 2
while i <= 10:
    total += i
    i += 2
print(total)
````

**答案：** `30`

---

### 4. 给定收入income=6000，用if-elif-else分级：≥10000"高收入"，≥5000"中等收入"，≥3000"中低收入"，否则"低收入"。

````python
# 请在下面填写代码：
income = 6000
if income >= 10000: print("高收入")
elif income >= 5000: print("中等收入")
else: print("低收入")
````

**答案：** `中等收入`

---

### 5. 用for循环遍历1到10，遇到3的倍数就跳过（用continue），打印其余数字。

````python
# 请在下面填写代码：
for i in range(1, 11):
    if i % 3 == 0: continue
    print(i)
````

**答案：** 1 2 4 5 7 8 10（跳过3、6、9）

## 四、代码纠错题

### 1. 下面的代码运行后报错，请找出问题并改正

```python
score = 75
if score >= 60
    print("及格")
```

````python
# 请在下面填写改正后的代码：
score = 75
if score >= 60:
    print("及格")
````

**改正说明：** if语句行尾必须有冒号 `:`。

---

### 2. 下面的代码运行后报错，请找出问题并改正

```python
for i in range(1, 6)
    print(i)
```

````python
# 请在下面填写改正后的代码：
for i in range(1, 6):
    print(i)
````

**改正说明：** for语句行尾也必须有冒号 `:`。

---

### 3. 下面的代码运行后变成死循环，请找出问题并改正

```python
i = 0
while i < 5:
    print(i)
```

````python
# 请在下面填写改正后的代码：
i = 0
while i < 5:
    print(i)
    i = i + 1
````

**改正说明：** 循环体里缺少 `i = i + 1`，条件永远满足，变成死循环。

---

### 4. 下面的代码输出不正确，请找出问题并改正

```python
for i in range(1, 10):
    if i = 5:
        break
    print(i)
```

````python
# 请在下面填写改正后的代码：
for i in range(1, 10):
    if i == 5:
        break
    print(i)
````

**改正说明：** 判断相等用 `==`，不是 `=`（赋值）。

---

### 5. 下面的代码输出不正确，请找出问题并改正

```python
score = 85
if score >= 60:
    print("及格")
elif score >= 80:
    print("良好")
```

````python
# 请在下面填写改正后的代码：
score = 85
if score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
````

**改正说明：** elif从上到下依次判断，85>=60先满足就走了"及格"，不会再看后面的"良好"。应把更大范围的条件放后面。
